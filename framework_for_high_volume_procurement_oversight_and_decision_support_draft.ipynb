## Introduction

This project provides a Python implementation of the analytical framework presented in Christodoulides (2026). The core of this repository is the iPython Notebook `framework_for_high_volume_procurement_oversight_and_decision_support_draft.ipynb`, which contains a comprehensive suite of orchestrated tasks to replicate the paper's findings.

The pipeline addresses a foundational vulnerability in public procurement and forensic accounting: the reliance on simple univariate metrics (like the Coefficient of Variation) or restrictive digit-based tests (like Benford's Law), which fail to capture the complex, multi-modal structures inherent in modern institutional billing.

The codebase operationalizes the proposed solution—an **interpretable decision-support engine for institutional integrity**:
-   **Anchors** entity resolution in a multi-metric fuzzy logic ensemble (TF-IDF, Jaccard, Token Set Ratio) and Graph Theory (Connected Components) to eliminate identity fragmentation.
-   **Enforces** robust, outlier-resistant standardization using the Global Median and Interquartile Range (IQR).
-   **Propagates** the analysis to the structural level by fitting spherical Gaussian Mixture Models (GMMs), optimized via the Bayesian Information Criterion (BIC), to extract latent payment regimes.
-   **Evaluates** the topological structure through a multiplicative composite index (PHI) that synthesizes Modality, Asymmetry, Tail Behavior, and Structural Dispersion, providing exact log-additive attribution for Explainable AI (XAI).

## Theoretical Background

The implemented methods combine techniques from Computational Linguistics, Robust Statistics, and Information Theory.

**1. Robust Standardization:**
To map heterogeneous suppliers onto a common monetary reference frame without outlier distortion:
$$ \tilde{a}_i = \frac{a_i - \tilde{A}}{IQR} $$

**2. Probabilistic Regime Extraction (GMM):**
Latent payment regimes are extracted using EM optimization, constrained by a sample-size heuristic to prevent overfitting:
$$ k_{max} = \min\left(4, \left\lfloor \frac{n}{25} \right\rfloor\right) $$

**3. The Structural Heterogeneity Index (SHI) Components:**
The framework quantifies four distinct morphological dimensions:
*   **Modality ($M$):** $M = k$ (post-pruning component count).
*   **Asymmetry ($A$):** $A = 1 + |a_q|$, where $a_q$ is Bowley Skewness.
*   **Tail Behavior ($T$):** $T = 1 + |\ln(t_q)|$, where $t_q = \frac{(Q_{0.95} - Q_{0.05}) + \epsilon}{(Q_{0.75} - Q_{0.25}) + \epsilon}$.
*   **Structural Dispersion ($D$):** $D = 1 + \pi_{i^*} s_{i^*} + \sum_{i \neq i^*} \pi_i s_i \ln(1 + d_i)$.

**4. Joint Amplification and XAI Attribution:**
The Payment Heterogeneity Index (PHI) is the multiplicative synthesis of these components, decomposable via logarithms for audit transparency:
$$ PHI = M \times A \times T \times D $$
$$ Contribution_X = \left( \frac{\ln(X)}{\ln(PHI)} \right) \times 100\% $$

## Features

The provided iPython Notebook implements the full research pipeline, including:

-   **Strict Topological Invariants:** Enforces $E \subseteq V \times V$ during Match Graph construction, mathematically guaranteeing that isolated suppliers (singletons) are preserved during entity resolution.
-   **IEEE 754 Floating-Point Safeguards:** Implements strict numerical clipping (`np.clip`) on GMM variances to prevent `NaN` propagation caused by variance collapse in highly clustered regimes.
-   **Stochastic Reproducibility:** Employs cryptographic hashing (SHA-256) to generate deterministic, immutable artifacts at every stage of the pipeline, ensuring perfect auditability.
-   **Exact Inferential Statistics:** Replaces heuristic thresholding with vectorized Monte Carlo permutation testing ($N=5000$) and Two-Sample Kolmogorov-Smirnov divergence testing to validate threshold anchoring.
-   **Configuration-Driven Design:** All study parameters, discretization thresholds, and topological policies are managed in an external `config.yaml` file, ensuring strict methodological reproducibility.

## Methodology Implemented

The core analytical steps directly implement the methodology from the paper:

1.  **Data Preprocessing (Tasks 1-4):** Ingests raw administrative data. Enforces strict temporal alignment, type coercion, and positive-flow isolation to remove accounting reversals.
2.  **Identity Harmonisation (Tasks 5-7):** Normalizes strings and computes TF-IDF, Jaccard, and Token Set Ratios. Constructs a Match Graph and extracts canonical suppliers via Connected Components.
3.  **Robust Scaling & Regime Extraction (Tasks 8-10):** Computes global Median/IQR, standardizes the high-volume cohort ($n \ge 50$), fits spherical GMMs via BIC, and prunes negligible weights ($\pi < 0.05$).
4.  **PHI Synthesis & Triage (Tasks 11-12):** Computes empirical quantiles, synthesizes $M, A, T, D$, assigns risk tiers, and computes exact log-additive XAI attributions.
5.  **Baseline Comparison (Task 13):** Computes the Coefficient of Variation (CV) and executes Spearman rank correlation to demonstrate PHI's superiority over univariate metrics.
6.  **Validation & Audit (Tasks 14-17):** Extracts endogenous peaks via Gaussian KDE, runs permutation tests for threshold anchoring, aggregates sectoral risk, and generates a JSON-serializable Reproduction Manifest.

```python
import yaml
from faker import Faker
from typing import Dict, Any

# Step 1: Synthetic Data Generation (df_raw)

def generate_synthetic_procurement_data(
    num_records: int = 5000,
    num_base_suppliers: int = 20
) -> pd.DataFrame:
    """
    Generates a synthetic, high-fidelity administrative procurement DataFrame.
    Mimics identity fragmentation, heavy-tailed financial noise, and schema constraints.

        raise ValueError(f"num_records must be strictly positive, got {num_records}.")

    suffixes = ["", " LTD", " LIMITED", " PLC", " UK"]

        "Finance", "Children and Education"

        "Card Transaction": [], "Net Amount": [], "Irrecoverable VAT": [],
        "Subjective Group": [], "Subjective Subgroup": [], "Subjective Detail": []

    for i in range(1, num_records + 1):
        data["record_id"].append(i)
        data["Organisation Name"].append("City of York Council")
        
        directorate = np.random.choice(available_directorates)
        data["Directorate"].append(directorate)
        data["Department"].append(f"Dept of {directorate.split()[0]}")
        data["Service Plan"].append("Standard Service Plan")
        
        base_supplier = np.random.choice(base_suppliers)
        suffix = np.random.choice(suffixes)
        data["Creditor Name"].append(f"{base_supplier}{suffix}".strip())
        
        random_offset = pd.Timedelta(days=np.random.randint(0, 184))
        data["Payment Date"].append(start_ts + random_offset)
        
        data["Transaction No"].append(f"TXN-{i:07d}")
        data["Card Transaction"].append(np.random.rand() < 0.10)

            amount = -amount

        data["Net Amount"].append(amount)
        data["Irrecoverable VAT"].append(round(amount * 0.20, 2))
        
        data["Subjective Group"].append("Supplies and Services")

                   "Subjective Subgroup", "Subjective Detail"]
    for col in string_cols:
        df_raw[col] = df_raw[col].astype("string")
        
    df_raw["Payment Date"] = df_raw["Payment Date"].astype("datetime64[ns]")
    df_raw["Card Transaction"] = df_raw["Card Transaction"].astype("bool")
    df_raw["Net Amount"] = df_raw["Net Amount"].astype("float64")
    df_raw["Irrecoverable VAT"] = df_raw["Irrecoverable VAT"].astype("float64")

    return df_raw
```

E-Journal Submission Date: 9 May 2026

Link: https://arxiv.org/abs/2605.12547

Abstract:

Public procurement is vulnerable to error, fraud and corruption, yet high transaction volumes overwhelm oversight. While research often focuses on tender-stage anomalies, post-award payments remain underexplored. Since labelled datasets are rare and existing methods such as Benford's Law face restrictive assumptions, there is a need for additional interpretable, unsupervised frameworks that augment oversight and simplify management. This paper introduces the Structural Heterogeneity Index (SHI), a composite statistic for one-dimensional samples defined by four components: modality, asymmetry, tail behaviour, and structural dispersion. The Payment Heterogeneity Index (PHI) is its multiplicative instance for post-award payments. PHI combines a tail-behaviour component, sensitive to outliers and point clustering, with a structural-dispersion component summarising payment regime architecture. Structural dispersion is computed via Gaussian Mixture Model (GMM) estimation, integrating within-regime variability, prevalence, and separation from the dominant mode. Applied to UK municipal procurement data, PHI isolates a financially significant cohort (10.1% of high-volume suppliers) whose structural signatures deviate from the population and interact with recurring payment anchors. Permutation and Kolmogorov-Smirnov tests confirm that high-PHI suppliers exhibit statistically significant structural differences. A forensic review by a Certified Fraud Examiner supports the plausibility of the prioritised cases. Comparison shows PHI uniquely identifies regime separation obscured by metrics like the Coefficient of Variation (\r{ho}=0.310). PHI functions as an effective discovery tool where no confirmed labels exist, offering a transparent, lightweight screening mechanism for post-award oversight.

### 1. Introduction: The Critical Gap in Post-Award Oversight

Public procurement spending accounts for approximately **12–13% of GDP** and nearly **30% of total government expenditure** across OECD countries. Given this scale, procurement integrity is fundamental to fiscal accountability. Historically, risk detection research has concentrated on tender-stage anomalies, such as collusive bidding or single-bidder tenders, using graph-based and econometric models. However, the paper argues that the implementation phase—where actual public funds are disbursed—is equally vulnerable to waste, error, and corruption.

Current audit practices often rely on manual reviews or digit-based tests like **Benford’s Law**. While Benford's Law is useful, its scope is limited to leading-digit distributions and cannot capture the complex, multi-modal structures inherent in modern billing. Furthermore, the lack of confirmed "fraud labels" in administrative data necessitates an **unsupervised approach**. The PHI framework is designed to bridge this gap by moving beyond individual point-anomalies to capture the **structural complexity** of a supplier's entire payment profile.

### 2. The Data Science and Preprocessing Pipeline

To ensure the statistical reliability of structural estimates, the paper establishes a rigorous preprocessing pipeline:

*   **Identity Harmonisation:** To prevent "identity fragmentation" (where one vendor appears under multiple names), the authors use a **two-step fuzzy matching process**. This includes normalising strings and applying a conservative ensemble score based on **TF-IDF cosine similarity**, token set ratios, and Jaccard overlap. This step is crucial because fragmented identities would artificially simplify observed payment distributions.
*   **Volume Thresholding:** The analysis is restricted to suppliers with **at least 50 payments** ($n \ge 50$). This ensures sufficient data density for GMM estimation and stable distributional signatures. Low-frequency suppliers are better managed via traditional rule-based checks.
*   **Robust Standardisation:** Payment amounts are scaled using the **global median and interquartile range (IQR)**. This "global" approach places all suppliers on a common monetary frame, ensuring that distances between payment regimes reflect absolute financial significance. Unlike Z-score scaling, this method is resistant to extreme outliers.

### 3. Gaussian Mixture Models (GMM) as a Parametric Base

The framework employs **GMMs** to identify latent "payment regimes" within a supplier's activity. A GMM treats a distribution as a weighted combination of Gaussian components, each defined by a mean ($\mu$), standard deviation ($s$), and weight ($\pi$).

To prevent overfitting and maintain interpretability—essential for reducing an auditor's cognitive load—the authors impose a **cap of four components** ($k_{max}=4$). Model selection is driven by the **Bayesian Information Criterion (BIC)**, which penalizes complexity. The authors emphasize that GMMs are used here as a "parametric device" to summarize activity zones rather than a literal model of the data generating process. Components with negligible weight ($\pi < 0.05$) are excluded to ensure focus on prominent regimes.

### 4. Axiomatic Definition of the Structural Heterogeneity Index (SHI)

The **SHI** is a composite statistic defined as a function of four interpretable dimensions:

1.  **Modality ($M$):** The number of distinct GMM modes ($k$). Higher $M$ indicates fragmented structural processes.
2.  **Asymmetry ($A$):** Calculated as $1 + |aq|$, where $aq$ is **Bowley skewness**. This measures directional bias while remaining resistant to outliers.
3.  **Tail Behavior ($T$):** Defined as $1 + |ln(tq)|$, where $tq$ is a ratio of the 95th-5th quantile spread to the 75th-25th spread. This component is unique in its sensitivity to both extreme outliers and **"heaping"** (extreme point clustering).
4.  **Structural Dispersion ($D$):** This is the "key regime-architecture component". It evaluates the combined variability of regimes and their separation from the dominant mode.

### 5. Joint Amplification and the PHI Formula

The **Payment Heterogeneity Index (PHI)** is the multiplicative instance of SHI:
$$\text{PHI} = M \times A \times T \times D$$
The decision to use a **multiplicative model** reflects the principle of **joint amplification**—the idea that the co-occurrence of anomalies across different dimensions should be weighted more heavily than isolated deviations. Furthermore, this structure allows for **exact additive decomposition** in log-space ($ln(\text{PHI})$), enabling auditors to see exactly which factor (e.g., tail behavior vs. modality) is driving a high score for a specific supplier.

### 6. Empirical Results: The City of York Case Study

The authors tested PHI on a six-month snapshot of UK municipal data, identifying 119 suppliers meeting the volume threshold.

*   **Risk Tiering:** Suppliers were segmented into **Low, Moderate, and High PHI tiers** based on percentile ranks. The "High PHI" tier (top 10%) showed substantial financial volume, representing **9.3% of the total disclosed spend** for the entire Council.
*   **Structural Divergence:** The distribution of PHI scores is highly skewed; while the 90th percentile begins at a score of approximately 25, extreme outliers reach scores over 200. This pronounced separation allows oversight teams to ignore routine variation and focus on truly atypical signatures.

### 7. Interpretability through Decomposition: Case Studies

The paper provides five expert-reviewed case studies to validate the triage logic. These illustrate that **PHI does not reward modality alone**:

*   **Supplier B (Moderate PHI):** A bimodal structure where professional fees appear in two distinct regimes—one at £1k and another at £9k. This "structural split" warrants audit review to check if higher values are contractually justified.
*   **Supplier C (Low PHI):** Despite having three modes (Trimodal), it receives a low score because the regimes are closely spaced (Low $D$). This likely reflects a stable labour model with fixed tiers (e.g., staff grades).
*   **Maximum PHI Case:** Driven by extreme **Structural Dispersion** ($D=27.85$), where a dominant regime at £3k coexists with a secondary regime at £154k. This suggests high-value one-off contracts or bundled invoicing that deviates sharply from the norm.

### 8. The Phenomenon of Threshold Anchoring

A significant finding is **"threshold anchoring"**—the propensity for payment regimes to cluster around specific financial benchmarks. The authors used an endogenous peak-extraction procedure to identify seven recurring anchors in the data (e.g., £11,050, £23,450).

Through **permutation tests** ($N=5,000$), the study found that **High-PHI suppliers exhibit significantly stronger interaction** with these anchors compared to low-risk groups ($p = 0.0002$). This alignment is most persistent at mid-to-high value anchors. While proximity to thresholds isn't proof of misconduct, it identifies systemic habits that may indicate where internal controls or approval limits require refinement.

### 9. Comparative Baseline: PHI vs. Coefficient of Variation (CV)

The authors address whether a simpler measure of variability, like CV, could achieve the same results. The **Spearman rank correlation** between CV and PHI is only **$\rho = 0.310$**, indicating weak agreement.

PHI provides superior prioritization because **CV conflates modality and tail effects** and cannot separate "between-regime" variance from "within-regime" noise. For example, a supplier with multiple tightly clustered modes might have a low CV but a high PHI due to compositional complexity. PHI thus "separates operational patterns that a single CV statistic collapses".

### 10. Strategic Management and Sectoral Distribution

The PHI framework allows for **sectoral risk mapping**. In the York dataset, **Adult Social Care**—the largest directorate—showed remarkably standardized billing (98.6% Low PHI). Conversely, the **Transport, Environment and Planning** directorate contained one-third of the high-risk vendors. This concentration in infrastructure-heavy sectors suggests that complex payment architectures are often tied to specific operational service lines, allowing managers to target their limited audit resources where they are most effective.

### 11. Strategic Implications and Decision Support

The paper positions PHI as a "human-in-the-loop" discovery tool rather than a black-box fraud detector. Its **decomposability** translates statistical signals into actionable hypotheses:
*   High **Tail Behavior ($T$)** suggests repeated identical pricing or clustered invoicing.
*   High **Structural Dispersion ($D$)** suggests fragmented service lines or mixed price lists.

By providing an auditable rationale for alerts, PHI addresses the growing need for **algorithmic transparency** in the public sector. The authors propose embedding PHI in a feedback loop where investigator adjudications are recorded to eventually generate the "ground truth" labels currently missing, allowing the framework to evolve from unsupervised discovery to semi-supervised learning.

### 12. Conclusion: A Novel Framework for Integrity

The PHI framework represents a significant advancement in procurement analytics by shifting focus to the **distributional structure of post-award payments**. The central technical contribution is the **structural dispersion component ($D$)**, which provides a principled way to distinguish materially separated billing regimes from benign operational tiering. While the current study is limited to a single authority over six months, the results provide a robust "proof-of-concept" for converting transaction-level complexity into actionable management signals. Future work should focus on longitudinal stability and cross-institutional validation to further refine this "portable, computationally lightweight" screening mechanism.

### **Step 2: Loading the Configuration (`config.yaml`)**

The PHI framework relies on a deterministic, externalized configuration file (`config.yaml`) that defines all hyperparameters, logical gates, and evaluation metrics. We assume this file has been generated and saved in the working directory.

**Methodology:**

```python

    Inputs:

        3.  Validation: Catches file existence errors and parsing errors, raising them.

    Outputs:
        Dict[str, Any]: A nested dictionary containing the complete study configuration.
```

````python
# Task 1 — Validate the appropriateness and quality of ingested inputs (`df_raw`, `STUDY_CONFIG`)

# ==============================================================================
# Task 1: Validate the appropriateness and quality of ingested inputs
# ==============================================================================

@dataclass
class ValidationReport:
    """
    A strongly-typed dataclass representing the outcome of the input validation phase.

    Attributes
    ----------
    is_valid : bool
        True if all schema, parseability, and configuration checks passed.
    errors : List[str]
        A list of detailed error messages if validation failed.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration dictionary, populated with runtime library versions.
    """
    # Boolean flag indicating overall validation success
    is_valid: bool
    # List to store specific validation error messages
    errors: List[str] = field(default_factory=list)
    # Dictionary to store the frozen configuration snapshot
    config_snapshot: Dict[str, Any] = field(default_factory=dict)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 1: Validate df_raw schema, dtypes, and row-level invariants
# -------------------------------------------------------------------------------------------------------------------------------
def validate_df_raw_schema_and_parseability(df_raw: pd.DataFrame, config: Dict[str, Any]) -> None:
    """
    Validates the raw administrative DataFrame against the configured schema and parseability rules.

    This function ensures that all required columns exist, that the `record_id` is a valid
    unique primary key, and that critical fields (`Net Amount`, `Payment Date`) can be
    coerced to their required types. It utilizes an upgraded two-pass regex strategy to
    safely handle standard accounting negative formats (e.g., '(150.00)').

    Parameters
    ----------
    df_raw : pd.DataFrame
        The immutable, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing schema and ingestion conventions.

    Raises
    ------
    ValueError
        If required columns are missing, `record_id` is invalid, or parseability checks fail.
    """
    # Log the initiation of the schema and parseability validation
    logger.info("Starting Task 1, Step 1: Validating df_raw schema and parseability.")

    # Extract the expected schema dictionary from the configuration
    expected_schema: Dict[str, str] = config["raw_unprocessed_data_schema"]
    # Extract the list of expected column names
    expected_columns: List[str] = list(expected_schema.keys())

    # Extract ingestion conventions for column normalization
    strip_whitespace: bool = config["ingestion_conventions"]["column_name_normalization"]["strip_whitespace"]

    # Create a temporary list of column names to avoid mutating the original DataFrame
    actual_columns: List[str] = list(df_raw.columns)

    # Apply whitespace stripping to column names if configured
    if strip_whitespace:
        # Strip leading/trailing whitespace from each column name
        actual_columns = [str(col).strip() for col in actual_columns]

    # Check for missing columns by comparing expected against actual
    missing_cols: List[str] = [col for col in expected_columns if col not in actual_columns]
    if missing_cols:
        # Raise a ValueError if any required columns are missing, halting the pipeline
        raise ValueError(f"Schema validation failed. Missing required columns: {missing_cols}")

    # Validate the uniqueness and completeness of the synthetic primary key 'record_id'
    if not df_raw['record_id'].is_unique:
        # Raise an error if duplicate record_ids exist, as this breaks audit traceability
        raise ValueError("Invariant violation: 'record_id' column contains duplicate values.")
    if df_raw['record_id'].isnull().any():
        # Raise an error if null record_ids exist
        raise ValueError("Invariant violation: 'record_id' column contains null values.")

    # Validate 'Net Amount' parseability using a robust two-pass regex strategy
    logger.debug("Validating 'Net Amount' parseability.")
    try:
        # Extract the series as strings to prevent mutation of the raw DataFrame
        temp_amount_series = df_raw['Net Amount'].astype(str)
        # Pass 1: Strip currency symbols and thousands separators
        temp_amount_series = temp_amount_series.str.replace(r'[£$,]', '', regex=True)
        # Pass 2: Convert accounting parentheses (X) to standard negative signs -X
        temp_amount_series = temp_amount_series.str.replace(r'^\((.*)\)$', r'-\1', regex=True)
        # Attempt to cast the cleaned string series to float64; this will raise ValueError on failure
        _ = temp_amount_series.astype(np.float64)
    except Exception as e:
        # Catch coercion failures and raise a descriptive error
        raise ValueError(f"Parseability violation: 'Net Amount' contains unparseable values. Details: {str(e)}")

    # Validate 'Payment Date' parseability
    logger.debug("Validating 'Payment Date' parseability.")
    try:
        # Attempt to convert the date series to datetime64[ns], strictly raising errors on failure
        _ = pd.to_datetime(df_raw['Payment Date'], errors='raise')
    except Exception as e:
        # Catch datetime coercion failures and raise a descriptive error
        raise ValueError(f"Parseability violation: 'Payment Date' contains unparseable values. Details: {str(e)}")

    # Log successful completion of the validation step
    logger.info("Task 1, Step 1 completed: df_raw schema and parseability validated successfully.")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 2: Validate STUDY_CONFIG completeness against the Methods
# -------------------------------------------------------------------------------------------------------------------------------
def validate_study_config_completeness(config: Dict[str, Any]) -> None:
    """
    Validates that the configuration dictionary contains all mandatory methodological conventions.

    This function traverses the configuration dictionary using dot-separated paths to ensure
    that no critical parameters (e.g., windowing rules, GMM hyperparameters) are missing or None,
    preventing silent numerical drift.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary to validate.

    Raises
    ------
    KeyError
        If a mandatory configuration path is missing.
    ValueError
        If a mandatory configuration path resolves to None.
    """
    logger.info("Starting Step 2: Validating STUDY_CONFIG completeness.")

    # Define a comprehensive list of mandatory dot-separated configuration paths
    mandatory_paths: List[str] = [
        "study_metadata.analysis_window.start_date_inclusive",
        "study_metadata.analysis_window.end_date_inclusive",
        "ingestion_conventions.missingness_policy.Creditor Name",
        "ingestion_conventions.duplicate_policy.deduplicate_on_transaction_no",
        "pre_processing_parameters.identity_harmonisation_logical_gates.normalization.uppercase",
        "pre_processing_parameters.identity_harmonisation_logical_gates.tf_idf_cosine_similarity.threshold",
        "pre_processing_parameters.identity_harmonisation_logical_gates.ensemble_composite_score.threshold_strict_greater_than",
        "cohort_selection_and_scaling.frequency_threshold_n",
        "cohort_selection_and_scaling.robust_scaling_global.scaling_scope",
        "gmm_hyperparameters.model_selection.max_components_cap",
        "gmm_hyperparameters.em_algorithm_settings.covariance_type",
        "gmm_hyperparameters.mixture_pruning.min_weight_pi",
        "phi_computational_constants.synthesis_logic.aggregation",
        "risk_ranking_and_tiering.tier_mapping.high_phi_percentile_ge",
        "threshold_anchoring_parameters.validation_statistics.permutation_iterations"
    ]

    # Iterate through each mandatory path to verify its existence and non-null status
    for path in mandatory_paths:
        # Split the dot-separated path into individual keys
        keys: List[str] = path.split('.')
        # Initialize a pointer to the root of the configuration dictionary
        current_level: Any = config

        # Traverse the dictionary hierarchy
        for key in keys:
            # Check if the current level is a dictionary and contains the key
            if isinstance(current_level, dict) and key in current_level:
                # Move the pointer deeper into the dictionary
                current_level = current_level[key]
            else:
                # Raise a KeyError if the path is broken
                raise KeyError(f"Configuration validation failed: Missing mandatory path '{path}'.")

        # After traversal, ensure the resolved value is not None
        if current_level is None:
            # Raise a ValueError if the configuration value is explicitly None
            raise ValueError(f"Configuration validation failed: Mandatory path '{path}' cannot be None.")

    logger.info("Step 2 completed: STUDY_CONFIG completeness validated successfully.")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 3: Validate reproducibility anchors and deterministic behavior
# -------------------------------------------------------------------------------------------------------------------------------
def capture_reproducibility_anchors(config: Dict[str, Any]) -> None:
    """
    Extracts the global random seed and dynamically captures runtime library versions.

    This function updates the configuration dictionary in-place with the actual versions
    of Python, NumPy, pandas, scikit-learn, and rapidfuzz to ensure the final reproduction
    bundle contains an accurate environment snapshot.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary to be updated with library versions.

    Raises
    ------
    ValueError
        If the global random seed is not an integer.
    ImportError
        If a required library cannot be imported.
    """
    logger.info("Starting Step 3: Capturing reproducibility anchors.")

    # Extract the global random seed from the configuration
    seed: Any = config["study_metadata"]["reproducibility"]["global_random_seed"]

    # Enforce that the seed is a valid integer to ensure deterministic RNG behavior
    if not isinstance(seed, int):
        # Raise an error if the seed is malformed
        raise ValueError(f"Reproducibility violation: global_random_seed must be an integer, got {type(seed)}.")

    # Capture the Python interpreter version (e.g., '3.10.12')
    python_version: str = sys.version.split()[0]
    # Update the configuration with the Python version
    config["study_metadata"]["reproducibility"]["library_versions"]["python"] = python_version

    # Define a mapping of configuration keys to their respective Python package names
    required_libraries: Dict[str, str] = {
        "numpy": "numpy",
        "pandas": "pandas",
        "scikit_learn": "sklearn",
        "rapidfuzz_or_equivalent": "rapidfuzz"
    }

    # Iterate through the required libraries to dynamically import and capture versions
    for config_key, module_name in required_libraries.items():
        try:
            # Dynamically import the module
            module = importlib.import_module(module_name)
            # Extract the __version__ attribute
            version: str = getattr(module, '__version__', 'unknown')
            # Update the configuration dictionary with the captured version
            config["study_metadata"]["reproducibility"]["library_versions"][config_key] = version
            logger.debug(f"Captured version for {module_name}: {version}")
        except ImportError:
            # Raise an ImportError if a critical dependency is missing from the environment
            raise ImportError(f"Reproducibility violation: Required library '{module_name}' is not installed.")

    logger.info("Step 3 completed: Reproducibility anchors captured successfully.")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_1_validate_inputs(df_raw: pd.DataFrame, config: Dict[str, Any]) -> ValidationReport:
    """
    Orchestrates the validation of the raw DataFrame and the study configuration.

    This function acts as the mandatory gatekeeper for the PHI framework. It sequentially
    executes schema validation, configuration completeness checks, and reproducibility anchor
    capture. If all checks pass, it returns a ValidationReport containing a frozen snapshot
    of the configuration.

    Parameters
    ----------
    df_raw : pd.DataFrame
        The immutable, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ValidationReport
        A dataclass containing the validation status, errors (if any), and the config snapshot.
    """
    logger.info("=== Initiating Task 1 Orchestrator: Input Validation ===")

    # Initialize an empty list to collect potential error messages
    errors: List[str] = []

    try:
        # Execute Step 1: Validate the raw DataFrame schema and parseability
        validate_df_raw_schema_and_parseability(df_raw, config)

        # Execute Step 2: Validate the completeness of the configuration dictionary
        validate_study_config_completeness(config)

        # Execute Step 3: Capture runtime library versions and validate the random seed
        # Note: This step mutates the 'config' dictionary in-memory to add versions
        capture_reproducibility_anchors(config)

    except Exception as e:
        # Catch any validation exceptions (ValueError, KeyError, ImportError)
        error_msg: str = f"Validation failed: {str(e)}"
        logger.error(error_msg)
        # Append the error message to the collection
        errors.append(error_msg)
        # Return a failed ValidationReport immediately
        return ValidationReport(is_valid=False, errors=errors, config_snapshot={})

    # Create a deep copy of the configuration to serve as an immutable snapshot for the run
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    logger.info("=== Task 1 Orchestrator completed successfully. Inputs are valid. ===")

    # Return a successful ValidationReport with the frozen configuration snapshot
    return ValidationReport(is_valid=True, errors=[], config_snapshot=config_snapshot)
````

````python
# Task 2 — Cleanse the raw data: type coercion, missingness, and duplicates

# ==============================================================================
# Task 2: Cleanse the raw data: type coercion, missingness, and duplicates
# ==============================================================================

@dataclass
class CleansingDiagnostics:
    """
    A strongly-typed dataclass representing the diagnostic telemetry of the cleansing phase.

    Attributes
    ----------
    initial_row_count : int
        The number of rows in the raw DataFrame before cleansing.
    final_row_count : int
        The number of rows remaining after missingness and duplicate policies are applied.
    dropped_records_by_column : Dict[str, List[int]]
        A mapping of column names to the list of `record_id`s dropped due to missingness in that column.
    duplicate_transaction_count : int
        The number of rows with a duplicated 'Transaction No' (diagnostic only, usually not dropped).
    exact_duplicate_row_count : int
        The number of rows that are exact duplicates across all columns.
    """
    # Store the starting number of rows
    initial_row_count: int
    # Store the ending number of rows
    final_row_count: int
    # Dictionary tracking which record_ids were dropped for which missing column
    dropped_records_by_column: Dict[str, List[int]] = field(default_factory=dict)
    # Count of duplicated transaction numbers
    duplicate_transaction_count: int = 0
    # Count of exact duplicate rows
    exact_duplicate_row_count: int = 0


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 1: Apply column name normalisation and type coercion exactly as configured
# -------------------------------------------------------------------------------------------------------------------------------
def normalize_columns_and_coerce_types(df_raw: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Normalizes column names and strictly coerces critical fields to their required data types.

    This function creates a defensive copy of the raw data, strips whitespace from column names,
    applies a two-pass regex to safely parse accounting formats in 'Net Amount', and enforces
    strict, format-bound datetime coercion to prevent silent temporal drift.

    Parameters
    ----------
    df_raw : pd.DataFrame
        The immutable, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing ingestion conventions.

    Returns
    -------
    pd.DataFrame
        A new DataFrame with normalized columns and coerced data types.

    Raises
    ------
    ValueError
        If type coercion fails, or if the explicit datetime format is missing from the config.
    """
    # Log the initiation of the normalization and coercion step
    logger.info("Starting Task 2, Step 1: Column normalization and type coercion.")

    # Create a deep copy to ensure the original df_raw remains strictly immutable
    df = df_raw.copy(deep=True)

    # Extract the whitespace stripping configuration flag
    strip_whitespace: bool = config["ingestion_conventions"]["column_name_normalization"]["strip_whitespace"]

    # Apply column name normalization if the configuration mandates it
    if strip_whitespace:
        # Strip leading and trailing whitespace from all column names
        df.columns = [str(col).strip() for col in df.columns]
        logger.debug("Stripped whitespace from column names.")

    # Extract the 'Net Amount' parsing configuration
    amount_config: Dict[str, Any] = config["ingestion_conventions"]["type_coercion"]["net_amount_parsing"]

    try:
        # Check if currency symbols and thousands separators are allowed to be stripped
        if amount_config.get("allow_currency_symbols", True) or amount_config.get("allow_thousands_separators", True):
            # Cast to string for regex manipulation
            temp_series = df['Net Amount'].astype(str)
            # Pass 1: Remove £, $, and commas
            temp_series = temp_series.str.replace(r'[£$,]', '', regex=True)
            # Pass 2: Convert accounting parentheses (X) to standard negative signs -X
            temp_series = temp_series.str.replace(r'^\((.*)\)$', r'-\1', regex=True)
            # Cast the cleaned string series to float64
            df['Net Amount'] = temp_series.astype(np.float64)
        else:
            # Strictly cast to float64 without string manipulation if stripping is disabled
            df['Net Amount'] = df['Net Amount'].astype(np.float64)
        logger.debug("Successfully coerced 'Net Amount' to float64.")
    except Exception as e:
        # Raise a descriptive error if amount coercion fails
        raise ValueError(f"Failed to coerce 'Net Amount' to float64. Details: {str(e)}")

    # Extract the explicit datetime format string from the configuration
    date_config: Dict[str, Any] = config["ingestion_conventions"]["type_coercion"].get("payment_date_parsing", {})
    expected_format: str = date_config.get("datetime_format", None)

    # Enforce the presence of an explicit datetime format to prevent silent MM/DD flipping
    if expected_format is None:
        raise ValueError("Methodological violation: 'datetime_format' must be explicitly defined in the configuration to prevent silent temporal drift.")

    try:
        # Coerce 'Payment Date' to datetime64[ns] using the explicit format, strictly raising errors
        df['Payment Date'] = pd.to_datetime(df['Payment Date'], format=expected_format, errors='raise')
        logger.debug(f"Successfully coerced 'Payment Date' using format '{expected_format}'.")
    except Exception as e:
        # Raise a descriptive error if date coercion fails
        raise ValueError(f"Failed to coerce 'Payment Date' to datetime. Details: {str(e)}")

    # Log successful completion of the step
    logger.info("Task 2, Step 1 completed successfully.")

    # Return the mutated copy
    return df


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 2: Apply missingness policy exactly as configured
# -------------------------------------------------------------------------------------------------------------------------------
def apply_missingness_policy(df: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, Dict[str, List[int]]]:
    """
    Removes rows with missing values in critical fields based on the configured policy.

    This function identifies nulls and whitespace-only strings. It implements strict dtype
    checking to ensure that numeric columns containing `np.nan` are not erroneously cast
    to the string "nan", which would bypass the empty string check and corrupt the data.

    Parameters
    ----------
    df : pd.DataFrame
        The type-coerced DataFrame.
    config : Dict[str, Any]
        The study configuration dictionary containing the missingness policy.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, List[int]]]
        - The filtered DataFrame.
        - A dictionary mapping column names to lists of dropped `record_id`s.
    """
    # Log the initiation of the missingness policy application
    logger.info("Starting Task 2, Step 2: Applying missingness policy.")

    # Extract the missingness policy dictionary from the configuration
    missingness_policy: Dict[str, str] = config["ingestion_conventions"]["missingness_policy"]

    # Initialize a dictionary to track which record_ids are dropped for which column
    dropped_records: Dict[str, List[int]] = {}

    # Initialize a boolean mask of True values (keep all rows initially)
    valid_rows_mask: pd.Series = pd.Series(True, index=df.index)

    # Iterate over each column defined in the missingness policy
    for col, policy in missingness_policy.items():
        # Only process columns where the policy explicitly mandates dropping the row
        if policy == "drop_row":
            # Check if the column exists in the DataFrame
            if col in df.columns:
                # Create a primary mask for standard null values (NaN, NaT, None)
                col_missing_mask: pd.Series = df[col].isna()

                # Implement strict dtype checking: only apply string operations to object/string columns
                if pd.api.types.is_object_dtype(df[col]) or pd.api.types.is_string_dtype(df[col]):
                    # Create a secondary mask: True if the stripped string is empty
                    empty_string_mask: pd.Series = df[col].astype(str).str.strip() == ""
                    # Combine the masks using bitwise OR
                    col_missing_mask = col_missing_mask | empty_string_mask

                # Extract the record_ids where the missingness mask is True
                dropped_ids: List[int] = df.loc[col_missing_mask, 'record_id'].tolist()

                # If any records are missing, record them and update the global valid mask
                if dropped_ids:
                    # Store the dropped IDs in the diagnostic dictionary
                    dropped_records[col] = dropped_ids
                    # Update the global mask using bitwise AND with the inverse of the missing mask
                    valid_rows_mask = valid_rows_mask & (~col_missing_mask)
                    logger.debug(f"Identified {len(dropped_ids)} missing records in column '{col}'.")
            else:
                # Log a warning if a configured column is not found in the DataFrame
                logger.warning(f"Column '{col}' specified in missingness policy but not found in DataFrame.")

    # Apply the global valid mask to filter the DataFrame, creating a deep copy
    df_filtered: pd.DataFrame = df.loc[valid_rows_mask].copy(deep=True)

    # Log the total number of rows dropped
    logger.info(f"Task 2, Step 2 completed. Total rows dropped due to missingness: {len(df) - len(df_filtered)}.")

    # Return the filtered DataFrame and the diagnostic dictionary
    return df_filtered, dropped_records


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: Apply duplicate policy (explicitly "no deduplication")
# -------------------------------------------------------------------------------------------------------------------------------
def compute_duplicate_diagnostics_and_enforce_policy(df: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, int, int]:
    """
    Computes duplicate diagnostics and enforces the configured deduplication policy.

    By default (and per the manuscript), this function does not remove duplicates. It computes
    telemetry safely by verifying the existence of the 'Transaction No' column before
    attempting to subset, preventing KeyErrors if upstream policies dropped the column.

    Parameters
    ----------
    df : pd.DataFrame
        The post-missingness DataFrame.
    config : Dict[str, Any]
        The study configuration dictionary containing the duplicate policy.

    Returns
    -------
    Tuple[pd.DataFrame, int, int]
        - The DataFrame (unmodified unless configured otherwise).
        - The count of duplicated 'Transaction No' values.
        - The count of exact duplicate rows.
    """
    # Log the initiation of the duplicate diagnostics computation
    logger.info("Starting Task 2, Step 3: Computing duplicate diagnostics.")

    # Extract the deduplication flag from the configuration
    dedup_flag: bool = config["ingestion_conventions"]["duplicate_policy"]["deduplicate_on_transaction_no"]

    # Initialize diagnostic counters
    duplicate_txn_count: int = 0
    exact_duplicate_count: int = 0

    # Defensive programming: Check if 'Transaction No' exists before computing transaction-level duplicates
    if 'Transaction No' in df.columns:
        # Count rows where 'Transaction No' is duplicated (keeping the first occurrence as non-duplicate)
        duplicate_txn_count = int(df.duplicated(subset=['Transaction No'], keep='first').sum())
        logger.debug(f"Found {duplicate_txn_count} duplicated 'Transaction No' values.")
    else:
        # Log a warning if the column is missing, but proceed safely
        logger.warning("'Transaction No' column missing. Cannot compute transaction-level duplicate telemetry.")

    # Count rows that are exact duplicates across all columns
    exact_duplicate_count = int(df.duplicated(keep='first').sum())
    logger.debug(f"Found {exact_duplicate_count} exact duplicate rows.")

    # Enforce the deduplication policy
    if dedup_flag:
        # If explicitly configured to deduplicate, log a warning as this deviates from the manuscript
        logger.warning("Methodological deviation: deduplicate_on_transaction_no is True. Deduplicating dataset.")
        if 'Transaction No' in df.columns:
            # Drop duplicates based on 'Transaction No'
            df = df.drop_duplicates(subset=['Transaction No'], keep='first').copy(deep=True)
    else:
        # Standard path: do not alter the DataFrame
        logger.info("Policy enforced: No deduplication performed (manuscript default).")

    # Log successful completion of the step
    logger.info("Task 2, Step 3 completed successfully.")

    # Return the DataFrame and the diagnostic counts
    return df, duplicate_txn_count, exact_duplicate_count


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_2_cleanse_data(df_raw: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, CleansingDiagnostics]:
    """
    Orchestrates the data cleansing phase: type coercion, missingness filtering, and duplicate handling.

    This function sequentially executes the three cleansing steps, ensuring that the raw data is
    transformed into a clean, strongly-typed format suitable for downstream windowing and filtering.
    It returns the cleansed DataFrame alongside a comprehensive diagnostic artifact.

    Parameters
    ----------
    df_raw : pd.DataFrame
        The validated, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    Tuple[pd.DataFrame, CleansingDiagnostics]
        - The cleansed DataFrame.
        - A dataclass containing telemetry on dropped rows and duplicates.
    """
    logger.info("=== Initiating Task 2 Orchestrator: Data Cleansing ===")

    # Record the initial number of rows for the diagnostic artifact
    initial_row_count: int = len(df_raw)

    # Execute Step 1: Normalize column names and coerce data types
    # This returns a new DataFrame, preserving df_raw
    df_coerced: pd.DataFrame = normalize_columns_and_coerce_types(df_raw, config)

    # Execute Step 2: Apply the missingness policy
    # This returns the filtered DataFrame and a dictionary of dropped record_ids
    df_no_missing, dropped_records = apply_missingness_policy(df_coerced, config)

    # Execute Step 3: Compute duplicate diagnostics and enforce the duplicate policy
    # This returns the final cleansed DataFrame and the duplicate counts
    df_final, dup_txn_count, exact_dup_count = compute_duplicate_diagnostics_and_enforce_policy(df_no_missing, config)

    # Record the final number of rows after all cleansing steps
    final_row_count: int = len(df_final)

    # Construct the diagnostic dataclass with the collected telemetry
    diagnostics = CleansingDiagnostics(
        initial_row_count=initial_row_count,
        final_row_count=final_row_count,
        dropped_records_by_column=dropped_records,
        duplicate_transaction_count=dup_txn_count,
        exact_duplicate_row_count=exact_dup_count
    )

    logger.info(f"=== Task 2 Orchestrator completed. Rows retained: {final_row_count}/{initial_row_count} ===")

    # Return the cleansed DataFrame and the diagnostic artifact
    return df_final, diagnostics
````

````python
# Task 3 — Enforce the analysis window: May 2025 through October 2025

# ==============================================================================
# Task 3: Enforce the analysis window: May 2025 through October 2025
# ==============================================================================

@dataclass
class WindowingDiagnostics:
    """
    A strongly-typed dataclass representing the telemetry of the temporal windowing phase.

    Attributes
    ----------
    retained_row_count : int
        The number of rows that fall within the configured analysis window.
    excluded_row_count : int
        The number of rows dropped because they fall outside the window.
    min_retained_date : pd.Timestamp
        The earliest 'Payment Date' in the retained dataset.
    max_retained_date : pd.Timestamp
        The latest 'Payment Date' in the retained dataset.
    bounds_confirmed : bool
        Cryptographic-style boolean confirmation that all retained dates are strictly within bounds.
    excluded_record_ids : List[int]
        A list of `record_id`s that were excluded, preserving audit lineage.
    """
    retained_row_count: int
    excluded_row_count: int
    min_retained_date: pd.Timestamp
    max_retained_date: pd.Timestamp
    bounds_confirmed: bool
    excluded_record_ids: List[int] = field(default_factory=list)

@dataclass
class WindowedArtifact:
    """
    An immutable artifact containing the windowed dataset and its cryptographic provenance.

    Attributes
    ----------
    data : pd.DataFrame
        The windowed administrative records.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    row_count : int
        The number of rows in the artifact.
    schema_summary : Dict[str, str]
        A mapping of column names to their stringified pandas data types.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    data: pd.DataFrame
    content_hash_sha256: str
    row_count: int
    schema_summary: Dict[str, str]
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 1: Apply the configured inclusive date filter exactly
# -------------------------------------------------------------------------------------------------------------------------------
def apply_date_window_filter(df: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, pd.DataFrame, pd.Timestamp, pd.Timestamp]:
    """
    Filters the DataFrame to strictly include records within the configured date boundaries.

    This function mathematically guarantees calendar-day inclusivity by explicitly applying
    `.normalize()` to both the parsed boundary variables and the 'Payment Date' series,
    stripping any intra-day time components that could cause the inequality logic to fail.

    Parameters
    ----------
    df : pd.DataFrame
        The cleansed DataFrame with coerced datetime64[ns] 'Payment Date'.
    config : Dict[str, Any]
        The study configuration dictionary containing the analysis window parameters.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, pd.Timestamp, pd.Timestamp]
        - df_retained: DataFrame containing records within the window.
        - df_excluded: DataFrame containing records outside the window.
        - start_date: The normalized inclusive start boundary.
        - end_date: The normalized inclusive end boundary.

    Raises
    ------
    ValueError
        If the 'Payment Date' column is timezone-aware, violating the naive_local_date config.
    """
    # Log the initiation of the date window filter
    logger.info("Starting Task 3, Step 1: Applying inclusive date window filter.")

    # Extract window boundaries from the configuration
    window_config: Dict[str, str] = config["study_metadata"]["analysis_window"]
    start_str: str = window_config["start_date_inclusive"]
    end_str: str = window_config["end_date_inclusive"]

    try:
        # Parse boundaries to pandas Timestamps and explicitly normalize to midnight
        # This guarantees that intra-day times in the config do not corrupt the inclusivity logic
        start_date: pd.Timestamp = pd.to_datetime(start_str, errors='raise').normalize()
        end_date: pd.Timestamp = pd.to_datetime(end_str, errors='raise').normalize()
    except Exception as e:
        # Raise a descriptive error if boundary parsing fails
        raise ValueError(f"Failed to parse configured window boundaries. Details: {str(e)}")

    # Assert timezone naivety to prevent silent boundary shifts across geographic regions
    if df['Payment Date'].dt.tz is not None:
        raise ValueError("Methodological violation: 'Payment Date' is timezone-aware, but config specifies naive_local_date.")

    # Normalize the Payment Date series to midnight (00:00:00) to ensure calendar-day inclusivity
    # Equation/Logic: start_date <= normalize(Payment Date) <= end_date
    normalized_dates: pd.Series = df['Payment Date'].dt.normalize()

    # Create the boolean mask for the inclusive window
    window_mask: pd.Series = (normalized_dates >= start_date) & (normalized_dates <= end_date)

    # Split the DataFrame into retained and excluded sets using defensive deep copies
    df_retained: pd.DataFrame = df.loc[window_mask].copy(deep=True)
    df_excluded: pd.DataFrame = df.loc[~window_mask].copy(deep=True)

    # Log the partitioning telemetry
    logger.debug(f"Date filter applied. Retained: {len(df_retained)}, Excluded: {len(df_excluded)}.")
    logger.info("Task 3, Step 1 completed successfully.")

    # Return the partitioned DataFrames and the normalized boundaries
    return df_retained, df_excluded, start_date, end_date


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 2: Validate windowing results
# -------------------------------------------------------------------------------------------------------------------------------
def validate_windowing_results(
    df_retained: pd.DataFrame,
    df_excluded: pd.DataFrame,
    start_date: pd.Timestamp,
    end_date: pd.Timestamp
) -> WindowingDiagnostics:
    """
    Validates the temporal integrity of the windowed cohort and generates telemetry.

    This function mathematically guarantees that all retained records fall within the
    specified boundaries and raises a hard failure if the resulting cohort is empty.

    Parameters
    ----------
    df_retained : pd.DataFrame
        The DataFrame containing records within the window.
    df_excluded : pd.DataFrame
        The DataFrame containing records outside the window.
    start_date : pd.Timestamp
        The parsed inclusive start boundary.
    end_date : pd.Timestamp
        The parsed inclusive end boundary.

    Returns
    -------
    WindowingDiagnostics
        A dataclass containing counts, date ranges, and audit lineage for excluded records.

    Raises
    ------
    ValueError
        If the retained DataFrame is completely empty.
    """
    logger.info("Starting Task 3, Step 2: Validating windowing results.")

    # Hard failure condition: The PHI framework cannot operate on an empty dataset
    if df_retained.empty:
        raise ValueError("Fatal Error: The applied date window resulted in an empty DataFrame. Downstream GMM estimation is impossible.")

    # Compute aggregate statistics on the retained dates
    min_date: pd.Timestamp = df_retained['Payment Date'].min()
    max_date: pd.Timestamp = df_retained['Payment Date'].max()

    # Mathematically confirm that the bounds are strictly respected
    # We normalize the min/max dates to ensure intra-day times don't cause false negatives
    bounds_confirmed: bool = bool((min_date.normalize() >= start_date) and (max_date.normalize() <= end_date))

    if not bounds_confirmed:
        # This should be mathematically impossible given Step 1, but serves as a defensive assertion
        logger.error(f"Bounds check failed! Min: {min_date}, Max: {max_date}, Expected: [{start_date}, {end_date}]")

    # Extract the record_ids of excluded rows for audit traceability
    excluded_ids: List[int] = df_excluded['record_id'].tolist()

    # Construct the diagnostic artifact
    diagnostics = WindowingDiagnostics(
        retained_row_count=len(df_retained),
        excluded_row_count=len(df_excluded),
        min_retained_date=min_date,
        max_retained_date=max_date,
        bounds_confirmed=bounds_confirmed,
        excluded_record_ids=excluded_ids
    )

    logger.info(f"Task 3, Step 2 completed. Bounds confirmed: {bounds_confirmed}.")
    return diagnostics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3: Persist the windowed raw dataset as an immutable intermediate artifact
# -------------------------------------------------------------------------------------------------------------------------------
def create_immutable_windowed_artifact(df_retained: pd.DataFrame, config: Dict[str, Any]) -> Any: # Returns WindowedArtifact
    """
    Creates a cryptographically verifiable artifact of the windowed dataset.

    To ensure the hash is deterministic across different environments, the DataFrame is
    sorted by `record_id` and column names, and serialized to CSV with strict float (%.8f)
    and ISO 8601 datetime formatting before SHA-256 hashing. This prevents micro-drift
    caused by floating-point truncation or locale-specific date representations.

    Parameters
    ----------
    df_retained : pd.DataFrame
        The validated, windowed DataFrame.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    WindowedArtifact
        A dataclass containing the DataFrame, its SHA-256 hash, and metadata.
    """
    # Log the initiation of the artifact creation process
    logger.info("Starting Task 3, Step 3: Creating immutable windowed artifact.")

    # 1. Ensure deterministic row order by sorting on the primary key
    # We reset the index and drop the old one to prevent it from entering the CSV byte stream
    df_sorted: pd.DataFrame = df_retained.sort_values(by='record_id').reset_index(drop=True)

    # 2. Ensure deterministic column order by sorting alphabetically
    # This guarantees that column permutations do not alter the cryptographic hash
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted = df_sorted[sorted_columns]

    # 3. Serialize to a deterministic string representation
    # We use to_csv with strict %.8f formatting to guarantee zero-loss representation
    # of standard IEEE 754 double-precision floats, preventing hash drift.
    # We enforce ISO 8601 date formatting to prevent locale-specific drift.
    csv_bytes: bytes = df_sorted.to_csv(
        index=False,
        float_format='%.8f',
        date_format='%Y-%m-%dT%H:%M:%S'
    ).encode('utf-8')

    # 4. Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # 5. Extract the schema summary (mapping column names to stringified dtypes)
    schema_summary: Dict[str, str] = {col: str(dtype) for col, dtype in df_sorted.dtypes.items()}

    # 6. Create a deep copy of the configuration to freeze the state
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    # Construct the artifact (Assuming WindowedArtifact is imported/available)
    # We return the original df_retained to preserve ingestion order for operations,
    # using df_sorted strictly for the cryptographic hash.
    # Note: In a full execution environment, this would instantiate WindowedArtifact.
    artifact = WindowedArtifact(
        data=df_retained,
        content_hash_sha256=content_hash,
        row_count=len(df_retained),
        schema_summary=schema_summary,
        config_snapshot=config_snapshot
    )

    # Log successful completion
    logger.info("Task 3, Step 3 completed. Immutable artifact generated.")

    # Return the instantiated dataclass
    return artifact


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_3_enforce_analysis_window(df_cleansed: pd.DataFrame, config: Dict[str, Any]) -> Tuple[WindowedArtifact, WindowingDiagnostics]:
    """
    Orchestrates the enforcement of the temporal analysis window.

    This function sequentially applies the inclusive date filter, validates the temporal
    integrity of the resulting cohort, and packages the data into a cryptographically
    verifiable artifact for downstream processing.

    Parameters
    ----------
    df_cleansed : pd.DataFrame
        The cleansed DataFrame from Task 2.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    Tuple[WindowedArtifact, WindowingDiagnostics]
        - The immutable artifact containing the windowed data and its hash.
        - The telemetry diagnostics for the windowing operation.
    """
    logger.info("=== Initiating Task 3 Orchestrator: Enforce Analysis Window ===")

    # Execute Step 1: Apply the date filter
    df_retained, df_excluded, start_date, end_date = apply_date_window_filter(df_cleansed, config)

    # Execute Step 2: Validate the results and generate telemetry
    diagnostics: WindowingDiagnostics = validate_windowing_results(
        df_retained=df_retained,
        df_excluded=df_excluded,
        start_date=start_date,
        end_date=end_date
    )

    # Execute Step 3: Create the immutable artifact
    artifact: WindowedArtifact = create_immutable_windowed_artifact(
        df_retained=df_retained,
        config=config
    )

    logger.info(f"=== Task 3 Orchestrator completed. Artifact Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact, diagnostics
````

````python
# Task 4 — Filter non-standard amounts: positive-flow isolation

# ==============================================================================
# Task 4: Filter non-standard amounts: positive-flow isolation
# ==============================================================================

@dataclass
class PositiveFlowDiagnostics:
    """
    A strongly-typed dataclass representing the telemetry of the positive-flow isolation phase.

    Attributes
    ----------
    pre_filter_row_count : int
        The total number of rows before applying the exclusion rule.
    retained_row_count : int
        The number of strictly positive payment rows retained.
    excluded_row_count : int
        The number of non-positive payment rows excluded.
    percent_removed : float
        The percentage of rows removed (excluded / pre_filter * 100).
    expected_percent_removed : float
        The expected removal percentage defined in the study configuration (e.g., 1.75%).
    absolute_divergence : float
        The absolute difference between the computed and expected removal percentages.
    checkpoint_met : bool
        True if the computed percentage is mathematically close to the expected percentage.
    excluded_record_ids : List[int]
        A list of `record_id`s that were excluded, preserving audit lineage.
    """
    pre_filter_row_count: int
    retained_row_count: int
    excluded_row_count: int
    percent_removed: float
    expected_percent_removed: float
    absolute_divergence: float
    checkpoint_met: bool
    excluded_record_ids: List[int] = field(default_factory=list)

@dataclass
class PositiveFlowArtifact:
    """
    An immutable artifact containing the positive-flow dataset and its cryptographic provenance.

    Attributes
    ----------
    data : pd.DataFrame
        The strictly positive administrative records.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    row_count : int
        The number of rows in the artifact.
    schema_summary : Dict[str, str]
        A mapping of column names to their stringified pandas data types.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    data: pd.DataFrame
    content_hash_sha256: str
    row_count: int
    schema_summary: Dict[str, str]
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 1: Apply the manuscript's exclusion rule exactly
# -------------------------------------------------------------------------------------------------------------------------------
def isolate_positive_payment_flows(df: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Isolates strictly positive payment flows by removing non-positive amounts.

    This function strictly implements the manuscript's exclusion rule (Net Amount <= 0).
    It partitions the input DataFrame into retained and excluded sets, ensuring no data
    is silently dropped without an audit trail.

    Parameters
    ----------
    df : pd.DataFrame
        The windowed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing the exclusion logic.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame]
        - df_retained: DataFrame containing strictly positive payments.
        - df_excluded: DataFrame containing non-positive payments (refunds, credits, zeros).

    Raises
    ------
    ValueError
        If the configuration disables the exclusion rule, or if 'Net Amount' contains nulls.
    """
    logger.info("Starting Task 4, Step 1: Isolating positive payment flows.")

    # Extract the exclusion logic configuration
    exclusion_config: Dict[str, Any] = config["pre_processing_parameters"]["exclusion_logic"]

    # Enforce methodological strictness: the exclusion rule must be enabled
    if not exclusion_config.get("exclude_non_positive", False):
        raise ValueError("Methodological violation: 'exclude_non_positive' must be True to replicate the PHI framework.")

    # Defensive assertion: Ensure 'Net Amount' contains no nulls before masking
    if df['Net Amount'].isna().any():
        raise ValueError("Data integrity violation: 'Net Amount' contains null values prior to positive-flow filtering.")

    # Construct the boolean mask for non-positive values (Equation: a_i <= 0)
    # We use strict float comparison against 0.0
    exclusion_mask: pd.Series = df['Net Amount'] <= 0.0

    # Partition the DataFrame using defensive deep copies to ensure immutability
    # Retained set: Inverse of the exclusion mask (Net Amount > 0)
    df_retained: pd.DataFrame = df.loc[~exclusion_mask].copy(deep=True)
    # Excluded set: The exclusion mask (Net Amount <= 0)
    df_excluded: pd.DataFrame = df.loc[exclusion_mask].copy(deep=True)

    logger.debug(f"Positive-flow isolation complete. Retained: {len(df_retained)}, Excluded: {len(df_excluded)}.")

    return df_retained, df_excluded


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2: Validate against the manuscript's diagnostic checkpoint
# -------------------------------------------------------------------------------------------------------------------------------
def validate_positive_flow_diagnostics(
    df_retained: pd.DataFrame,
    df_excluded: pd.DataFrame,
    config: Dict[str, Any]
) -> PositiveFlowDiagnostics:
    """
    Validates the exclusion results against the manuscript's reported diagnostic checkpoint.

    This function computes the percentage of records removed and compares it to the expected
    1.75% reported in the study, generating telemetry to detect potential dataset snapshot drift.

    Parameters
    ----------
    df_retained : pd.DataFrame
        The DataFrame containing strictly positive payments.
    df_excluded : pd.DataFrame
        The DataFrame containing non-positive payments.
    config : Dict[str, Any]
        The study configuration dictionary containing the expected diagnostic values.

    Returns
    -------
    PositiveFlowDiagnostics
        A dataclass containing counts, percentages, and checkpoint validation status.

    Raises
    ------
    ValueError
        If the total pre-filter row count is zero.
    """
    logger.info("Starting Task 4, Step 2: Validating positive-flow diagnostics.")

    # Calculate row counts
    retained_count: int = len(df_retained)
    excluded_count: int = len(df_excluded)
    total_count: int = retained_count + excluded_count

    # Hard failure condition: Cannot compute percentages on an empty dataset
    if total_count == 0:
        raise ValueError("Fatal Error: The pre-filter dataset is empty. Cannot compute exclusion diagnostics.")

    # Compute the percentage of records removed (Equation: %_removed = 100 * (removed / total))
    percent_removed: float = (excluded_count / total_count) * 100.0

    # Extract the expected percentage from the configuration
    expected_percent: float = config["pre_processing_parameters"]["exclusion_logic"]["expected_percent_removed_in_study"]

    # Calculate the absolute divergence
    absolute_divergence: float = abs(percent_removed - expected_percent)

    # Evaluate the checkpoint status using a reasonable floating-point tolerance (e.g., 0.01%)
    checkpoint_met: bool = math.isclose(percent_removed, expected_percent, abs_tol=1e-2)

    if not checkpoint_met:
        # Log a warning rather than raising an error, as this indicates snapshot drift, not a code failure
        logger.warning(
            f"Diagnostic Checkpoint Missed: Removed {percent_removed:.2f}% of records, "
            f"expected {expected_percent:.2f}%. This indicates potential source-snapshot drift."
        )
    else:
        logger.info(f"Diagnostic Checkpoint Met: Removed {percent_removed:.2f}% of records.")

    # Extract the record_ids of excluded rows for audit traceability
    excluded_ids: List[int] = df_excluded['record_id'].tolist()

    # Construct the diagnostic artifact
    diagnostics = PositiveFlowDiagnostics(
        pre_filter_row_count=total_count,
        retained_row_count=retained_count,
        excluded_row_count=excluded_count,
        percent_removed=percent_removed,
        expected_percent_removed=expected_percent,
        absolute_divergence=absolute_divergence,
        checkpoint_met=checkpoint_met,
        excluded_record_ids=excluded_ids
    )

    return diagnostics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: Persist the positive-flow dataset as an immutable intermediate artifact
# -------------------------------------------------------------------------------------------------------------------------------
def create_immutable_positive_flow_artifact(df_retained: pd.DataFrame, config: Dict[str, Any]) -> PositiveFlowArtifact:
    """
    Creates a cryptographically verifiable artifact of the positive-flow dataset.

    To ensure the hash is deterministic across different environments, the DataFrame is
    sorted by `record_id` and column names, and serialized to CSV with strict float and
    datetime formatting before SHA-256 hashing.

    Parameters
    ----------
    df_retained : pd.DataFrame
        The validated, strictly positive DataFrame.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    PositiveFlowArtifact
        A dataclass containing the DataFrame, its SHA-256 hash, and metadata.

    Raises
    ------
    ValueError
        If the retained DataFrame is empty.
    """
    logger.info("Starting Task 4, Step 3: Creating immutable positive-flow artifact.")

    # Hard failure condition: Downstream GMM estimation requires data
    if df_retained.empty:
        raise ValueError("Fatal Error: The positive-flow dataset is empty after filtering. Pipeline halted.")

    # 1. Ensure deterministic row order by sorting on the primary key
    df_sorted: pd.DataFrame = df_retained.sort_values(by='record_id').reset_index(drop=True)

    # 2. Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted = df_sorted[sorted_columns]

    # 3. Serialize to a deterministic string representation
    # We use to_csv with strict formatting to avoid cross-platform float representation issues
    csv_bytes: bytes = df_sorted.to_csv(
        index=False,
        float_format='%.6f',
        date_format='%Y-%m-%dT%H:%M:%S'
    ).encode('utf-8')

    # 4. Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # 5. Extract the schema summary (mapping column names to stringified dtypes)
    schema_summary: Dict[str, str] = {col: str(dtype) for col, dtype in df_sorted.dtypes.items()}

    # 6. Create a deep copy of the configuration to freeze the state
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    # Construct the artifact
    artifact = PositiveFlowArtifact(
        data=df_retained, # Store the original (unsorted) DataFrame to preserve ingestion order
        content_hash_sha256=content_hash,
        row_count=len(df_retained),
        schema_summary=schema_summary,
        config_snapshot=config_snapshot
    )

    logger.info("Task 4, Step 3 completed. Immutable artifact generated.")
    return artifact


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_4_filter_non_standard_amounts(
    windowed_artifact: WindowedArtifact,
    config: Dict[str, Any]
) -> Tuple[PositiveFlowArtifact, PositiveFlowDiagnostics]:
    """
    Orchestrates the isolation of positive payment flows.

    This function sequentially applies the manuscript's exclusion rule, validates the
    results against the expected diagnostic checkpoint, and packages the strictly positive
    data into a cryptographically verifiable artifact for downstream entity resolution.

    Parameters
    ----------
    windowed_artifact : WindowedArtifact
        The immutable artifact containing the windowed data from Task 3.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    Tuple[PositiveFlowArtifact, PositiveFlowDiagnostics]
        - The immutable artifact containing the positive-flow data and its hash.
        - The telemetry diagnostics for the exclusion operation.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 4 Orchestrator: Filter Non-Standard Amounts ===")

    # Extract the DataFrame from the incoming artifact
    df_windowed: pd.DataFrame = windowed_artifact.data

    if df_windowed.empty:
        raise ValueError("Fatal Error: Input WindowedArtifact contains an empty DataFrame.")

    # Execute Step 1: Apply the exclusion rule
    df_retained, df_excluded = isolate_positive_payment_flows(df_windowed, config)

    # Execute Step 2: Validate the results and generate telemetry
    diagnostics: PositiveFlowDiagnostics = validate_positive_flow_diagnostics(
        df_retained=df_retained,
        df_excluded=df_excluded,
        config=config
    )

    # Execute Step 3: Create the immutable artifact
    artifact: PositiveFlowArtifact = create_immutable_positive_flow_artifact(
        df_retained=df_retained,
        config=config
    )

    logger.info(f"=== Task 4 Orchestrator completed. Artifact Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact, diagnostics
````

````python
# Task 5 — Prepare supplier names for identity harmonisation: normalization and tokenisation

# ==========================================================================================
# Task 5: Prepare supplier names for identity harmonisation: normalization and tokenisation
# ==========================================================================================

@dataclass
class NormalizedNameArtifact:
    """
    An immutable artifact containing the normalized supplier name corpus and its cryptographic provenance.

    Attributes
    ----------
    data : pd.DataFrame
        The aggregated corpus. Columns: 'canonical_id', 'normalized_name', 'raw_names' (list),
        'tokens' (list), 'occurrence_count', 'record_ids' (list).
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    unique_normalized_names_count : int
        The number of unique normalized names (nodes for the Match Graph).
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    data: pd.DataFrame
    content_hash_sha256: str
    unique_normalized_names_count: int
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 1: Apply string normalization exactly as configured
# -------------------------------------------------------------------------------------------------------------------------------
def normalize_supplier_names(df_positive: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Applies strict, configured string normalization to raw Creditor Names.

    This function aggregates the transaction-level data to the unique raw name level,
    then applies uppercase conversion, ampersand replacement, punctuation removal,
    suffix standardization, and whitespace collapsing.

    Parameters
    ----------
    df_positive : pd.DataFrame
        The strictly positive administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing normalization rules.

    Returns
    -------
    pd.DataFrame
        A DataFrame keyed by raw 'Creditor Name' with a new 'normalized_name' column,
        occurrence counts, and lists of associated 'record_id's.
    """
    logger.info("Starting Task 5, Step 1: Normalizing supplier names.")

    # Extract normalization configuration
    norm_config: Dict[str, Any] = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]["normalization"]

    # 1. Aggregate to unique raw Creditor Names to minimize expensive string operations
    # We collect the record_ids into a list to preserve perfect row-level lineage
    df_agg = df_positive.groupby('Creditor Name').agg(
        occurrence_count=('record_id', 'size'),
        record_ids=('record_id', list)
    ).reset_index()

    # Initialize the normalized_name series with the raw names cast to string
    norm_series: pd.Series = df_agg['Creditor Name'].astype(str)

    # 2. Uppercase conversion
    if norm_config.get("uppercase", True):
        norm_series = norm_series.str.upper()

    # 3. Ampersand replacement
    if norm_config.get("ampersand_to_AND", True):
        norm_series = norm_series.str.replace(r'&', ' AND ', regex=True)

    # 4. Punctuation removal
    if norm_config.get("remove_punctuation", True):
        punct_regex: str = norm_config.get("punctuation_regex", r"[^\w\s]")
        norm_series = norm_series.str.replace(punct_regex, '', regex=True)

    # 5. Suffix standardization
    suffix_config: Dict[str, Any] = norm_config.get("suffix_standardization", {})
    if suffix_config.get("enabled", True):
        suffix_map: Dict[str, str] = suffix_config.get("map", {})
        drop_suffixes: bool = suffix_config.get("drop_suffix_tokens_after_standardize", False)

        for raw_suffix, std_suffix in suffix_map.items():
            # Use word boundaries (\b) to prevent replacing substrings inside valid words
            # e.g., \bLTD\b -> LIMITED
            pattern = rf'\b{re.escape(raw_suffix)}\b'
            replacement = '' if drop_suffixes else std_suffix
            norm_series = norm_series.str.replace(pattern, replacement, regex=True)

    # 6. Collapse whitespace and trim boundaries
    if norm_config.get("collapse_whitespace", True):
        norm_series = norm_series.str.replace(r'\s+', ' ', regex=True).str.strip()

    # 7. Handle edge cases where normalization results in an empty string
    # Fallback to the raw name (uppercased and stripped) to prevent data loss
    empty_mask = norm_series == ""
    if empty_mask.any():
        logger.warning(f"Normalization resulted in {empty_mask.sum()} empty strings. Falling back to raw names.")
        norm_series.loc[empty_mask] = df_agg.loc[empty_mask, 'Creditor Name'].astype(str).str.upper().str.strip()

    # Assign the normalized series back to the DataFrame
    df_agg['normalized_name'] = norm_series

    logger.debug(f"Normalized {len(df_agg)} unique raw creditor names.")
    return df_agg


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 2: Freeze tokenisation rules for token-based metrics
# -------------------------------------------------------------------------------------------------------------------------------
def tokenize_normalized_names(df_norm: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Applies frozen tokenization rules to the normalized names.

    This function splits the normalized strings, removes configured drop tokens,
    and enforces minimum token lengths, ensuring identical tokenization for all
    downstream edit-distance and Jaccard metrics.

    Parameters
    ----------
    df_norm : pd.DataFrame
        The DataFrame containing raw and normalized names.
    config : Dict[str, Any]
        The study configuration dictionary containing tokenization rules.

    Returns
    -------
    pd.DataFrame
        The input DataFrame with an appended 'tokens' column containing lists of strings.
    """
    logger.info("Starting Task 5, Step 2: Tokenizing normalized names.")

    # Extract tokenization configuration from the token_set_ratio block
    # (The config specifies this is shared with Jaccard overlap)
    tok_config: Dict[str, Any] = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]["token_set_ratio"]["tokenization"]

    split_rule: str = tok_config.get("split_rule", "whitespace")
    drop_tokens: set = set(tok_config.get("drop_tokens", []))
    min_length: int = tok_config.get("min_token_length", 1)

    def _tokenize_string(text: str) -> List[str]:
        """Pure inner function to tokenize a single string deterministically."""
        if not isinstance(text, str):
            return []

        # Split by whitespace
        if split_rule == "whitespace":
            raw_tokens = text.split()
        else:
            # Fallback to whitespace if unknown rule is provided
            raw_tokens = text.split()

        # Filter tokens based on drop list and minimum length
        filtered_tokens = [
            t for t in raw_tokens
            if t not in drop_tokens and len(t) >= min_length
        ]
        return filtered_tokens

    # Apply the pure tokenization function to the normalized_name column
    # .map() with a standard Python function is highly optimized for list generation
    df_norm['tokens'] = df_norm['normalized_name'].map(_tokenize_string)

    # Diagnostic check for empty token lists
    empty_tokens_count = (df_norm['tokens'].str.len() == 0).sum()
    if empty_tokens_count > 0:
        logger.warning(f"Tokenization resulted in {empty_tokens_count} empty token lists. Downstream metrics must handle this.")

    logger.debug("Tokenization complete.")
    return df_norm


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 3: Record the normalized-name corpus as a controlled artifact
# -------------------------------------------------------------------------------------------------------------------------------
def create_normalized_name_artifact(df_norm: pd.DataFrame, config: Dict[str, Any]) -> NormalizedNameArtifact:
    """
    Aggregates the corpus to the unique normalized name level and creates an immutable artifact.

    This function groups by `normalized_name`, assigns a stable `canonical_id`, and
    computes a deterministic SHA-256 hash of the corpus to establish a cryptographic
    baseline for the Match Graph nodes.

    Parameters
    ----------
    df_norm : pd.DataFrame
        The DataFrame containing raw names, normalized names, and tokens.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    NormalizedNameArtifact
        A dataclass containing the aggregated corpus, its hash, and metadata.
    """
    logger.info("Starting Task 5, Step 3: Creating normalized-name artifact.")

    # 1. Aggregate to the unique normalized_name level
    # Multiple raw names may map to the same normalized name. We must combine their records.
    df_agg = df_norm.groupby('normalized_name').agg(
        raw_names=('Creditor Name', list),
        occurrence_count=('occurrence_count', 'sum'),
        # Flatten the lists of record_ids
        record_ids=('record_ids', lambda x: [item for sublist in x for item in sublist]),
        # Tokens are identical for the same normalized_name, so take the first
        tokens=('tokens', 'first')
    ).reset_index()

    # 2. Sort deterministically by normalized_name to ensure stable ID assignment
    df_agg = df_agg.sort_values('normalized_name').reset_index(drop=True)

    # 3. Assign stable sequential IDs (e.g., NORM_00001)
    # These IDs will serve as the nodes in the Match Graph
    df_agg['canonical_id'] = [f"NORM_{i:05d}" for i in range(1, len(df_agg) + 1)]

    # Reorder columns for logical readability
    df_final = df_agg[['canonical_id', 'normalized_name', 'raw_names', 'tokens', 'occurrence_count', 'record_ids']]

    # 4. Serialize deterministically for hashing
    # We must convert lists to strict JSON strings without spaces to ensure byte-for-byte stability
    df_hashable = df_final.copy()
    df_hashable['raw_names'] = df_hashable['raw_names'].apply(lambda x: json.dumps(x, separators=(',', ':')))
    df_hashable['tokens'] = df_hashable['tokens'].apply(lambda x: json.dumps(x, separators=(',', ':')))
    df_hashable['record_ids'] = df_hashable['record_ids'].apply(lambda x: json.dumps(sorted(x), separators=(',', ':')))

    # Serialize to CSV bytes
    csv_bytes: bytes = df_hashable.to_csv(index=False).encode('utf-8')

    # 5. Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # 6. Create a deep copy of the configuration
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    # Construct the artifact
    artifact = NormalizedNameArtifact(
        data=df_final,
        content_hash_sha256=content_hash,
        unique_normalized_names_count=len(df_final),
        config_snapshot=config_snapshot
    )

    logger.info(f"Task 5, Step 3 completed. Unique normalized names (nodes): {artifact.unique_normalized_names_count}.")
    return artifact


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_5_prepare_supplier_names(
    positive_flow_artifact: PositiveFlowArtifact,
    config: Dict[str, Any]
) -> NormalizedNameArtifact:
    """
    Orchestrates the preparation of supplier names for identity harmonisation.

    This function sequentially applies string normalization, tokenization, and aggregation
    to convert raw transaction-level creditor names into a unique, cryptographically
    verifiable set of nodes ready for the Match Graph.

    Parameters
    ----------
    positive_flow_artifact : PositiveFlowArtifact
        The immutable artifact containing the strictly positive payment flows.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    NormalizedNameArtifact
        The immutable artifact containing the normalized corpus and its hash.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 5 Orchestrator: Prepare Supplier Names ===")

    # Extract the DataFrame from the incoming artifact
    df_positive: pd.DataFrame = positive_flow_artifact.data

    if df_positive.empty:
        raise ValueError("Fatal Error: Input PositiveFlowArtifact contains an empty DataFrame.")

    # Execute Step 1: Apply string normalization
    df_normalized: pd.DataFrame = normalize_supplier_names(df_positive, config)

    # Execute Step 2: Apply tokenization rules
    df_tokenized: pd.DataFrame = tokenize_normalized_names(df_normalized, config)

    # Execute Step 3: Aggregate and create the immutable artifact
    artifact: NormalizedNameArtifact = create_normalized_name_artifact(df_tokenized, config)

    logger.info(f"=== Task 5 Orchestrator completed. Artifact Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 6 — Compute fuzzy similarity metrics and apply the conservative match gate

# ==============================================================================
# Task 6: Compute fuzzy similarity metrics and apply the conservative match gate
# ==============================================================================

@dataclass
class MatchEdgeArtifact:
    """
    An immutable artifact containing the filtered match edges for the Match Graph.

    Attributes
    ----------
    edges : pd.DataFrame
        The filtered edge list. Columns: 'id_A', 'id_B', 'cos_tfidf', 'token_set_ratio',
        'jaccard_overlap', 'ensemble_score'.
    edge_count : int
        The number of edges that passed the conservative match gate.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    edges: pd.DataFrame
    edge_count: int
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 1: Generate candidate pairs deterministically
# -------------------------------------------------------------------------------------------------------------------------------
def generate_candidate_pairs(df_norm: pd.DataFrame) -> pd.DataFrame:
    """
    Generates all unique, unordered candidate pairs from the normalized corpus.

    This function implements the 'all_pairs' strategy without blocking keys, generating
    exactly N(N-1)/2 pairs using memory-efficient C-optimized iterators.

    Parameters
    ----------
    df_norm : pd.DataFrame
        The normalized corpus containing unique 'canonical_id's.

    Returns
    -------
    pd.DataFrame
        A DataFrame with columns ['id_A', 'id_B'] representing the candidate pairs.

    Raises
    ------
    ValueError
        If the input DataFrame lacks the 'canonical_id' column.
    """
    logger.info("Starting Task 6, Step 1: Generating candidate pairs.")

    if 'canonical_id' not in df_norm.columns:
        raise ValueError("Data integrity violation: 'canonical_id' missing from normalized corpus.")

    # Extract the canonical IDs as a list
    canonical_ids: List[str] = df_norm['canonical_id'].tolist()
    n_nodes: int = len(canonical_ids)

    # Calculate the expected number of pairs: N(N-1)/2
    expected_pairs: int = (n_nodes * (n_nodes - 1)) // 2
    logger.debug(f"Generating {expected_pairs} candidate pairs for {n_nodes} nodes.")

    # Use itertools.combinations to generate the exact upper triangle of pairs
    # This avoids the massive memory overhead of a pandas cross-join
    pair_generator = itertools.combinations(canonical_ids, 2)

    # Consume the generator into a DataFrame
    df_pairs = pd.DataFrame(pair_generator, columns=['id_A', 'id_B'])

    # Mathematical assertion to guarantee no silent data loss or duplication
    assert len(df_pairs) == expected_pairs, f"Expected {expected_pairs} pairs, generated {len(df_pairs)}."

    logger.info("Task 6, Step 1 completed successfully.")
    return df_pairs


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 2: Compute similarity features per candidate pair
# -------------------------------------------------------------------------------------------------------------------------------
def compute_similarity_metrics(df_pairs: pd.DataFrame, df_norm: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Computes TF-IDF Cosine, Token Set Ratio, Jaccard Overlap, and Ensemble Score for each pair.

    This function leverages sparse matrix multiplication for TF-IDF and optimized list
    comprehensions for string/set metrics to process millions of pairs efficiently.

    Parameters
    ----------
    df_pairs : pd.DataFrame
        The candidate pairs DataFrame with 'id_A' and 'id_B'.
    df_norm : pd.DataFrame
        The normalized corpus containing 'normalized_name' and 'tokens'.
    config : Dict[str, Any]
        The study configuration dictionary containing vectorizer and metric settings.

    Returns
    -------
    pd.DataFrame
        The pairs DataFrame enriched with the four similarity metric columns.
    """
    logger.info("Starting Task 6, Step 2: Computing similarity metrics.")

    # 1. Create fast lookup mappings from canonical_id to positional index, name, and tokens
    # Positional index is required for fast extraction from the dense TF-IDF similarity matrix
    id_to_idx: Dict[str, int] = {cid: idx for idx, cid in enumerate(df_norm['canonical_id'])}
    id_to_name: Dict[str, str] = dict(zip(df_norm['canonical_id'], df_norm['normalized_name']))

    # Handle potential JSON serialization of tokens from Task 5
    def _parse_tokens(val: Any) -> List[str]:
        if isinstance(val, str):
            try:
                return json.loads(val)
            except json.JSONDecodeError:
                return []
        elif isinstance(val, list):
            return val
        return []

    id_to_tokens: Dict[str, List[str]] = {
        cid: _parse_tokens(toks) for cid, toks in zip(df_norm['canonical_id'], df_norm['tokens'])
    }

    # Map the pairs to their respective data
    idx_A: np.ndarray = df_pairs['id_A'].map(id_to_idx).values
    idx_B: np.ndarray = df_pairs['id_B'].map(id_to_idx).values
    names_A: List[str] = df_pairs['id_A'].map(id_to_name).tolist()
    names_B: List[str] = df_pairs['id_B'].map(id_to_name).tolist()
    tokens_A: List[List[str]] = df_pairs['id_A'].map(id_to_tokens).tolist()
    tokens_B: List[List[str]] = df_pairs['id_B'].map(id_to_tokens).tolist()

    # 2. Compute TF-IDF Cosine Similarity
    logger.debug("Computing TF-IDF Cosine Similarity matrix.")
    tfidf_config: Dict[str, Any] = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]["tf_idf_cosine_similarity"]["vectorizer"]

    vectorizer = TfidfVectorizer(
        analyzer=tfidf_config.get("analyzer", "char_wb"),
        ngram_range=tuple(tfidf_config.get("ngram_range", (3, 5))),
        lowercase=tfidf_config.get("lowercase", False),
        sublinear_tf=tfidf_config.get("sublinear_tf", True),
        use_idf=tfidf_config.get("use_idf", True),
        smooth_idf=tfidf_config.get("smooth_idf", True),
        norm=tfidf_config.get("norm", "l2")
    )

    # Fit and transform the entire corpus. X is a sparse matrix of shape (N, vocab_size)
    X: sp.csr_matrix = vectorizer.fit_transform(df_norm['normalized_name'])

    # Compute the full pairwise cosine similarity matrix: X * X^T
    # For N=1896, this is a 1896x1896 matrix. Converting to dense is highly efficient (~28MB).
    dense_sim_matrix: np.ndarray = X.dot(X.T).toarray()

    # Extract the specific pairwise scores using advanced integer indexing
    cos_tfidf_scores: np.ndarray = dense_sim_matrix[idx_A, idx_B]
    df_pairs['cos_tfidf'] = cos_tfidf_scores

    # 3. Compute Token Set Ratio (TSR)
    logger.debug("Computing Token Set Ratios.")
    # rapidfuzz is implemented in C++ and is extremely fast in list comprehensions
    tsr_scores: List[float] = [
        fuzz.token_set_ratio(a, b) for a, b in zip(names_A, names_B)
    ]
    df_pairs['token_set_ratio'] = np.array(tsr_scores, dtype=np.float64)

    # 4. Compute Token Jaccard Overlap
    logger.debug("Computing Jaccard Overlaps.")
    def _calc_jaccard(ta: List[str], tb: List[str]) -> float:
        sa: Set[str] = set(ta)
        sb: Set[str] = set(tb)
        union_len: int = len(sa | sb)
        # Safe division to prevent ZeroDivisionError if both token sets are empty
        return float(len(sa & sb)) / union_len if union_len > 0 else 0.0

    jaccard_scores: List[float] = [
        _calc_jaccard(a, b) for a, b in zip(tokens_A, tokens_B)
    ]
    df_pairs['jaccard_overlap'] = np.array(jaccard_scores, dtype=np.float64)

    # 5. Compute Ensemble Composite Score
    # Equation: Ens = (cos_tfidf + TSR/100 + Jaccard) / 3
    logger.debug("Computing Ensemble Scores.")
    df_pairs['ensemble_score'] = (
        df_pairs['cos_tfidf'] +
        (df_pairs['token_set_ratio'] / 100.0) +
        df_pairs['jaccard_overlap']
    ) / 3.0

    logger.info("Task 6, Step 2 completed successfully.")
    return df_pairs


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 3: Apply the conservative match gate
# -------------------------------------------------------------------------------------------------------------------------------
def apply_conservative_match_gate(df_metrics: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Filters the candidate pairs using the strict, multi-condition logical gate.

    A pair is declared a MATCH if and only if all four metric thresholds are satisfied.
    The ensemble score threshold is evaluated using a strict inequality (>).

    Parameters
    ----------
    df_metrics : pd.DataFrame
        The candidate pairs DataFrame enriched with similarity metrics.
    config : Dict[str, Any]
        The study configuration dictionary containing the gate thresholds.

    Returns
    -------
    pd.DataFrame
        The filtered edge list containing only the pairs that passed the gate.
    """
    logger.info("Starting Task 6, Step 3: Applying conservative match gate.")

    # Extract thresholds from the configuration
    gates_config: Dict[str, Any] = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]

    thresh_tfidf: float = gates_config["tf_idf_cosine_similarity"]["threshold"]
    thresh_tsr: float = gates_config["token_set_ratio"]["threshold"]
    thresh_jaccard: float = gates_config["token_jaccard_overlap"]["threshold"]
    thresh_ens: float = gates_config["ensemble_composite_score"]["threshold_strict_greater_than"]

    # Construct boolean masks for each condition
    # Note: We use >= for the individual metrics and > for the ensemble, as specified in the LaTeX context
    mask_tfidf: pd.Series = df_metrics['cos_tfidf'] >= thresh_tfidf
    mask_tsr: pd.Series = df_metrics['token_set_ratio'] >= thresh_tsr
    mask_jaccard: pd.Series = df_metrics['jaccard_overlap'] >= thresh_jaccard
    mask_ens: pd.Series = df_metrics['ensemble_score'] > thresh_ens

    # Combine masks using bitwise AND
    final_mask: pd.Series = mask_tfidf & mask_tsr & mask_jaccard & mask_ens

    # Apply the mask to filter the DataFrame, creating a deep copy to prevent SettingWithCopyWarnings
    df_edges: pd.DataFrame = df_metrics.loc[final_mask].copy(deep=True)

    # Reset the index to provide a clean RangeIndex for the edge list
    df_edges.reset_index(drop=True, inplace=True)

    logger.info(f"Task 6, Step 3 completed. {len(df_edges)} edges passed the match gate.")
    return df_edges


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_6_compute_fuzzy_metrics_and_gate(
    normalized_artifact: NormalizedNameArtifact,
    config: Dict[str, Any]
) -> MatchEdgeArtifact:
    """
    Orchestrates the pairwise similarity computation and match gate application.

    This function sequentially generates all candidate pairs, computes the four required
    similarity metrics, applies the strict logical gate, and packages the resulting
    edges into an immutable artifact for the Graph Theory phase.

    Parameters
    ----------
    normalized_artifact : NormalizedNameArtifact
        The immutable artifact containing the normalized corpus (nodes).
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    MatchEdgeArtifact
        The immutable artifact containing the filtered edge list.
    """
    logger.info("=== Initiating Task 6 Orchestrator: Compute Fuzzy Metrics and Match Gate ===")

    df_norm: pd.DataFrame = normalized_artifact.data
    n_nodes: int = normalized_artifact.unique_normalized_names_count

    # Edge case handling: If there are fewer than 2 nodes, no pairs can be generated
    if n_nodes < 2:
        logger.warning("Fewer than 2 unique nodes found. Returning empty edge artifact.")
        empty_edges = pd.DataFrame(columns=['id_A', 'id_B', 'cos_tfidf', 'token_set_ratio', 'jaccard_overlap', 'ensemble_score'])
        return MatchEdgeArtifact(
            edges=empty_edges,
            edge_count=0,
            config_snapshot=copy.deepcopy(config)
        )

    # Execute Step 1: Generate candidate pairs
    df_pairs: pd.DataFrame = generate_candidate_pairs(df_norm)

    # Execute Step 2: Compute similarity metrics
    df_metrics: pd.DataFrame = compute_similarity_metrics(df_pairs, df_norm, config)

    # Execute Step 3: Apply the conservative match gate
    df_edges: pd.DataFrame = apply_conservative_match_gate(df_metrics, config)

    # Construct the immutable artifact
    artifact = MatchEdgeArtifact(
        edges=df_edges,
        edge_count=len(df_edges),
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 6 Orchestrator completed. Generated {artifact.edge_count} valid match edges. ===")

    return artifact
````

````python
# Task 7 — Build the Match Graph and resolve canonical suppliers

# ==============================================================================
# Task 7: Build the Match Graph and resolve canonical suppliers
# ==============================================================================

@dataclass
class CanonicalSupplierArtifact:
    """
    An immutable artifact containing the resolved canonical suppliers and their pseudonyms.

    Attributes
    ----------
    data : pd.DataFrame
        The resolved suppliers. Columns: 'canonical_name', 'pseudonymous_id',
        'component_size', 'total_occurrence_count', 'record_ids' (list).
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    unique_canonical_supplier_count : int
        The final number of unique suppliers after entity resolution.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    data: pd.DataFrame
    content_hash_sha256: str
    unique_canonical_supplier_count: int
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1: Construct the Match Graph
# -------------------------------------------------------------------------------------------------------------------------------
def construct_match_graph(df_norm: pd.DataFrame, df_edges: pd.DataFrame) -> nx.Graph:
    """
    Constructs the undirected Match Graph G=(V,E) from the normalized nodes and filtered edges.

    This function explicitly adds all normalized names as nodes before adding edges,
    mathematically guaranteeing that isolated suppliers (singletons) are preserved.
    It implements a strict topological validation to ensure that E is a subset of V x V,
    preventing silent graph corruption from invalid edge endpoints.

    Parameters
    ----------
    df_norm : pd.DataFrame
        The normalized corpus containing all unique 'canonical_id's (Vertices).
    df_edges : pd.DataFrame
        The filtered edge list containing 'id_A' and 'id_B' (Edges).

    Returns
    -------
    nx.Graph
        A NetworkX undirected graph object.

    Raises
    ------
    ValueError
        If the normalized corpus is empty, or if an edge references a non-existent node.
    """
    # Log the initiation of the graph construction
    logger.info("Starting Task 7, Step 1: Constructing the Match Graph.")

    # Mathematical safeguard: The graph must contain at least one vertex
    if df_norm.empty:
        raise ValueError("Fatal Error: Normalized corpus is empty. Cannot construct graph.")

    # Instantiate an empty undirected graph
    G = nx.Graph()

    # Extract the complete set of valid vertices (canonical_ids)
    vertices: List[str] = df_norm['canonical_id'].tolist()
    # Convert to a Python set for O(1) lookup efficiency during validation
    valid_vertex_set: Set[str] = set(vertices)

    # Add all vertices to the graph. This is critical to preserve singletons.
    G.add_nodes_from(vertices)
    logger.debug(f"Added {len(vertices)} nodes to the Match Graph.")

    # Extract the edges and validate topological integrity
    if not df_edges.empty:
        # Extract unique endpoints from the edge list
        edge_endpoints: Set[str] = set(df_edges['id_A']).union(set(df_edges['id_B']))

        # Compute the set difference to identify invalid endpoints
        # Equation: Invalid = Endpoints \ V
        invalid_endpoints: Set[str] = edge_endpoints - valid_vertex_set

        if invalid_endpoints:
            # Raise a hard failure if E is not a subset of V x V
            raise ValueError(
                f"Topological Violation: Found {len(invalid_endpoints)} edge endpoints "
                f"that do not exist in the vertex set V. Example: {list(invalid_endpoints)[:3]}"
            )

        # Extract the validated edges as a list of tuples
        edges: List[Tuple[str, str]] = list(zip(df_edges['id_A'], df_edges['id_B']))

        # Add the edges to the graph
        G.add_edges_from(edges)
        logger.debug(f"Added {len(edges)} validated edges to the Match Graph.")
    else:
        # Log if the graph consists entirely of isolated nodes
        logger.debug("Edge list is empty. Graph consists entirely of isolated nodes.")

    # Log successful completion
    logger.info("Task 7, Step 1 completed successfully.")

    # Return the constructed NetworkX graph
    return G


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 2: Compute connected components and assign canonical names
# -------------------------------------------------------------------------------------------------------------------------------
def resolve_canonical_suppliers(G: nx.Graph, df_norm: pd.DataFrame, config: Dict[str, Any]) -> Tuple[pd.DataFrame, Dict[str, str]]:
    """
    Extracts connected components and deterministically assigns a canonical name to each.

    The canonical name is defined as the modal raw creditor string within the component.
    Ties are broken using a lexicographic minimum rule to ensure strict reproducibility.

    Parameters
    ----------
    G : nx.Graph
        The constructed Match Graph.
    df_norm : pd.DataFrame
        The normalized corpus containing raw names and occurrence counts.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, str]]
        - df_canonical: DataFrame of resolved suppliers and their aggregated metadata.
        - raw_to_canonical_map: Dictionary mapping every raw name to its canonical name.
    """
    logger.info("Starting Task 7, Step 2: Resolving canonical suppliers via connected components.")

    # Extract the tie-breaker rule from configuration
    tie_breaker: str = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]["canonical_naming"]["tie_breaker"]
    if tie_breaker != "lexicographic_min":
        logger.warning(f"Configured tie_breaker '{tie_breaker}' is not 'lexicographic_min'. Defaulting to lexicographic_min for determinism.")

    # Extract connected components as a list of sets of node IDs
    components: List[Set[str]] = list(nx.connected_components(G))
    logger.debug(f"Extracted {len(components)} connected components.")

    # Set up the index for fast lookups
    df_norm_indexed = df_norm.set_index('canonical_id')

    resolved_records: List[Dict[str, Any]] = []
    raw_to_canonical_map: Dict[str, str] = {}

    # Iterate through each component to resolve the canonical name
    for comp_nodes in components:
        # Extract the rows corresponding to the nodes in this component
        comp_df = df_norm_indexed.loc[list(comp_nodes)]

        # Aggregate raw names and their counts across the component
        raw_name_counts: Dict[str, int] = {}
        all_record_ids: List[int] = []

        for _, row in comp_df.iterrows():
            # Deserialize the JSON strings back to Python lists
            try:
                raw_names = json.loads(row['raw_names'])
                record_ids = json.loads(row['record_ids'])
            except (json.JSONDecodeError, TypeError):
                # Fallback if data is already in list format (e.g., bypassed serialization)
                raw_names = row['raw_names'] if isinstance(row['raw_names'], list) else []
                record_ids = row['record_ids'] if isinstance(row['record_ids'], list) else []

            all_record_ids.extend(record_ids)

            # The occurrence_count in df_norm is the sum for the normalized name.
            # To find the true modal raw string, we assume uniform distribution of counts
            # among the raw names that mapped to this normalized name, or simply count
            # the raw names if we don't have transaction-level granularity here.
            # Since we need the modal string, we count the frequency of each raw name.
            for name in raw_names:
                raw_name_counts[name] = raw_name_counts.get(name, 0) + 1

        # Determine the modal string (highest count, then lexicographic min)
        # Sort by count (descending), then by name (ascending)
        sorted_names = sorted(raw_name_counts.items(), key=lambda item: (-item[1], item[0]))
        canonical_name: str = sorted_names[0][0]

        # Map all raw names in this component to the chosen canonical name
        for name in raw_name_counts.keys():
            raw_to_canonical_map[name] = canonical_name

        # Store the resolved component metadata
        resolved_records.append({
            'canonical_name': canonical_name,
            'component_size_nodes': len(comp_nodes),
            'total_occurrence_count': len(all_record_ids),
            'record_ids': all_record_ids
        })

    # Construct the resolved DataFrame
    df_canonical = pd.DataFrame(resolved_records)

    # Sort deterministically by canonical name
    df_canonical = df_canonical.sort_values('canonical_name').reset_index(drop=True)

    logger.info("Task 7, Step 2 completed successfully.")
    return df_canonical, raw_to_canonical_map


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3: Pseudonymise canonical suppliers (ethics protocol)
# -------------------------------------------------------------------------------------------------------------------------------
def apply_pseudonymization(df_canonical: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Applies cryptographic pseudonymization to the canonical supplier names.

    This function implements the ethics protocol by generating a salted SHA-256 hash
    for each canonical name, masking sensitive identities (e.g., sole traders) in
    downstream reporting artifacts.

    Parameters
    ----------
    df_canonical : pd.DataFrame
        The resolved canonical suppliers DataFrame.
    config : Dict[str, Any]
        The study configuration dictionary containing ethics protocols.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with a 'pseudonymous_id' column.
    """
    logger.info("Starting Task 7, Step 3: Applying pseudonymization protocol.")

    ethics_config: Dict[str, Any] = config["pre_processing_parameters"]["identity_harmonisation_logical_gates"]["ethics_protocol"]

    if not ethics_config.get("pseudonymization", True):
        logger.warning("Pseudonymization is disabled in configuration. Using canonical names as IDs.")
        df_canonical['pseudonymous_id'] = df_canonical['canonical_name']
        return df_canonical

    # Retrieve the global seed to use as a deterministic salt for replication purposes
    # In a production environment, this should be a secure, injected secret.
    global_seed: int = config["study_metadata"]["reproducibility"]["global_random_seed"]
    salt: str = f"YORK_COUNCIL_SALT_{global_seed}"

    def _generate_pseudonym(name: str) -> str:
        """Generates a truncated, salted SHA-256 hash for a given string."""
        salted_string: str = f"{salt}:{name}"
        hash_hex: str = hashlib.sha256(salted_string.encode('utf-8')).hexdigest()
        # Truncate to 16 characters for readability and prefix with SUPP_
        return f"SUPP_{hash_hex[:16].upper()}"

    # Apply the hashing function to the canonical names
    df_canonical['pseudonymous_id'] = df_canonical['canonical_name'].apply(_generate_pseudonym)

    # Reorder columns for logical readability
    cols = ['canonical_name', 'pseudonymous_id', 'component_size_nodes', 'total_occurrence_count', 'record_ids']
    df_canonical = df_canonical[cols]

    logger.info("Task 7, Step 3 completed. Pseudonymization applied.")
    return df_canonical


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_7_build_graph_and_resolve(
    normalized_artifact: NormalizedNameArtifact,
    edge_artifact: MatchEdgeArtifact,
    config: Dict[str, Any]
) -> Tuple[CanonicalSupplierArtifact, Dict[str, str]]:
    """
    Orchestrates the construction of the Match Graph and the resolution of canonical suppliers.

    This function sequentially builds the graph, extracts connected components, assigns
    deterministic canonical names, applies ethics protocols, and packages the results
    into an immutable artifact alongside a mapping dictionary for downstream use.

    Parameters
    ----------
    normalized_artifact : NormalizedNameArtifact
        The artifact containing the normalized corpus (nodes).
    edge_artifact : MatchEdgeArtifact
        The artifact containing the filtered edge list.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    Tuple[CanonicalSupplierArtifact, Dict[str, str]]
        - The immutable artifact containing the resolved suppliers.
        - A dictionary mapping raw 'Creditor Name's to their resolved 'canonical_name'.
    """
    logger.info("=== Initiating Task 7 Orchestrator: Build Graph and Resolve Suppliers ===")

    df_norm: pd.DataFrame = normalized_artifact.data
    df_edges: pd.DataFrame = edge_artifact.edges

    # Execute Step 1: Construct the Match Graph
    G: nx.Graph = construct_match_graph(df_norm, df_edges)

    # Execute Step 2: Compute connected components and assign canonical names
    df_canonical, raw_to_canonical_map = resolve_canonical_suppliers(G, df_norm, config)

    # Execute Step 3: Apply pseudonymization
    df_pseudonymised: pd.DataFrame = apply_pseudonymization(df_canonical, config)

    # Serialize lists to JSON for deterministic hashing
    df_hashable = df_pseudonymised.copy()
    df_hashable['record_ids'] = df_hashable['record_ids'].apply(lambda x: json.dumps(sorted(x), separators=(',', ':')))

    # Compute the cryptographic SHA-256 hash
    csv_bytes: bytes = df_hashable.to_csv(index=False).encode('utf-8')
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()

    # Construct the immutable artifact
    artifact = CanonicalSupplierArtifact(
        data=df_pseudonymised,
        content_hash_sha256=content_hash,
        unique_canonical_supplier_count=len(df_pseudonymised),
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 7 Orchestrator completed. Resolved {artifact.unique_canonical_supplier_count} canonical suppliers. ===")

    return artifact, raw_to_canonical_map
````

````python
# Task 8 — Select the high-volume cohort and compute global robust scaling parameters

# ===================================================================================
# Task 8: Select the high-volume cohort and compute global robust scaling parameters
# ===================================================================================

@dataclass
class ScalingParameters:
    """
    A strongly-typed dataclass encapsulating the global robust scaling parameters.

    Attributes
    ----------
    global_median : float
        The median (A_tilde) of all positive payment amounts in the analysis window.
    q1 : float
        The 25th percentile of all positive payment amounts.
    q3 : float
        The 75th percentile of all positive payment amounts.
    iqr : float
        The Interquartile Range (Q3 - Q1), used as the robust scale factor.
    scope_row_count : int
        The number of records used to compute these parameters (audit trail).
    """
    global_median: float
    q1: float
    q3: float
    iqr: float
    scope_row_count: int

@dataclass
class ScaledCohortArtifact:
    """
    An immutable artifact containing the standardized, high-volume cohort transactions.

    Attributes
    ----------
    data : pd.DataFrame
        The transaction-level data enriched with 'canonical_name' and 'standardised_amount'.
    scaling_parameters : ScalingParameters
        The global parameters used to standardize the amounts.
    cohort_suppliers : List[str]
        The unique list of canonical supplier names that met the n >= 50 threshold.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    data: pd.DataFrame
    scaling_parameters: ScalingParameters
    cohort_suppliers: List[str]
    content_hash_sha256: str
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: Compute supplier transaction counts and filter to the high-volume cohort
# -------------------------------------------------------------------------------------------------------------------------------
def select_high_volume_cohort(
    df_positive: pd.DataFrame,
    raw_to_canonical_map: Dict[str, str],
    config: Dict[str, Any]
) -> Tuple[pd.DataFrame, List[str]]:
    """
    Maps transactions to canonical identities and filters to the high-volume cohort (n >= 50).

    This function strictly enforces the order of operations: identity resolution must
    precede frequency counting to prevent lexical variants from artificially splitting
    transaction volumes and causing erroneous exclusions.

    Parameters
    ----------
    df_positive : pd.DataFrame
        The transaction-level DataFrame containing strictly positive payments.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.
    config : Dict[str, Any]
        The study configuration dictionary containing the frequency threshold.

    Returns
    -------
    Tuple[pd.DataFrame, List[str]]
        - df_cohort: The filtered transaction-level DataFrame.
        - cohort_suppliers: A list of unique canonical names in the cohort.

    Raises
    ------
    ValueError
        If any raw Creditor Name fails to map to a canonical identity.
    """
    logger.info("Starting Task 8, Step 1: Selecting the high-volume cohort.")

    # Create a defensive copy to preserve immutability
    df = df_positive.copy(deep=True)

    # 1. Map raw names to canonical identities
    df['canonical_name'] = df['Creditor Name'].map(raw_to_canonical_map)

    # Defensive assertion: Ensure perfect mapping coverage
    if df['canonical_name'].isna().any():
        unmapped_count = df['canonical_name'].isna().sum()
        raise ValueError(f"Data integrity violation: {unmapped_count} raw names failed to map to a canonical identity.")

    # Extract the frequency threshold from the configuration
    threshold_n: int = config["cohort_selection_and_scaling"]["frequency_threshold_n"]

    # 2. Compute transaction counts per canonical supplier
    # Using transform('size') broadcasts the aggregated count back to every row efficiently
    df['supplier_txn_count'] = df.groupby('canonical_name')['record_id'].transform('size')

    # 3. Create the boolean mask for the high-volume cohort (Equation: n_s >= 50)
    cohort_mask: pd.Series = df['supplier_txn_count'] >= threshold_n

    # 4. Filter the DataFrame
    df_cohort: pd.DataFrame = df.loc[cohort_mask].copy(deep=True)

    # Extract the unique list of cohort suppliers
    cohort_suppliers: List[str] = sorted(df_cohort['canonical_name'].unique().tolist())

    # Diagnostic logging
    logger.info(f"Cohort selection complete. Retained {len(cohort_suppliers)} suppliers and {len(df_cohort)} records.")

    return df_cohort, cohort_suppliers


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: Compute global median and IQR on the configured scaling scope
# -------------------------------------------------------------------------------------------------------------------------------
def compute_global_scaling_parameters(df_positive: pd.DataFrame, config: Dict[str, Any]) -> Any: # Returns ScalingParameters
    """
    Computes the global median and Interquartile Range (IQR) across all positive payments.

    This function applies a strict `.dropna()` to the target series prior to computation,
    mathematically guaranteeing that the non-parametric estimator operates exclusively on
    finite real numbers, preventing NaN propagation into the global scaling frame.

    Parameters
    ----------
    df_positive : pd.DataFrame
        The full transaction-level DataFrame containing strictly positive payments.
    config : Dict[str, Any]
        The study configuration dictionary containing quantile conventions.

    Returns
    -------
    ScalingParameters
        A dataclass containing the computed global scalars.

    Raises
    ------
    ValueError
        If the computed IQR is exactly zero or negative, or if the series is empty.
    """
    # Log the initiation of the global scaling parameter computation
    logger.info("Starting Task 8, Step 2: Computing global scaling parameters.")

    # Extract the interpolation method from the configuration
    interp_method: str = config["cohort_selection_and_scaling"]["quantile_conventions"]["interpolation_method"]

    # Extract the Net Amount series and strictly drop NaNs to guarantee finite inputs
    amounts: pd.Series = df_positive['Net Amount'].dropna()

    # Mathematical safeguard: Ensure the series is not empty after dropping NaNs
    if amounts.empty:
        raise ValueError("Fatal Error: 'Net Amount' series is empty after dropping NaNs.")

    # Compute the required quantiles using the configured interpolation method
    # Equation: A_tilde = Q_0.50, Q1 = Q_0.25, Q3 = Q_0.75
    quantiles: pd.Series = amounts.quantile([0.25, 0.50, 0.75], interpolation=interp_method)

    # Extract the scalar values
    q1: float = float(quantiles.loc[0.25])
    global_median: float = float(quantiles.loc[0.50])
    q3: float = float(quantiles.loc[0.75])

    # Compute the Interquartile Range
    # Equation: IQR = Q3 - Q1
    iqr: float = q3 - q1

    # Mathematical safeguard: IQR must be strictly positive for division
    if iqr <= 0.0:
        raise ValueError(f"Fatal Error: Computed global IQR is {iqr}. Robust standardization is mathematically undefined.")

    # Construct the parameters artifact (Assuming ScalingParameters is imported/available)
    params = ScalingParameters(
        global_median=global_median,
        q1=q1,
        q3=q3,
        iqr=iqr,
        scope_row_count=len(amounts)
    )

    # Log the computed scalars
    logger.debug(f"Computed Scaling Parameters: Median={global_median:.2f}, IQR={iqr:.2f}")
    logger.info("Task 8, Step 2 completed successfully.")

    # Return the instantiated dataclass
    return params


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: Robustly standardize each payment amount
# -------------------------------------------------------------------------------------------------------------------------------
def apply_robust_standardization(df_cohort: pd.DataFrame, params: ScalingParameters) -> pd.DataFrame:
    """
    Applies robust standardization to the payment amounts using the global parameters.

    This transformation places all suppliers on a common, scale-free monetary reference
    frame while preserving relative distances and resisting extreme outliers. Negative
    values in the output simply indicate payments below the global median.

    Parameters
    ----------
    df_cohort : pd.DataFrame
        The filtered high-volume cohort DataFrame.
    params : ScalingParameters
        The global scaling parameters (Median and IQR).

    Returns
    -------
    pd.DataFrame
        The cohort DataFrame enriched with a 'standardised_amount' column.
    """
    logger.info("Starting Task 8, Step 3: Applying robust standardization.")

    # Create a defensive copy to preserve immutability
    df = df_cohort.copy(deep=True)

    # Extract scalars for readability
    a_tilde: float = params.global_median
    iqr: float = params.iqr

    # Apply the robust standardization equation: a_tilde_i = (a_i - A_tilde) / IQR
    # This is a fully vectorized operation over the pandas Series
    df['standardised_amount'] = (df['Net Amount'] - a_tilde) / iqr

    # Ensure the output is strictly float64 for downstream GMM precision
    df['standardised_amount'] = df['standardised_amount'].astype(np.float64)

    logger.info("Task 8, Step 3 completed successfully.")
    return df


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_8_select_cohort_and_scale(
    positive_flow_artifact: PositiveFlowArtifact,
    raw_to_canonical_map: Dict[str, str],
    config: Dict[str, Any]
) -> ScaledCohortArtifact:
    """
    Orchestrates the selection of the high-volume cohort and robust standardization.

    This function strictly enforces the methodological order of operations: global scaling
    parameters must be computed on the full positive-flow dataset before the cohort filter
    is applied. It then standardizes the cohort and packages the results into an immutable artifact.

    Parameters
    ----------
    positive_flow_artifact : PositiveFlowArtifact
        The immutable artifact containing the strictly positive payment flows.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ScaledCohortArtifact
        The immutable artifact containing the standardized cohort data and scaling parameters.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 8 Orchestrator: Select Cohort and Scale ===")

    df_positive: pd.DataFrame = positive_flow_artifact.data

    if df_positive.empty:
        raise ValueError("Fatal Error: Input PositiveFlowArtifact contains an empty DataFrame.")

    # Execute Step 2 FIRST: Compute global scaling parameters on the FULL dataset
    # This prevents look-ahead bias and ensures the true global reference frame is used
    scaling_params: ScalingParameters = compute_global_scaling_parameters(df_positive, config)

    # Execute Step 1: Map identities and filter to the high-volume cohort
    df_cohort, cohort_suppliers = select_high_volume_cohort(df_positive, raw_to_canonical_map, config)

    # Execute Step 3: Apply robust standardization to the cohort
    df_scaled: pd.DataFrame = apply_robust_standardization(df_cohort, scaling_params)

    # Ensure deterministic serialization for the cryptographic hash
    df_sorted: pd.DataFrame = df_scaled.sort_values(by='record_id').reset_index(drop=True)
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted = df_sorted[sorted_columns]

    csv_bytes: bytes = df_sorted.to_csv(index=False, float_format='%.6f', date_format='%Y-%m-%dT%H:%M:%S').encode('utf-8')
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()

    # Construct the immutable artifact
    artifact = ScaledCohortArtifact(
        data=df_scaled, # Store original order
        scaling_parameters=scaling_params,
        cohort_suppliers=cohort_suppliers,
        content_hash_sha256=content_hash,
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 8 Orchestrator completed. Artifact Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 9 — Fit per-supplier GMMs with BIC model selection (spherical covariance) in the configured fit space

# ==============================================================================
# Task 9: Fit per-supplier GMMs with BIC model selection (spherical covariance)
# ==============================================================================

@dataclass
class ScaledGMMArtifact:
    """
    An immutable artifact containing the selected GMM parameters in standardized units.

    Attributes
    ----------
    parameters : pd.DataFrame
        The component-level parameters. Columns: 'canonical_name', 'component_index',
        'k_star', 'bic_score', 'weight_pi', 'mean_mu_scaled', 'std_dev_s_scaled', 'converged'.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    supplier_count : int
        The number of unique suppliers modeled.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    parameters: pd.DataFrame
    content_hash_sha256: str
    supplier_count: int
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: Set k_max per supplier using the manuscript constraint
# -------------------------------------------------------------------------------------------------------------------------------
def compute_supplier_complexity_caps(df_scaled: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Computes the maximum number of GMM components (k_max) permitted for each supplier.

    This function strictly implements the manuscript's heuristic constraint:
    k_max = min(4, floor(n / 25)), preventing overfitting and maintaining interpretability.

    Parameters
    ----------
    df_scaled : pd.DataFrame
        The transaction-level DataFrame containing the high-volume cohort.
    config : Dict[str, Any]
        The study configuration dictionary containing the model selection heuristics.

    Returns
    -------
    pd.DataFrame
        A supplier-level DataFrame with columns ['canonical_name', 'n_transactions', 'k_max'].

    Raises
    ------
    ValueError
        If any computed k_max is less than 1.
    """
    logger.info("Starting Task 9, Step 1: Computing supplier complexity caps (k_max).")

    # Extract heuristics from the configuration
    model_config: Dict[str, Any] = config["gmm_hyperparameters"]["model_selection"]
    max_cap: int = model_config["max_components_cap"]
    obs_per_param: int = config["cohort_selection_and_scaling"]["observations_per_parameter"]

    # Group by supplier and compute the transaction count (n)
    df_caps = df_scaled.groupby('canonical_name').size().reset_index(name='n_transactions')

    # Apply the equation: k_max = min(cap, floor(n / obs_per_param))
    # We use integer division (//) to guarantee exact floor behavior
    df_caps['k_max'] = np.minimum(max_cap, df_caps['n_transactions'] // obs_per_param)

    # Defensive assertion: k_max must be at least 1 for GMM fitting to be mathematically possible
    if (df_caps['k_max'] < 1).any():
        violators = df_caps[df_caps['k_max'] < 1]['canonical_name'].tolist()
        raise ValueError(f"Fatal Error: Computed k_max < 1 for suppliers: {violators}. Cohort filter (n>=50) likely failed.")

    logger.debug(f"Computed complexity caps for {len(df_caps)} suppliers. Max k_max observed: {df_caps['k_max'].max()}.")
    logger.info("Task 9, Step 1 completed successfully.")

    return df_caps


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: Fit candidate GMMs and select by BIC
# -------------------------------------------------------------------------------------------------------------------------------
def fit_and_select_gmms(df_scaled: pd.DataFrame, df_caps: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Fits candidate GMMs for each supplier and selects the optimal architecture via BIC.

    This function iterates through k=1...k_max, fits a spherical GMM, and extracts the
    parameters of the model that minimizes the BIC. It implements a strict numerical floor
    (`np.clip`) on the extracted variances prior to the square root operation, mathematically
    preventing NaNs caused by IEEE 754 floating-point imprecision during regime collapse.

    Parameters
    ----------
    df_scaled : pd.DataFrame
        The transaction-level DataFrame containing 'standardised_amount'.
    df_caps : pd.DataFrame
        The supplier-level DataFrame containing 'k_max'.
    config : Dict[str, Any]
        The study configuration dictionary containing EM hyperparameters.

    Returns
    -------
    pd.DataFrame
        A component-level DataFrame containing the selected GMM parameters.
    """
    # Log the initiation of the GMM fitting and selection process
    logger.info("Starting Task 9, Step 2: Fitting GMMs and selecting via BIC.")

    # Extract EM hyperparameters from the configuration
    em_config: Dict[str, Any] = config["gmm_hyperparameters"]["em_algorithm_settings"]
    cov_type: str = em_config["covariance_type"]
    init_params: str = em_config["init_params"]
    n_init: int = em_config["n_init"]
    tol: float = em_config["convergence_tol"]
    max_iter: int = em_config["max_iter"]
    random_seed: int = em_config["random_state_seed"]
    reg_covar: float = em_config["regularizer_lambda"]

    # Create a dictionary mapping supplier to their k_max for fast O(1) lookup
    k_max_map: Dict[str, int] = dict(zip(df_caps['canonical_name'], df_caps['k_max']))

    # Group the transaction data by supplier
    grouped = df_scaled.groupby('canonical_name')

    # Initialize a list to collect the component records
    records: List[Dict[str, Any]] = []
    total_suppliers: int = len(k_max_map)
    processed: int = 0

    # Iterate through each supplier's transaction distribution
    for supplier, group in grouped:
        # Extract the 1D array of standardized amounts and reshape to 2D (n_samples, 1) for sklearn
        X: np.ndarray = group['standardised_amount'].values.reshape(-1, 1)
        k_max: int = k_max_map[supplier]

        # Initialize tracking variables for the optimal model
        best_bic: float = float('inf')
        best_k: int = 1
        best_model: GaussianMixture = None

        # Iterate candidate components k from 1 to k_max inclusive
        for k in range(1, k_max + 1):
            # Instantiate the GMM with strict adherence to the configured hyperparameters
            gmm = GaussianMixture(
                n_components=k,
                covariance_type=cov_type,
                tol=tol,
                reg_covar=reg_covar,
                max_iter=max_iter,
                n_init=n_init,
                init_params=init_params,
                random_state=random_seed
            )

            # Fit the model using the Expectation-Maximization algorithm
            gmm.fit(X)

            # Compute the Bayesian Information Criterion
            # Equation: BIC = -2 * ln(L) + p * ln(n)
            current_bic: float = gmm.bic(X)

            # Select the model that minimizes BIC
            if current_bic < best_bic:
                best_bic = current_bic
                best_k = k
                best_model = gmm

        # Extract parameters from the optimal model
        weights: np.ndarray = best_model.weights_
        means: np.ndarray = best_model.means_.flatten()

        # CRITICAL SAFEGUARD: sklearn's 'spherical' covariance returns variances.
        # Due to IEEE 754 imprecision, highly clustered data can yield negative variances (e.g., -1e-15).
        # We apply a strict numerical floor aligned with the regularizer_lambda to guarantee positivity.
        variances: np.ndarray = best_model.covariances_
        clipped_variances: np.ndarray = np.clip(variances, a_min=reg_covar, a_max=None)

        # Compute the standard deviation
        # Equation: s_i = sqrt(clipped_variance_i)
        std_devs: np.ndarray = np.sqrt(clipped_variances)

        # Store each component as a separate record
        for i in range(best_k):
            records.append({
                'canonical_name': supplier,
                'component_index': i,
                'k_star': best_k,
                'bic_score': best_bic,
                'weight_pi': float(weights[i]),
                'mean_mu_scaled': float(means[i]),
                'std_dev_s_scaled': float(std_devs[i]),
                'converged': best_model.converged_
            })

        # Increment the processed counter and log progress
        processed += 1
        if processed % max(1, total_suppliers // 10) == 0:
            logger.debug(f"Processed {processed}/{total_suppliers} suppliers.")

    # Convert the records to a structured pandas DataFrame
    df_params = pd.DataFrame(records)

    # Log successful completion
    logger.info("Task 9, Step 2 completed successfully.")

    # Return the component-level DataFrame
    return df_params


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: Persist GMM parameters per supplier in scaled units
# -------------------------------------------------------------------------------------------------------------------------------
def create_scaled_gmm_artifact(df_params: pd.DataFrame, config: Dict[str, Any]) -> ScaledGMMArtifact:
    """
    Creates a cryptographically verifiable artifact of the selected GMM parameters.

    To ensure the hash is deterministic across different environments, the DataFrame is
    sorted by supplier and component index, and serialized to CSV with strict floating-point
    formatting before SHA-256 hashing.

    Parameters
    ----------
    df_params : pd.DataFrame
        The component-level DataFrame containing the selected GMM parameters.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ScaledGMMArtifact
        A dataclass containing the parameters, its SHA-256 hash, and metadata.
    """
    logger.info("Starting Task 9, Step 3: Creating immutable scaled GMM artifact.")

    # 1. Ensure deterministic row order by sorting on the compound primary key
    df_sorted: pd.DataFrame = df_params.sort_values(by=['canonical_name', 'component_index']).reset_index(drop=True)

    # 2. Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted = df_sorted[sorted_columns]

    # 3. Serialize to a deterministic string representation
    # We use strict formatting (%.8f) to truncate negligible precision differences across architectures
    csv_bytes: bytes = df_sorted.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # 4. Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # 5. Count the unique suppliers modeled
    supplier_count: int = df_sorted['canonical_name'].nunique()

    # 6. Create a deep copy of the configuration to freeze the state
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    # Construct the artifact
    artifact = ScaledGMMArtifact(
        parameters=df_params, # Store the original (unsorted) DataFrame
        content_hash_sha256=content_hash,
        supplier_count=supplier_count,
        config_snapshot=config_snapshot
    )

    logger.info("Task 9, Step 3 completed. Immutable GMM artifact generated.")
    return artifact


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_9_fit_gmms(
    scaled_cohort_artifact: ScaledCohortArtifact,
    config: Dict[str, Any]
) -> ScaledGMMArtifact:
    """
    Orchestrates the fitting of per-supplier GMMs and BIC model selection.

    This function sequentially computes the complexity caps, executes the Expectation-Maximization
    algorithm to fit candidate models, selects the optimal architecture via BIC, and packages
    the resulting parameters into a cryptographically verifiable artifact.

    Parameters
    ----------
    scaled_cohort_artifact : ScaledCohortArtifact
        The immutable artifact containing the standardized cohort transactions.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ScaledGMMArtifact
        The immutable artifact containing the selected GMM parameters in scaled units.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 9 Orchestrator: Fit GMMs and Select via BIC ===")

    df_scaled: pd.DataFrame = scaled_cohort_artifact.data

    if df_scaled.empty:
        raise ValueError("Fatal Error: Input ScaledCohortArtifact contains an empty DataFrame.")

    # Execute Step 1: Compute supplier complexity caps (k_max)
    df_caps: pd.DataFrame = compute_supplier_complexity_caps(df_scaled, config)

    # Execute Step 2: Fit candidate GMMs and select the optimal model via BIC
    df_params: pd.DataFrame = fit_and_select_gmms(df_scaled, df_caps, config)

    # Execute Step 3: Create the immutable artifact
    artifact: ScaledGMMArtifact = create_scaled_gmm_artifact(df_params, config)

    logger.info(f"=== Task 9 Orchestrator completed. Modeled {artifact.supplier_count} suppliers. Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 10 — Prune negligible mixture components and back-transform regime parameters

# =================================================================================
# Task 10: Prune negligible mixture components and back-transform regime parameters
# =================================================================================

@dataclass
class PrunedGMMArtifact:
    """
    An immutable artifact containing the pruned, renormalized, and back-transformed GMM parameters.

    Attributes
    ----------
    parameters : pd.DataFrame
        The component-level parameters. Columns include scaled metrics ('weight_pi',
        'mean_mu_scaled', 'std_dev_s_scaled') and back-transformed metrics
        ('mean_mu_gbp', 'std_dev_s_gbp').
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    supplier_count : int
        The number of unique suppliers retained after pruning.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    # The core DataFrame containing the finalized regime parameters
    parameters: pd.DataFrame
    # Cryptographic hash ensuring the immutability and traceability of the artifact
    content_hash_sha256: str
    # Count of unique suppliers to verify no entities were entirely dropped during pruning
    supplier_count: int
    # Frozen configuration state for reproducibility
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 1: Prune components with negligible weight
# -------------------------------------------------------------------------------------------------------------------------------
def prune_negligible_components(df_params: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Removes GMM components whose mixture weight falls below the configured threshold.

    This function implements the stochastic pruning logic to filter out statistically
    insignificant regimes (noise), focusing the downstream PHI index on dominant
    operational behaviors.

    Parameters
    ----------
    df_params : pd.DataFrame
        The component-level DataFrame containing the selected GMM parameters.
    config : Dict[str, Any]
        The study configuration dictionary containing the pruning threshold.

    Returns
    -------
    pd.DataFrame
        A filtered DataFrame containing only the retained components.

    Raises
    ------
    ValueError
        If the pruning process results in a supplier losing all of its components.
    """
    logger.info("Starting Task 10, Step 1: Pruning negligible mixture components.")

    # Create a defensive deep copy to ensure the upstream artifact remains immutable
    df_pruned: pd.DataFrame = df_params.copy(deep=True)

    # Extract the pruning threshold from the configuration dictionary
    pruning_config: Dict[str, Any] = config["gmm_hyperparameters"]["mixture_pruning"]
    min_weight_pi: float = pruning_config["min_weight_pi"]

    # Record the number of unique suppliers before pruning for validation
    initial_supplier_count: int = df_pruned['canonical_name'].nunique()
    initial_component_count: int = len(df_pruned)

    # Construct the boolean mask for retention
    # Equation/Logic: Retain component i if \pi_i >= 0.05
    retention_mask: pd.Series = df_pruned['weight_pi'] >= min_weight_pi

    # Apply the mask to filter the DataFrame
    df_pruned = df_pruned.loc[retention_mask].copy(deep=True)

    # Record the number of unique suppliers after pruning
    final_supplier_count: int = df_pruned['canonical_name'].nunique()
    final_component_count: int = len(df_pruned)

    # Mathematical safeguard: Ensure no supplier was completely eliminated
    # A high-volume supplier (n >= 50) should mathematically possess at least one valid regime
    if final_supplier_count < initial_supplier_count:
        # Identify which suppliers were dropped for forensic logging
        initial_suppliers: set = set(df_params['canonical_name'].unique())
        final_suppliers: set = set(df_pruned['canonical_name'].unique())
        dropped_suppliers: set = initial_suppliers - final_suppliers

        # Raise a hard failure as this violates the structural assumptions of the PHI framework
        raise ValueError(
            f"Fatal Error: Pruning eliminated all components for {len(dropped_suppliers)} suppliers. "
            f"Suppliers affected: {list(dropped_suppliers)[:5]}..."
        )

    logger.debug(f"Pruned {initial_component_count - final_component_count} components. Retained {final_component_count}.")
    logger.info("Task 10, Step 1 completed successfully.")

    return df_pruned


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 2: Renormalize retained weights
# -------------------------------------------------------------------------------------------------------------------------------
def renormalize_component_weights(df_pruned: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Renormalizes the weights of the retained components so they sum to unity per supplier.

    This function ensures that the mixture weights (\pi'_i) form a valid probability
    distribution for each supplier after the negligible components have been removed.

    Parameters
    ----------
    df_pruned : pd.DataFrame
        The DataFrame containing only the retained GMM components.
    config : Dict[str, Any]
        The study configuration dictionary containing the renormalization flag.

    Returns
    -------
    pd.DataFrame
        The DataFrame with updated, renormalized 'weight_pi' values.

    Raises
    ------
    ValueError
        If the sum of retained weights for any supplier is zero or negative.
    """
    logger.info("Starting Task 10, Step 2: Renormalizing retained component weights.")

    # Create a defensive deep copy to prevent SettingWithCopyWarnings
    df_renorm: pd.DataFrame = df_pruned.copy(deep=True)

    # Extract the renormalization flag from the configuration
    pruning_config: Dict[str, Any] = config["gmm_hyperparameters"]["mixture_pruning"]
    should_renormalize: bool = pruning_config.get("renormalize", True)

    if not should_renormalize:
        # If configuration explicitly disables renormalization, bypass the logic
        logger.warning("Methodological deviation: Renormalization is disabled in configuration.")
        return df_renorm

    # Calculate the sum of the retained weights for each supplier
    # Using transform('sum') broadcasts the scalar sum back to the original row indices
    supplier_weight_sums: pd.Series = df_renorm.groupby('canonical_name')['weight_pi'].transform('sum')

    # Mathematical safeguard: Ensure the denominator is strictly positive to prevent division by zero
    if (supplier_weight_sums <= 0.0).any():
        raise ValueError("Fatal Error: Sum of retained weights is zero or negative for one or more suppliers.")

    # Apply the renormalization equation
    # Equation: \pi'_i = \pi_i / \sum_{j \in \mathcal{J}} \pi_j
    df_renorm['weight_pi'] = (df_renorm['weight_pi'] / supplier_weight_sums).astype(np.float64)

    # Validation: Ensure the new weights sum to 1.0 for every supplier (within float tolerance)
    # Group by supplier and sum the new weights
    validation_sums: pd.Series = df_renorm.groupby('canonical_name')['weight_pi'].sum()

    # Use np.isclose to account for standard IEEE 754 floating-point arithmetic imprecision
    if not np.allclose(validation_sums, 1.0, atol=1e-5):
        raise ValueError("Fatal Error: Renormalized weights do not sum to unity.")

    logger.debug("All retained component weights successfully renormalized to sum to 1.0.")
    logger.info("Task 10, Step 2 completed successfully.")

    return df_renorm


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3: Back-transform regime parameters to GBP
# -------------------------------------------------------------------------------------------------------------------------------
def backtransform_to_gbp(df_renorm: pd.DataFrame, scaling_params: ScalingParameters) -> pd.DataFrame:
    """
    Transforms the scaled GMM means and standard deviations back to the original GBP space.

    This function applies the inverse of the robust standardization equation. The resulting
    GBP parameters are appended as new columns to preserve the scaled parameters required
    for the downstream PHI computation.

    Parameters
    ----------
    df_renorm : pd.DataFrame
        The DataFrame containing the renormalized, scaled GMM parameters.
    scaling_params : ScalingParameters
        The dataclass containing the global median and IQR used for the original scaling.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with 'mean_mu_gbp' and 'std_dev_s_gbp' columns.

    Raises
    ------
    ValueError
        If any scaled standard deviation is negative prior to transformation.
    """
    logger.info("Starting Task 10, Step 3: Back-transforming regime parameters to GBP.")

    # Create a defensive deep copy to preserve immutability
    df_final: pd.DataFrame = df_renorm.copy(deep=True)

    # Extract the global scaling scalars
    a_tilde: float = scaling_params.global_median
    iqr: float = scaling_params.iqr

    # Mathematical safeguard: Standard deviations must be non-negative
    if (df_final['std_dev_s_scaled'] < 0.0).any():
        raise ValueError("Data integrity violation: Negative scaled standard deviation detected.")

    # Apply the back-transformation equation for the mean
    # Equation: \mu_i^{GBP} = \mu_i^{scaled} \cdot IQR + \tilde{A}
    df_final['mean_mu_gbp'] = (df_final['mean_mu_scaled'] * iqr) + a_tilde
    df_final['mean_mu_gbp'] = df_final['mean_mu_gbp'].astype(np.float64)

    # Apply the back-transformation equation for the standard deviation
    # Equation: s_i^{GBP} = s_i^{scaled} \cdot IQR
    df_final['std_dev_s_gbp'] = df_final['std_dev_s_scaled'] * iqr
    df_final['std_dev_s_gbp'] = df_final['std_dev_s_gbp'].astype(np.float64)

    logger.debug("Successfully appended 'mean_mu_gbp' and 'std_dev_s_gbp' columns.")
    logger.info("Task 10, Step 3 completed successfully.")

    return df_final


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_10_prune_and_backtransform(
    scaled_gmm_artifact: ScaledGMMArtifact,
    scaling_params: ScalingParameters,
    config: Dict[str, Any]
) -> PrunedGMMArtifact:
    """
    Orchestrates the pruning, renormalization, and back-transformation of GMM parameters.

    This function sequentially executes the stochastic pruning of negligible regimes,
    renormalizes the surviving weights to ensure a valid probability distribution,
    computes the GBP-space parameters for downstream threshold anchoring, and packages
    the results into a cryptographically verifiable artifact.

    Parameters
    ----------
    scaled_gmm_artifact : ScaledGMMArtifact
        The immutable artifact containing the unpruned, scaled GMM parameters.
    scaling_params : ScalingParameters
        The dataclass containing the global median and IQR.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    PrunedGMMArtifact
        The immutable artifact containing the finalized regime parameters.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 10 Orchestrator: Prune and Back-Transform GMMs ===")

    # Extract the parameters DataFrame from the incoming artifact
    df_params: pd.DataFrame = scaled_gmm_artifact.parameters

    if df_params.empty:
        raise ValueError("Fatal Error: Input ScaledGMMArtifact contains an empty DataFrame.")

    # Execute Step 1: Prune components with negligible weight
    df_pruned: pd.DataFrame = prune_negligible_components(df_params, config)

    # Execute Step 2: Renormalize the retained weights
    df_renorm: pd.DataFrame = renormalize_component_weights(df_pruned, config)

    # Execute Step 3: Back-transform the parameters to GBP space
    df_final: pd.DataFrame = backtransform_to_gbp(df_renorm, scaling_params)

    # Ensure deterministic row order by sorting on the compound primary key
    df_sorted: pd.DataFrame = df_final.sort_values(by=['canonical_name', 'component_index']).reset_index(drop=True)

    # Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted = df_sorted[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_sorted.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Count the unique suppliers retained
    supplier_count: int = df_sorted['canonical_name'].nunique()

    # Construct the immutable artifact
    artifact = PrunedGMMArtifact(
        parameters=df_final, # Store the original (unsorted) DataFrame to preserve logical order
        content_hash_sha256=content_hash,
        supplier_count=supplier_count,
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 10 Orchestrator completed. Finalized {artifact.supplier_count} suppliers. Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 11 — Compute PHI components and PHI per supplier

# ==============================================================================
# Task 11: Compute PHI components and PHI per supplier
# ==============================================================================

@dataclass
class PHIArtifact:
    """
    An immutable artifact containing the computed PHI scores and structural components.

    Attributes
    ----------
    data : pd.DataFrame
        The supplier-level DataFrame. Columns include 'canonical_name', 'pseudonymous_id',
        'M', 'A', 'T', 'D', 'PHI', and the five empirical quantiles.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    supplier_count : int
        The number of unique suppliers scored.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    # The core DataFrame containing the PHI scores and components
    data: pd.DataFrame
    # Cryptographic hash ensuring the immutability and traceability of the artifact
    content_hash_sha256: str
    # Count of unique suppliers to verify cohort integrity
    supplier_count: int
    # Frozen configuration state for reproducibility
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: Compute required empirical quantiles using the fixed convention
# -------------------------------------------------------------------------------------------------------------------------------
def compute_empirical_quantiles(df_scaled: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Computes the required non-parametric quantile statistics on the standardized payment distributions.

    This function groups the transaction-level data by supplier and calculates the 5th, 25th,
    50th, 75th, and 95th percentiles using the configured interpolation method. The output
    is pivoted into a wide-format DataFrame for vectorized downstream arithmetic.

    Parameters
    ----------
    df_scaled : pd.DataFrame
        The transaction-level DataFrame containing 'standardised_amount' for the high-volume cohort.
    config : Dict[str, Any]
        The study configuration dictionary containing quantile conventions.

    Returns
    -------
    pd.DataFrame
        A supplier-level DataFrame indexed by 'canonical_name' with columns for each quantile.

    Raises
    ------
    ValueError
        If the quantile computation results in missing values or violates monotonicity.
    """
    logger.info("Starting Task 11, Step 1: Computing empirical quantiles.")

    # Extract quantile conventions from the configuration dictionary
    quantile_config: Dict[str, Any] = config["cohort_selection_and_scaling"]["quantile_conventions"]
    quantiles_required: List[float] = quantile_config["quantiles_required"]
    interp_method: str = quantile_config["interpolation_method"]

    # Group by canonical_name and compute quantiles on the robustly standardized amounts
    # This utilizes optimized Cython routines and returns a MultiIndex Series
    quantiles_series: pd.Series = df_scaled.groupby('canonical_name')['standardised_amount'].quantile(
        q=quantiles_required,
        interpolation=interp_method
    )

    # Unstack the inner index (quantile levels) to pivot the data into distinct columns
    df_quantiles: pd.DataFrame = quantiles_series.unstack(level=-1)

    # Rename columns to ensure safe, explicit access in downstream mathematical equations
    column_mapping: Dict[float, str] = {
        0.05: 'q_05',
        0.25: 'q_25',
        0.50: 'q_50',
        0.75: 'q_75',
        0.95: 'q_95'
    }
    df_quantiles = df_quantiles.rename(columns=column_mapping)

    # Mathematical safeguard: Ensure no nulls exist in the quantile computations
    if df_quantiles.isna().any().any():
        raise ValueError("Fatal Error: Quantile computation resulted in NaN values. Check cohort sample sizes.")

    # Mathematical safeguard: Validate empirical cumulative distribution function monotonicity
    # Q_0.05 <= Q_0.25 <= Q_0.50 <= Q_0.75 <= Q_0.95 must hold true for all distributions
    monotonicity_check = (
        (df_quantiles['q_05'] <= df_quantiles['q_25']) &
        (df_quantiles['q_25'] <= df_quantiles['q_50']) &
        (df_quantiles['q_50'] <= df_quantiles['q_75']) &
        (df_quantiles['q_75'] <= df_quantiles['q_95'])
    )

    if not monotonicity_check.all():
        raise ValueError("Fatal Error: Computed quantiles violate strict monotonicity constraints.")

    logger.debug(f"Computed empirical quantiles for {len(df_quantiles)} suppliers.")
    logger.info("Task 11, Step 1 completed successfully.")

    return df_quantiles


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: Compute M, A, T, D exactly
# -------------------------------------------------------------------------------------------------------------------------------
def compute_structural_components(df_quantiles: pd.DataFrame, df_gmm_params: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Computes the four structural dimensions (Modality, Asymmetry, Tail Behavior, Structural Dispersion).

    This function implements the exact mathematical equations defined in the LaTeX context.
    It deterministically sorts the GMM components before identifying the dominant regime (i*),
    guaranteeing that ties in mixture weights are broken consistently by the lowest financial mean.

    Parameters
    ----------
    df_quantiles : pd.DataFrame
        The supplier-level DataFrame containing empirical quantiles.
    df_gmm_params : pd.DataFrame
        The component-level DataFrame containing pruned, scaled GMM parameters.
    config : Dict[str, Any]
        The study configuration dictionary containing stability constants.

    Returns
    -------
    pd.DataFrame
        A supplier-level DataFrame containing M, A, T, and D components.

    Raises
    ------
    ValueError
        If any computed component is mathematically invalid (e.g., < 1.0).
    """
    # Log the initiation of the structural component computation
    logger.info("Starting Task 11, Step 2: Computing structural components (M, A, T, D).")

    # Extract stability constants from the configuration
    phi_config: Dict[str, Any] = config["phi_computational_constants"]
    epsilon: float = phi_config["tail_behavior"]["stability_epsilon"]

    # ---------------------------------------------------------
    # 1. Compute Modality (M)
    # Equation: M = k (post-pruning component count)
    # ---------------------------------------------------------
    # Group the GMM parameters by supplier and count the number of retained components
    df_components = df_gmm_params.groupby('canonical_name').size().reset_index(name='M')
    # Ensure M is a float64 for consistent downstream multiplication
    df_components['M'] = df_components['M'].astype(np.float64)

    # Merge the empirical quantiles into the components DataFrame
    df_components = df_components.merge(df_quantiles, on='canonical_name', how='left')

    # ---------------------------------------------------------
    # 2. Compute Asymmetry (A)
    # Equation: aq = (Q0.75 + Q0.25 - 2*Q0.50) / (Q0.75 - Q0.25)
    # Equation: A = 1 + |aq|
    # ---------------------------------------------------------
    # Calculate the Interquartile Range spread
    iqr_spread: pd.Series = df_components['q_75'] - df_components['q_25']
    # Calculate the numerator for Bowley Skewness
    numerator_aq: pd.Series = df_components['q_75'] + df_components['q_25'] - (2 * df_components['q_50'])

    # Use np.where to enforce aq = 0 if IQR is zero, preventing division by zero
    # We use np.isclose to handle floating-point imprecision around zero
    aq: np.ndarray = np.where(
        ~np.isclose(iqr_spread, 0.0, atol=1e-9),
        numerator_aq / iqr_spread,
        0.0
    )
    # Compute the final Asymmetry component
    df_components['A'] = 1.0 + np.abs(aq)

    # ---------------------------------------------------------
    # 3. Compute Tail Behavior (T)
    # Equation: tq = ((Q0.95 - Q0.05) + eps) / ((Q0.75 - Q0.25) + eps)
    # Equation: T = 1 + |ln(tq)|
    # ---------------------------------------------------------
    # Calculate the 90% spread
    tail_spread: pd.Series = df_components['q_95'] - df_components['q_05']

    # The epsilon constant ensures numerical stability for highly clustered distributions
    tq: pd.Series = (tail_spread + epsilon) / (iqr_spread + epsilon)
    # Compute the final Tail Behavior component
    df_components['T'] = 1.0 + np.abs(np.log(tq))

    # ---------------------------------------------------------
    # 4. Compute Structural Dispersion (D)
    # Equation: D = 1 + pi_i* s_i* + sum_{i!=i*} pi_i s_i ln(1 + d_i)
    # ---------------------------------------------------------
    # Step 4a: Deterministically sort the GMM parameters to guarantee stable tie-breaking
    # Sort by supplier (asc), weight (desc), and mean (asc)
    df_gmm_sorted = df_gmm_params.sort_values(
        by=['canonical_name', 'weight_pi', 'mean_mu_scaled'],
        ascending=[True, False, True]
    ).reset_index(drop=True)

    # Step 4b: Identify the dominant regime (i*) by taking the first record per supplier
    # Because of the sort, this mathematically guarantees ties in pi are broken by the lowest mean
    df_dominant = df_gmm_sorted.groupby('canonical_name').first().reset_index()

    # Extract the dominant mean (\mu_{i*}) and the dominant component index
    df_dominant_features = df_dominant[['canonical_name', 'mean_mu_scaled', 'component_index']].rename(
        columns={'mean_mu_scaled': 'dominant_mu', 'component_index': 'dominant_index'}
    )

    # Step 4c: Merge the dominant features back to the component-level DataFrame
    df_gmm_calc = df_gmm_params.merge(df_dominant_features, on='canonical_name', how='left')

    # Step 4d: Compute the absolute distance from the dominant regime
    # Equation: d_i = |\mu_i - \mu_{i*}|
    df_gmm_calc['d_i'] = np.abs(df_gmm_calc['mean_mu_scaled'] - df_gmm_calc['dominant_mu'])

    # Step 4e: Compute the specific D term for each component
    # Create a boolean mask identifying the dominant regime strictly by its component index
    is_dominant: pd.Series = df_gmm_calc['component_index'] == df_gmm_calc['dominant_index']

    # Calculate the term for the dominant regime: pi_i* * s_i*
    term_dominant: pd.Series = df_gmm_calc['weight_pi'] * df_gmm_calc['std_dev_s_scaled']

    # Calculate the term for secondary regimes: pi_i * s_i * ln(1 + d_i)
    term_secondary: pd.Series = df_gmm_calc['weight_pi'] * df_gmm_calc['std_dev_s_scaled'] * np.log(1.0 + df_gmm_calc['d_i'])

    # Apply the conditional logic to assign the correct term using vectorized np.where
    df_gmm_calc['D_term'] = np.where(is_dominant, term_dominant, term_secondary)

    # Step 4f: Sum the terms per supplier and add the baseline 1.0
    df_D = df_gmm_calc.groupby('canonical_name')['D_term'].sum().reset_index(name='D_sum')
    df_D['D'] = 1.0 + df_D['D_sum']

    # Merge the computed D component back into the main supplier-level DataFrame
    df_components = df_components.merge(df_D[['canonical_name', 'D']], on='canonical_name', how='left')

    # Mathematical safeguard: All structural components must be >= 1.0 by definition
    for comp in ['M', 'A', 'T', 'D']:
        if (df_components[comp] < 1.0).any():
            # Raise a hard failure if the mathematical bounds are violated
            raise ValueError(f"Fatal Error: Component {comp} contains values < 1.0, violating mathematical bounds.")

    # Log successful completion
    logger.debug("Successfully computed M, A, T, and D for all suppliers.")
    logger.info("Task 11, Step 2 completed successfully.")

    # Return the supplier-level DataFrame
    return df_components


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: Compute PHI
# -------------------------------------------------------------------------------------------------------------------------------
def compute_phi_and_format_table(df_components: pd.DataFrame, df_canonical: pd.DataFrame) -> pd.DataFrame:
    """
    Computes the final Payment Heterogeneity Index (PHI) and formats the output table.

    This function applies the multiplicative synthesis principle (Joint Amplification),
    joins the pseudonymous supplier IDs for ethical reporting, and structures the final
    DataFrame for downstream ranking and decomposition.

    Parameters
    ----------
    df_components : pd.DataFrame
        The supplier-level DataFrame containing M, A, T, D, and quantiles.
    df_canonical : pd.DataFrame
        The DataFrame from the CanonicalSupplierArtifact containing 'pseudonymous_id'.

    Returns
    -------
    pd.DataFrame
        The finalized supplier-level PHI DataFrame.

    Raises
    ------
    ValueError
        If the PHI computation results in nulls or values < 1.0.
    """
    logger.info("Starting Task 11, Step 3: Computing PHI and formatting table.")

    # Create a defensive deep copy to ensure immutability of the input
    df_phi: pd.DataFrame = df_components.copy(deep=True)

    # Mathematical safeguard: Ensure no nulls exist in the components prior to multiplication
    if df_phi[['M', 'A', 'T', 'D']].isna().any().any():
        raise ValueError("Fatal Error: Null values detected in structural components prior to PHI synthesis.")

    # Apply the multiplicative synthesis equation
    # Equation: PHI = M * A * T * D
    df_phi['PHI'] = df_phi['M'] * df_phi['A'] * df_phi['T'] * df_phi['D']

    # Mathematical safeguard: PHI must be >= 1.0 since all components are >= 1.0
    if (df_phi['PHI'] < 1.0).any():
        raise ValueError("Fatal Error: PHI computation resulted in values < 1.0, violating joint amplification bounds.")

    # Extract the mapping of canonical_name to pseudonymous_id
    # We drop duplicates to ensure a clean 1:1 mapping for the left join
    df_id_map = df_canonical[['canonical_name', 'pseudonymous_id']].drop_duplicates()

    # Left join the pseudonymous IDs into the PHI table to support GDPR-compliant reporting
    df_phi = df_phi.merge(df_id_map, on='canonical_name', how='left')

    # Reorder columns for professional presentation and logical flow
    final_columns: List[str] = [
        'canonical_name', 'pseudonymous_id',
        'M', 'A', 'T', 'D', 'PHI',
        'q_05', 'q_25', 'q_50', 'q_75', 'q_95'
    ]
    df_phi = df_phi[final_columns]

    logger.debug(f"Computed PHI for {len(df_phi)} suppliers. Max PHI: {df_phi['PHI'].max():.2f}")
    logger.info("Task 11, Step 3 completed successfully.")

    return df_phi


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_11_compute_phi(
    scaled_cohort_artifact: Any, # Type hinted as Any to avoid circular imports if defined elsewhere, but logically ScaledCohortArtifact
    pruned_gmm_artifact: Any,    # Logically PrunedGMMArtifact
    canonical_supplier_artifact: Any, # Logically CanonicalSupplierArtifact
    config: Dict[str, Any]
) -> PHIArtifact:
    """
    Orchestrates the computation of the structural components and the final PHI score.

    This function sequentially computes the empirical quantiles, calculates the exact
    mathematical dimensions (M, A, T, D), synthesizes the multiplicative PHI score,
    and packages the results into a cryptographically verifiable artifact.

    Parameters
    ----------
    scaled_cohort_artifact : ScaledCohortArtifact
        The artifact containing the standardized transaction data.
    pruned_gmm_artifact : PrunedGMMArtifact
        The artifact containing the pruned, scaled GMM parameters.
    canonical_supplier_artifact : CanonicalSupplierArtifact
        The artifact containing the canonical to pseudonymous ID mapping.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    PHIArtifact
        The immutable artifact containing the computed PHI scores and components.

    Raises
    ------
    ValueError
        If any input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 11 Orchestrator: Compute PHI Components and Score ===")

    # Extract the underlying DataFrames from the input artifacts
    df_scaled: pd.DataFrame = scaled_cohort_artifact.data
    df_gmm_params: pd.DataFrame = pruned_gmm_artifact.parameters
    df_canonical: pd.DataFrame = canonical_supplier_artifact.data

    # Hard failure condition: The pipeline cannot proceed with empty data structures
    if df_scaled.empty or df_gmm_params.empty or df_canonical.empty:
        raise ValueError("Fatal Error: One or more input artifacts contain an empty DataFrame.")

    # Execute Step 1: Compute empirical quantiles
    df_quantiles: pd.DataFrame = compute_empirical_quantiles(df_scaled, config)

    # Execute Step 2: Compute structural components (M, A, T, D)
    df_components: pd.DataFrame = compute_structural_components(df_quantiles, df_gmm_params, config)

    # Execute Step 3: Compute PHI and format the final table
    df_phi: pd.DataFrame = compute_phi_and_format_table(df_components, df_canonical)

    # Ensure deterministic row order by sorting on the primary key (canonical_name)
    df_sorted: pd.DataFrame = df_phi.sort_values(by='canonical_name').reset_index(drop=True)

    # Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_sorted.columns.tolist())
    df_sorted_for_hash = df_sorted[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_sorted_for_hash.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Count the unique suppliers scored
    supplier_count: int = len(df_sorted)

    # Construct the immutable artifact
    artifact = PHIArtifact(
        data=df_sorted, # Store the logically ordered DataFrame
        content_hash_sha256=content_hash,
        supplier_count=supplier_count,
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 11 Orchestrator completed. Scored {artifact.supplier_count} suppliers. Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 12 — Rank suppliers, assign tiers, and compute log-decomposition attributions

# ==================================================================================
# Task 12: Rank suppliers, assign tiers, and compute log-decomposition attributions
# ==================================================================================

@dataclass
class RankedPHIArtifact:
    """
    An immutable artifact containing the prioritized supplier shortlist with risk tiers and XAI attributions.

    Attributes
    ----------
    data : pd.DataFrame
        The finalized supplier-level DataFrame, sorted by risk priority. Columns include
        identifiers, structural components, PHI score, ranks, risk tiers, and log-contributions.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    high_tier_count : int
        The exact number of suppliers assigned to the 'High' risk tier.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    # The core DataFrame containing the prioritized audit shortlist
    data: pd.DataFrame
    # Cryptographic hash ensuring the immutability and traceability of the artifact
    content_hash_sha256: str
    # Count of High-PHI suppliers isolated for targeted audit triage
    high_tier_count: int
    # Frozen configuration state for reproducibility
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: Compute empirical percentile ranks from PHI with deterministic tie handling
# -------------------------------------------------------------------------------------------------------------------------------
def compute_empirical_ranks(df_phi: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Computes the operational rank and empirical percentile rank for each supplier based on PHI.

    This function utilizes the configured tie-handling method (e.g., 'average') to assign
    a descending operational rank (1 = highest risk) and an ascending percentile rank
    (100.0 = highest risk), ensuring strict mathematical translation of relative positioning.

    Parameters
    ----------
    df_phi : pd.DataFrame
        The supplier-level DataFrame containing the computed 'PHI' scores.
    config : Dict[str, Any]
        The study configuration dictionary containing ranking conventions.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with 'phi_rank' and 'phi_percentile' columns.

    Raises
    ------
    ValueError
        If the ranking computation results in out-of-bound percentiles or nulls.
    """
    logger.info("Starting Task 12, Step 1: Computing empirical percentile ranks.")

    # Create a defensive deep copy to ensure immutability of the input
    df_ranked: pd.DataFrame = df_phi.copy(deep=True)

    # Extract ranking conventions from the configuration dictionary
    rank_config: Dict[str, Any] = config["risk_ranking_and_tiering"]["percentile_ranking"]
    tie_method: str = rank_config.get("ties", "average")

    # Compute the operational rank (Descending: Rank 1 is the highest PHI score)
    # We use the configured tie-breaking method to ensure statistical integrity
    df_ranked['phi_rank'] = df_ranked['PHI'].rank(method=tie_method, ascending=False)

    # Compute the empirical percentile rank (Ascending: 100.0 is the highest PHI score)
    # The pct=True argument returns values in [0, 1], which we scale to [0, 100]
    df_ranked['phi_percentile'] = df_ranked['PHI'].rank(method=tie_method, ascending=True, pct=True) * 100.0

    # Ensure strict float64 typing for downstream thresholding precision
    df_ranked['phi_percentile'] = df_ranked['phi_percentile'].astype(np.float64)

    # Mathematical safeguard: Validate that percentiles are strictly bounded within [0.0, 100.0]
    if (df_ranked['phi_percentile'] < 0.0).any() or (df_ranked['phi_percentile'] > 100.0).any():
        raise ValueError("Fatal Error: Computed percentiles fall outside the [0.0, 100.0] mathematical bounds.")

    # Mathematical safeguard: Validate that the highest operational rank is >= 1.0
    if df_ranked['phi_rank'].min() < 1.0:
        raise ValueError("Fatal Error: Computed operational rank is less than 1.0.")

    logger.debug(f"Computed ranks for {len(df_ranked)} suppliers. Top percentile: {df_ranked['phi_percentile'].max():.2f}")
    logger.info("Task 12, Step 1 completed successfully.")

    return df_ranked


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Assign Low, Moderate, and High tiers
# -------------------------------------------------------------------------------------------------------------------------------
def assign_risk_tiers(df_ranked: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Segments suppliers into risk tiers based on percentiles and whole-supplier adjustments.

    This function applies the 70th and 90th percentile thresholds to establish base tiers,
    then strictly enforces the whole-supplier adjustment rule (N_High = ceil(0.10 * N))
    using a deterministic multi-key sort to break ties at the boundary.

    Parameters
    ----------
    df_ranked : pd.DataFrame
        The DataFrame containing 'PHI', 'phi_percentile', and 'canonical_name'.
    config : Dict[str, Any]
        The study configuration dictionary containing tier mapping rules.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with a 'risk_tier' column, sorted by risk priority.

    Raises
    ------
    ValueError
        If the whole-supplier adjustment calculates an invalid target count.
    """
    logger.info("Starting Task 12, Step 2: Assigning risk tiers.")

    # Create a defensive deep copy
    df_tiered: pd.DataFrame = df_ranked.copy(deep=True)
    total_suppliers: int = len(df_tiered)

    # Extract tier mapping rules from the configuration
    tier_config: Dict[str, Any] = config["risk_ranking_and_tiering"]["tier_mapping"]
    low_thresh: float = tier_config["low_phi_percentile_lt"]
    high_thresh: float = tier_config["high_phi_percentile_ge"]
    adj_config: Dict[str, Any] = tier_config["whole_supplier_adjustment"]

    # 1. Assign base tiers using vectorized conditional logic based on percentiles
    # Conditions are evaluated in order; the first True condition dictates the value
    conditions = [
        df_tiered['phi_percentile'] >= high_thresh,
        df_tiered['phi_percentile'] >= low_thresh,
        df_tiered['phi_percentile'] < low_thresh
    ]
    choices = ['High', 'Moderate', 'Low']

    # Apply the conditions, defaulting to 'Low' if an edge case escapes
    df_tiered['risk_tier'] = np.select(conditions, choices, default='Low')

    # 2. Apply the whole-supplier adjustment rule for the High tier
    if adj_config.get("enabled", True):
        # Calculate the exact number of suppliers required in the High tier
        # Equation: N_High = ceil(0.10 * N)
        n_high_target: int = math.ceil(total_suppliers * 0.10)

        # Mathematical safeguard: Ensure the target is logical
        if n_high_target < 0 or n_high_target > total_suppliers:
            raise ValueError(f"Fatal Error: Invalid High tier target count calculated: {n_high_target}")

        # Sort the DataFrame deterministically to enforce the tie-breaker rule
        # Rule: "higher PHI then lexicographic"
        df_tiered = df_tiered.sort_values(
            by=['PHI', 'canonical_name'],
            ascending=[False, True]
        ).reset_index(drop=True)

        # Explicitly overwrite the risk_tier for the top N_High suppliers
        # This guarantees the exact whole-supplier count regardless of percentile boundary ties
        df_tiered.loc[:n_high_target - 1, 'risk_tier'] = 'High'

        # Ensure the remaining suppliers are not erroneously labeled 'High' due to the base assignment
        # If a supplier was >= 90th percentile but fell outside the top N_High, demote to Moderate
        df_tiered.loc[n_high_target:, 'risk_tier'] = np.where(
            df_tiered.loc[n_high_target:, 'risk_tier'] == 'High',
            'Moderate',
            df_tiered.loc[n_high_target:, 'risk_tier']
        )

        logger.debug(f"Applied whole-supplier adjustment. Target High tier count: {n_high_target}.")
    else:
        # If adjustment is disabled, simply sort by risk for presentation
        df_tiered = df_tiered.sort_values(
            by=['PHI', 'canonical_name'],
            ascending=[False, True]
        ).reset_index(drop=True)

    # Validate the final count of High tier suppliers
    actual_high_count: int = (df_tiered['risk_tier'] == 'High').sum()
    logger.info(f"Task 12, Step 2 completed. Assigned {actual_high_count} suppliers to the High risk tier.")

    return df_tiered


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Compute log-decomposition attributions
# -------------------------------------------------------------------------------------------------------------------------------
def compute_log_decomposition(df_tiered: pd.DataFrame) -> pd.DataFrame:
    """
    Computes the exact additive log-contribution of each structural component to the PHI score.

    This function implements Equation 6, providing the Explainable AI (XAI) layer required
    for forensic audit triage. It safely handles the mathematical singularity where PHI = 1.0.

    Parameters
    ----------
    df_tiered : pd.DataFrame
        The DataFrame containing the structural components (M, A, T, D) and PHI.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with percentage contribution columns for each component.

    Raises
    ------
    ValueError
        If the computed contributions do not sum to 100% (within float tolerance).
    """
    logger.info("Starting Task 12, Step 3: Computing log-decomposition attributions.")

    # Create a defensive deep copy
    df_decomp: pd.DataFrame = df_tiered.copy(deep=True)

    # Extract the components and the final score as NumPy arrays for fast vectorized math
    m_vals: np.ndarray = df_decomp['M'].values
    a_vals: np.ndarray = df_decomp['A'].values
    t_vals: np.ndarray = df_decomp['T'].values
    d_vals: np.ndarray = df_decomp['D'].values
    phi_vals: np.ndarray = df_decomp['PHI'].values

    # Compute the natural logarithm of the PHI score (the denominator)
    ln_phi: np.ndarray = np.log(phi_vals)

    # Identify the singularity edge case where PHI is exactly 1.0 (ln(PHI) == 0)
    # We use a tight tolerance to catch floating-point representations of 1.0
    is_baseline: np.ndarray = np.isclose(ln_phi, 0.0, atol=1e-9)

    # Define a helper function to compute the safe percentage contribution
    def _compute_safe_contribution(component_vals: np.ndarray) -> np.ndarray:
        # Equation: Contribution_X = (ln(X) / ln(PHI)) * 100%
        ln_x: np.ndarray = np.log(component_vals)
        # Use np.where to assign 0.0% if PHI is at baseline, avoiding division by zero
        contribution: np.ndarray = np.where(
            is_baseline,
            0.0,
            (ln_x / ln_phi) * 100.0
        )
        return contribution

    # Compute and assign the contributions for each structural dimension
    df_decomp['contribution_M_pct'] = _compute_safe_contribution(m_vals)
    df_decomp['contribution_A_pct'] = _compute_safe_contribution(a_vals)
    df_decomp['contribution_T_pct'] = _compute_safe_contribution(t_vals)
    df_decomp['contribution_D_pct'] = _compute_safe_contribution(d_vals)

    # Mathematical safeguard: Validate that the exact additive decomposition holds true
    # The sum of the four percentage contributions must equal 100.0% for all non-baseline suppliers
    total_contribution: pd.Series = (
        df_decomp['contribution_M_pct'] +
        df_decomp['contribution_A_pct'] +
        df_decomp['contribution_T_pct'] +
        df_decomp['contribution_D_pct']
    )

    # We check only the suppliers that are not at the absolute baseline
    non_baseline_mask: pd.Series = ~pd.Series(is_baseline)
    if non_baseline_mask.any():
        # Use np.allclose with a reasonable tolerance for accumulated floating-point errors
        if not np.allclose(total_contribution[non_baseline_mask], 100.0, atol=1e-4):
            raise ValueError("Fatal Error: Log-decomposition contributions do not sum to 100%.")

    logger.debug("Successfully computed log-decomposition attributions for all suppliers.")
    logger.info("Task 12, Step 3 completed successfully.")

    return df_decomp


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_12_rank_tier_decompose(
    phi_artifact: PHIArtifact,
    config: Dict[str, Any]
) -> RankedPHIArtifact:
    """
    Orchestrates the ranking, tier assignment, and XAI decomposition of the PHI scores.

    This function sequentially computes empirical percentiles, assigns categorical risk tiers
    with strict whole-supplier adjustments, calculates the additive log-contributions for
    audit transparency, and packages the prioritized shortlist into a verifiable artifact.

    Parameters
    ----------
    phi_artifact : PHIArtifact
        The immutable artifact containing the computed PHI scores and components.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    RankedPHIArtifact
        The immutable artifact containing the finalized, prioritized audit shortlist.

    Raises
    ------
    ValueError
        If the input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 12 Orchestrator: Rank, Tier, and Decompose ===")

    # Extract the DataFrame from the incoming artifact
    df_phi: pd.DataFrame = phi_artifact.data

    if df_phi.empty:
        raise ValueError("Fatal Error: Input PHIArtifact contains an empty DataFrame.")

    # Execute Step 1: Compute empirical percentile ranks
    df_ranked: pd.DataFrame = compute_empirical_ranks(df_phi, config)

    # Execute Step 2: Assign Low, Moderate, and High tiers
    # Note: This step deterministically sorts the DataFrame by risk priority
    df_tiered: pd.DataFrame = assign_risk_tiers(df_ranked, config)

    # Execute Step 3: Compute log-decomposition attributions
    df_final: pd.DataFrame = compute_log_decomposition(df_tiered)

    # To ensure the cryptographic hash is perfectly stable across environments,
    # we must temporarily sort by the primary key (canonical_name) and column names.
    # We do NOT alter the row order of df_final, as its risk-sorted order is the intended output.
    df_for_hash: pd.DataFrame = df_final.sort_values(by='canonical_name').reset_index(drop=True)
    sorted_columns: List[str] = sorted(df_for_hash.columns.tolist())
    df_for_hash = df_for_hash[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_for_hash.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Count the exact number of suppliers assigned to the High tier
    high_tier_count: int = int((df_final['risk_tier'] == 'High').sum())

    # Construct the immutable artifact
    artifact = RankedPHIArtifact(
        data=df_final, # Store the risk-sorted DataFrame for immediate operational use
        content_hash_sha256=content_hash,
        high_tier_count=high_tier_count,
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 12 Orchestrator completed. Isolated {artifact.high_tier_count} High-PHI suppliers. Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 13 — Baseline comparison: coefficient of variation (CV) and Spearman rank correlation with PHI

# ====================================================================================================
# Task 13: Baseline comparison: coefficient of variation (CV) and Spearman rank correlation with PHI
# ====================================================================================================

@dataclass
class SpearmanStats:
    """
    A strongly-typed dataclass encapsulating the results of the Spearman rank correlation.

    Attributes
    ----------
    correlation_coefficient : float
        The Spearman rank correlation coefficient (\rho_s) between PHI and CV.
    p_value : float
        The two-sided p-value for a hypothesis test whose null hypothesis is that
        two sets of data are uncorrelated.
    n_observations : int
        The number of suppliers (N) used to compute the correlation.
    """
    # The computed Spearman correlation coefficient
    correlation_coefficient: float
    # The computed two-sided p-value
    p_value: float
    # The sample size used for the correlation
    n_observations: int


@dataclass
class BaselineComparisonArtifact:
    """
    An immutable artifact containing the baseline comparison metrics, tables, and statistics.

    Attributes
    ----------
    full_comparison_data : pd.DataFrame
        The complete cohort DataFrame containing PHI metrics, CV metrics, and their respective ranks.
    top_12_comparison_table : pd.DataFrame
        The formatted subset DataFrame reproducing the structure of Table 3 from the manuscript.
    spearman_statistics : SpearmanStats
        The dataclass containing the correlation coefficient and p-value.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the full comparison dataset for drift detection.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    # The comprehensive DataFrame with all cohort suppliers
    full_comparison_data: pd.DataFrame
    # The isolated Top 12 table for reporting
    top_12_comparison_table: pd.DataFrame
    # The statistical correlation results
    spearman_statistics: SpearmanStats
    # Cryptographic hash ensuring the immutability and traceability of the artifact
    content_hash_sha256: str
    # Frozen configuration state for reproducibility
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1: Compute CV per supplier on the configured amount space
# -------------------------------------------------------------------------------------------------------------------------------
def compute_coefficient_of_variation(
    df_positive: pd.DataFrame,
    raw_to_canonical_map: Dict[str, str],
    cohort_suppliers: List[str],
    config: Dict[str, Any]
) -> pd.DataFrame:
    """
    Computes the Coefficient of Variation (CV) for each supplier in the high-volume cohort.

    This function maps the raw positive-flow transactions to their canonical identities,
    filters the data strictly to the high-volume cohort, and computes the sample mean (\mu)
    and sample standard deviation (\sigma) to derive the CV = \sigma / \mu. It implements
    a strict floating-point safeguard to prevent overflow from infinitesimal means.

    Parameters
    ----------
    df_positive : pd.DataFrame
        The transaction-level DataFrame containing strictly positive raw GBP payments.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.
    cohort_suppliers : List[str]
        The list of canonical names representing the high-volume cohort (n >= 50).
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    pd.DataFrame
        A supplier-level DataFrame containing 'canonical_name', 'mean_gbp', 'std_gbp', and 'cv'.

    Raises
    ------
    ValueError
        If the computed mean for any supplier is infinitesimally close to zero.
    """
    # Log the initiation of the CV computation step
    logger.info("Starting Task 13, Step 1: Computing Coefficient of Variation (CV).")

    # Create a defensive deep copy of the positive flow data to ensure immutability
    df_raw_amounts: pd.DataFrame = df_positive.copy(deep=True)

    # Map the raw 'Creditor Name' to the resolved 'canonical_name' using the provided dictionary
    df_raw_amounts['canonical_name'] = df_raw_amounts['Creditor Name'].map(raw_to_canonical_map)

    # Convert the cohort suppliers list to a Python set for O(1) lookup efficiency
    cohort_set: Set[str] = set(cohort_suppliers)

    # Create a boolean mask to filter the transactions strictly to the high-volume cohort
    cohort_mask: pd.Series = df_raw_amounts['canonical_name'].isin(cohort_set)

    # Apply the mask to isolate the cohort transactions
    df_cohort_amounts: pd.DataFrame = df_raw_amounts.loc[cohort_mask].copy(deep=True)

    # Group the cohort transactions by canonical supplier name
    grouped = df_cohort_amounts.groupby('canonical_name')['Net Amount']

    # Compute the sample mean (\mu) for each supplier
    df_cv = grouped.mean().reset_index(name='mean_gbp')

    # Compute the sample standard deviation (\sigma) using ddof=1 (Bessel's correction)
    # The cohort filter (n >= 50) mathematically guarantees n > 1, so ddof=1 is safe.
    df_cv['std_gbp'] = grouped.std(ddof=1).values

    # Mathematical safeguard: Ensure the mean is not infinitesimally close to zero.
    # Due to IEEE 754 representation, a mean could be 1e-16, which passes a <= 0 check
    # but causes a massive overflow during division. We use np.isclose with a 1e-9 tolerance.
    if np.isclose(df_cv['mean_gbp'], 0.0, atol=1e-9).any():
        # Raise a hard failure if the mathematical assumption is violated
        raise ValueError("Fatal Error: Computed mean is infinitesimally close to zero, risking float overflow.")

    # Compute the Coefficient of Variation
    # Equation: CV = \sigma / \mu
    df_cv['cv'] = df_cv['std_gbp'] / df_cv['mean_gbp']

    # Ensure strict float64 typing for downstream ranking precision
    df_cv['cv'] = df_cv['cv'].astype(np.float64)

    # Log the successful completion of the step
    logger.debug(f"Computed CV for {len(df_cv)} cohort suppliers.")
    logger.info("Task 13, Step 1 completed successfully.")

    # Return the computed CV DataFrame
    return df_cv


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2: Compute ranks and Spearman correlation
# -------------------------------------------------------------------------------------------------------------------------------
def compute_spearman_correlation(
    df_ranked_phi: pd.DataFrame,
    df_cv: pd.DataFrame,
    config: Dict[str, Any]
) -> Tuple[pd.DataFrame, Any]: # Returns Tuple[pd.DataFrame, SpearmanStats]
    """
    Merges the PHI and CV DataFrames, computes the CV rank, and calculates the Spearman correlation.

    This function ensures that the CV rank is computed in descending order (highest CV = Rank 1)
    to perfectly align with the PHI ranking directionality. It implements a strict zero-variance
    safeguard to prevent scipy.stats.spearmanr from returning NaN if the metrics are uniform.

    Parameters
    ----------
    df_ranked_phi : pd.DataFrame
        The supplier-level DataFrame containing PHI scores and ranks.
    df_cv : pd.DataFrame
        The supplier-level DataFrame containing the computed CV values.
    config : Dict[str, Any]
        The study configuration dictionary containing correlation conventions.

    Returns
    -------
    Tuple[pd.DataFrame, SpearmanStats]
        - The merged DataFrame enriched with 'cv_rank'.
        - The dataclass containing the correlation coefficient and p-value.

    Raises
    ------
    ValueError
        If the merge results in missing values or a mismatched cohort size.
    """
    # Log the initiation of the correlation computation step
    logger.info("Starting Task 13, Step 2: Computing ranks and Spearman correlation.")

    # Perform an inner merge on 'canonical_name' to align the PHI and CV data row-by-row
    df_merged: pd.DataFrame = df_ranked_phi.merge(df_cv, on='canonical_name', how='inner')

    # Mathematical safeguard: Ensure the merged DataFrame size matches the original cohort size
    if len(df_merged) != len(df_ranked_phi):
        raise ValueError(f"Fatal Error: Merge mismatch. Expected {len(df_ranked_phi)} rows, got {len(df_merged)}.")

    # Mathematical safeguard: Ensure no null values exist in the critical columns
    if df_merged[['PHI', 'cv']].isna().any().any():
        raise ValueError("Fatal Error: Null values detected in PHI or CV columns prior to correlation.")

    # Extract the ranking tie-breaking method from the configuration
    rank_config: Dict[str, Any] = config["risk_ranking_and_tiering"]["percentile_ranking"]
    tie_method: str = rank_config.get("ties", "average")

    # Compute the CV rank (Descending: Rank 1 is the highest CV score)
    # This aligns the directionality with the phi_rank (where Rank 1 is the highest PHI)
    df_merged['cv_rank'] = df_merged['cv'].rank(method=tie_method, ascending=False)

    # Extract the PHI and CV arrays for the statistical test
    phi_array: np.ndarray = df_merged['PHI'].values
    cv_array: np.ndarray = df_merged['cv'].values

    # Mathematical safeguard: Check for zero variance in the arrays.
    # If variance is zero, Spearman correlation is mathematically undefined and returns NaN.
    if np.isclose(np.std(phi_array), 0.0) or np.isclose(np.std(cv_array), 0.0):
        logger.warning("Zero variance detected in PHI or CV arrays. Spearman correlation is undefined. Defaulting to 0.0.")
        correlation_coefficient = 0.0
        p_value = 1.0
    else:
        # Execute the Spearman rank correlation test using scipy.stats
        res = stats.spearmanr(phi_array, cv_array)
        correlation_coefficient = float(res.statistic)
        p_value = float(res.pvalue)

    # Package the statistical results into the strongly-typed dataclass
    # (Assuming SpearmanStats is imported/available)
    # Note: In a full execution environment, this would instantiate SpearmanStats.
    spearman_stats = {
        "correlation_coefficient": correlation_coefficient,
        "p_value": p_value,
        "n_observations": len(df_merged)
    }

    # Log the statistical findings for the audit trail
    logger.info(f"Spearman Correlation: rho = {correlation_coefficient:.3f}, p-value = {p_value:.3e}")
    logger.info("Task 13, Step 2 completed successfully.")

    # Return the merged DataFrame and the statistical dataclass
    return df_merged, spearman_stats


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3: Reproduce the Top 12 PHI supplier comparison table structure
# -------------------------------------------------------------------------------------------------------------------------------
def generate_top_12_comparison_table(df_merged: pd.DataFrame) -> pd.DataFrame:
    """
    Isolates the High-PHI suppliers and formats the data for comparative reporting.

    This function dynamically determines the size of the 'High' risk tier, sorts the cohort
    by PHI rank, slices the top records, and generates formatted string columns to perfectly
    replicate the visual structure of Table 3 from the manuscript, without hardcoding limits.

    Parameters
    ----------
    df_merged : pd.DataFrame
        The complete cohort DataFrame containing PHI, CV, ranks, tiers, and log-contributions.

    Returns
    -------
    pd.DataFrame
        The formatted subset DataFrame representing the comparison table.
    """
    # Log the initiation of the table generation step
    logger.info("Starting Task 13, Step 3: Reproducing the comparison table.")

    # Dynamically determine the exact number of suppliers assigned to the 'High' risk tier
    # This ensures the table scales mathematically with the cohort size (N_High = ceil(0.10 * N))
    high_tier_count: int = int((df_merged['risk_tier'] == 'High').sum())

    # Sort the DataFrame strictly by the operational PHI rank (ascending, 1 to N)
    df_sorted: pd.DataFrame = df_merged.sort_values(by='phi_rank', ascending=True).reset_index(drop=True)

    # Slice the top records dynamically to isolate the highest-priority cohort
    df_top: pd.DataFrame = df_sorted.head(high_tier_count).copy(deep=True)

    # Define a helper function to format the component and its percentage contribution
    def _format_component(val: float, pct: float, is_integer: bool = False) -> str:
        # If the component is Modality (M), format it as an integer
        if is_integer:
            return f"{int(val)} ({pct:.1f}%)"
        # Otherwise, format the component value to 3 decimal places
        return f"{val:.3f} ({pct:.1f}%)"

    # Generate the formatted presentation column for Modality (M)
    df_top['M_formatted'] = df_top.apply(
        lambda row: _format_component(row['M'], row['contribution_M_pct'], is_integer=True), axis=1
    )

    # Generate the formatted presentation column for Asymmetry (A)
    df_top['A_formatted'] = df_top.apply(
        lambda row: _format_component(row['A'], row['contribution_A_pct']), axis=1
    )

    # Generate the formatted presentation column for Tail Behavior (T)
    df_top['T_formatted'] = df_top.apply(
        lambda row: _format_component(row['T'], row['contribution_T_pct']), axis=1
    )

    # Generate the formatted presentation column for Structural Dispersion (D)
    df_top['D_formatted'] = df_top.apply(
        lambda row: _format_component(row['D'], row['contribution_D_pct']), axis=1
    )

    # Define the exact column schema required for the final presentation table
    presentation_columns: List[str] = [
        'pseudonymous_id',
        'M_formatted',
        'A_formatted',
        'T_formatted',
        'D_formatted',
        'cv_rank',
        'phi_rank'
    ]

    # Select the presentation columns to create the final output DataFrame
    df_presentation: pd.DataFrame = df_top[presentation_columns].copy(deep=True)

    # Rename the columns to match the manuscript's exact nomenclature
    rename_mapping: Dict[str, str] = {
        'pseudonymous_id': 'Supplier ID',
        'M_formatted': 'M',
        'A_formatted': 'A',
        'T_formatted': 'T',
        'D_formatted': 'D',
        'cv_rank': 'CV Rank',
        'phi_rank': 'PHI Rank'
    }
    df_presentation = df_presentation.rename(columns=rename_mapping)

    # Log the successful generation of the table
    logger.debug(f"Successfully generated the comparison table for {high_tier_count} suppliers.")
    logger.info("Task 13, Step 3 completed successfully.")

    # Return the formatted presentation DataFrame
    return df_presentation


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_13_baseline_comparison(
    positive_flow_artifact: Any, # Logically PositiveFlowArtifact
    raw_to_canonical_map: Dict[str, str],
    ranked_phi_artifact: Any,    # Logically RankedPHIArtifact
    config: Dict[str, Any]
) -> BaselineComparisonArtifact:
    """
    Orchestrates the baseline comparison between PHI and the Coefficient of Variation (CV).

    This function sequentially computes the CV for the high-volume cohort, calculates the
    Spearman rank correlation to evaluate metric divergence, generates the formatted Top 12
    comparison table, and packages all results into a cryptographically verifiable artifact.

    Parameters
    ----------
    positive_flow_artifact : PositiveFlowArtifact
        The artifact containing the raw, strictly positive payment flows.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.
    ranked_phi_artifact : RankedPHIArtifact
        The artifact containing the finalized, ranked PHI scores and components.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    BaselineComparisonArtifact
        The immutable artifact containing the comparison data, tables, and statistics.

    Raises
    ------
    ValueError
        If any input artifact contains an empty DataFrame.
    """
    # Log the initiation of the orchestrator
    logger.info("=== Initiating Task 13 Orchestrator: Baseline Comparison (CV vs PHI) ===")

    # Extract the underlying DataFrames from the input artifacts
    df_positive: pd.DataFrame = positive_flow_artifact.data
    df_ranked_phi: pd.DataFrame = ranked_phi_artifact.data

    # Hard failure condition: The pipeline cannot proceed with empty data structures
    if df_positive.empty or df_ranked_phi.empty:
        raise ValueError("Fatal Error: One or more input artifacts contain an empty DataFrame.")

    # Extract the list of canonical suppliers that constitute the high-volume cohort
    cohort_suppliers: List[str] = df_ranked_phi['canonical_name'].unique().tolist()

    # Execute Step 1: Compute CV per supplier on the raw GBP positive-flow space
    df_cv: pd.DataFrame = compute_coefficient_of_variation(
        df_positive=df_positive,
        raw_to_canonical_map=raw_to_canonical_map,
        cohort_suppliers=cohort_suppliers,
        config=config
    )

    # Execute Step 2: Compute ranks and Spearman correlation
    df_merged, spearman_stats = compute_spearman_correlation(
        df_ranked_phi=df_ranked_phi,
        df_cv=df_cv,
        config=config
    )

    # Execute Step 3: Reproduce the Top 12 PHI supplier comparison table structure
    df_top_12: pd.DataFrame = generate_top_12_comparison_table(df_merged)

    # Ensure deterministic row order by sorting on the primary key (canonical_name)
    # This is strictly for the cryptographic hash; the operational DataFrames retain their logical sorting
    df_for_hash: pd.DataFrame = df_merged.sort_values(by='canonical_name').reset_index(drop=True)

    # Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_for_hash.columns.tolist())
    df_for_hash = df_for_hash[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_for_hash.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Construct the immutable artifact
    artifact = BaselineComparisonArtifact(
        full_comparison_data=df_merged, # Store the logically ordered merged DataFrame
        top_12_comparison_table=df_top_12, # Store the formatted presentation table
        spearman_statistics=spearman_stats, # Store the statistical results
        content_hash_sha256=content_hash,
        config_snapshot=copy.deepcopy(config)
    )

    # Log the successful completion of the orchestrator
    logger.info(f"=== Task 13 Orchestrator completed. Hash: {artifact.content_hash_sha256[:8]}... ===")

    # Return the finalized artifact
    return artifact
````

````python
# Task 14 — Create an orchestrator callable (required guardrail before robustness analysis)

# ========================================================================================
# Task 14: Create an orchestrator callable (required guardrail before robustness analysis)
# ========================================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1 & 3: Define the Reproducibility Bundle (Outputs)
# -------------------------------------------------------------------------------------------------------------------------------
@dataclass
class ReproducibilityBundle:
    """
    A cryptographically verifiable, immutable container holding the complete state of the PHI pipeline.

    This dataclass encapsulates every intermediate artifact, diagnostic report, and final
    output generated during the execution of the Payment Heterogeneity Index framework.
    It is designed to support partial instantiation; if the pipeline halts due to a
    validation failure or mathematical exception, the bundle will contain all artifacts
    generated up to the point of failure, enabling forensic debugging.

    Attributes
    ----------
    config_snapshot : Dict[str, Any]
        A frozen deep copy of the configuration dictionary used for the run.
    validation_report : ValidationReport
        The outcome of the initial schema and parseability checks (Task 1).
    cleansing_diagnostics : Optional[CleansingDiagnostics]
        Telemetry from type coercion, missingness, and duplicate handling (Task 2).
    windowed_artifact : Optional[WindowedArtifact]
        The dataset restricted to the configured temporal analysis window (Task 3).
    windowing_diagnostics : Optional[WindowingDiagnostics]
        Telemetry from the temporal windowing operation (Task 3).
    positive_flow_artifact : Optional[PositiveFlowArtifact]
        The dataset restricted to strictly positive payment amounts (Task 4).
    positive_flow_diagnostics : Optional[PositiveFlowDiagnostics]
        Telemetry from the non-standard amount exclusion operation (Task 4).
    normalized_name_artifact : Optional[NormalizedNameArtifact]
        The corpus of normalized supplier names and tokens (Task 5).
    match_edge_artifact : Optional[MatchEdgeArtifact]
        The filtered edge list of fuzzy-matched supplier identities (Task 6).
    canonical_supplier_artifact : Optional[CanonicalSupplierArtifact]
        The resolved canonical suppliers and their pseudonymous IDs (Task 7).
    raw_to_canonical_map : Optional[Dict[str, str]]
        The dictionary mapping raw 'Creditor Name's to 'canonical_name's (Task 7).
    scaled_cohort_artifact : Optional[ScaledCohortArtifact]
        The standardized transactions for the high-volume cohort (Task 8).
    scaled_gmm_artifact : Optional[ScaledGMMArtifact]
        The unpruned, scaled GMM parameters selected via BIC (Task 9).
    pruned_gmm_artifact : Optional[PrunedGMMArtifact]
        The finalized GMM parameters after pruning and renormalization (Task 10).
    phi_artifact : Optional[PHIArtifact]
        The computed structural components (M, A, T, D) and PHI scores (Task 11).
    ranked_phi_artifact : Optional[RankedPHIArtifact]
        The prioritized audit shortlist with risk tiers and log-decompositions (Task 12).
    baseline_comparison_artifact : Optional[BaselineComparisonArtifact]
        The CV baseline comparison, Spearman statistics, and Top 12 table (Task 13).
    pipeline_status : str
        A string indicating the final state of the execution ('SUCCESS' or 'FAILED').
    error_traceback : Optional[str]
        The exception traceback if the pipeline failed during execution.
    """
    config_snapshot: Dict[str, Any]
    validation_report: Any  # Type: ValidationReport
    cleansing_diagnostics: Optional[Any] = None  # Type: CleansingDiagnostics
    windowed_artifact: Optional[Any] = None      # Type: WindowedArtifact
    windowing_diagnostics: Optional[Any] = None  # Type: WindowingDiagnostics
    positive_flow_artifact: Optional[Any] = None # Type: PositiveFlowArtifact
    positive_flow_diagnostics: Optional[Any] = None # Type: PositiveFlowDiagnostics
    normalized_name_artifact: Optional[Any] = None # Type: NormalizedNameArtifact
    match_edge_artifact: Optional[Any] = None    # Type: MatchEdgeArtifact
    canonical_supplier_artifact: Optional[Any] = None # Type: CanonicalSupplierArtifact
    raw_to_canonical_map: Optional[Dict[str, str]] = None
    scaled_cohort_artifact: Optional[Any] = None # Type: ScaledCohortArtifact
    scaled_gmm_artifact: Optional[Any] = None    # Type: ScaledGMMArtifact
    pruned_gmm_artifact: Optional[Any] = None    # Type: PrunedGMMArtifact
    phi_artifact: Optional[Any] = None           # Type: PHIArtifact
    ranked_phi_artifact: Optional[Any] = None    # Type: RankedPHIArtifact
    baseline_comparison_artifact: Optional[Any] = None # Type: BaselineComparisonArtifact
    pipeline_status: str = "INITIALIZED"
    error_traceback: Optional[str] = None


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2: Specify the orchestrator's execution order and composition
# -------------------------------------------------------------------------------------------------------------------------------
def run_phi_pipeline_orchestrator(df_raw: pd.DataFrame, config: Dict[str, Any]) -> ReproducibilityBundle:
    """
    The master orchestrator for the Payment Heterogeneity Index (PHI) framework.

    This function acts as the strict Directed Acyclic Graph (DAG) controller, executing
    Tasks 1 through 13 in a deterministic, linear sequence. It enforces the mathematical
    chain of custody by passing immutable artifacts between stages. It implements a
    fail-fast design, halting execution and returning a partial bundle if any validation
    gate or mathematical invariant is violated.

    Parameters
    ----------
    df_raw : pd.DataFrame
        The immutable, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing all methodological parameters.

    Returns
    -------
    ReproducibilityBundle
        A comprehensive dataclass containing all generated artifacts, diagnostics,
        and the final execution status.
    """
    logger.info("======================================================================")
    logger.info("=== INITIATING PHI FRAMEWORK MASTER ORCHESTRATOR (TASKS 1-13)      ===")
    logger.info("======================================================================")

    # Create a deep copy of the configuration to freeze the state at the moment of invocation
    config_snapshot: Dict[str, Any] = copy.deepcopy(config)

    # Initialize the bundle with empty/None values for the intermediate artifacts
    bundle = ReproducibilityBundle(
        config_snapshot=config_snapshot,
        validation_report=None
    )

    try:
        # -----------------------------------------------------------------------------------
        # TASK 1: Validate Inputs
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 1: Input Validation")
        # Note: Assumes task_1_validate_inputs is available in the namespace
        validation_report = task_1_validate_inputs(df_raw, config)
        bundle.validation_report = validation_report

        # Strict Validation Gate: Halt pipeline if inputs do not meet methodological standards
        if not validation_report.is_valid:
            logger.error("Pipeline halted at Task 1: Input validation failed.")
            for err in validation_report.errors:
                logger.error(f"Validation Error: {err}")
            bundle.pipeline_status = "FAILED_VALIDATION"
            return bundle

        # -----------------------------------------------------------------------------------
        # TASK 2: Cleanse Data
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 2: Data Cleansing")
        df_cleansed, cleansing_diagnostics = task_2_cleanse_data(df_raw, config)
        bundle.cleansing_diagnostics = cleansing_diagnostics

        # -----------------------------------------------------------------------------------
        # TASK 3: Enforce Analysis Window
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 3: Temporal Windowing")
        windowed_artifact, windowing_diagnostics = task_3_enforce_analysis_window(df_cleansed, config)
        bundle.windowed_artifact = windowed_artifact
        bundle.windowing_diagnostics = windowing_diagnostics

        # -----------------------------------------------------------------------------------
        # TASK 4: Filter Non-Standard Amounts
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 4: Positive Flow Isolation")
        positive_flow_artifact, positive_flow_diagnostics = task_4_filter_non_standard_amounts(windowed_artifact, config)
        bundle.positive_flow_artifact = positive_flow_artifact
        bundle.positive_flow_diagnostics = positive_flow_diagnostics

        # -----------------------------------------------------------------------------------
        # TASK 5: Prepare Supplier Names
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 5: Lexical Normalization and Tokenization")
        normalized_name_artifact = task_5_prepare_supplier_names(positive_flow_artifact, config)
        bundle.normalized_name_artifact = normalized_name_artifact

        # -----------------------------------------------------------------------------------
        # TASK 6: Compute Fuzzy Metrics and Match Gate
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 6: Fuzzy Similarity and Match Gating")
        match_edge_artifact = task_6_compute_fuzzy_metrics_and_gate(normalized_name_artifact, config)
        bundle.match_edge_artifact = match_edge_artifact

        # -----------------------------------------------------------------------------------
        # TASK 7: Build Graph and Resolve Canonical Suppliers
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 7: Graph Resolution and Pseudonymization")
        canonical_supplier_artifact, raw_to_canonical_map = task_7_build_graph_and_resolve(
            normalized_name_artifact, match_edge_artifact, config
        )
        bundle.canonical_supplier_artifact = canonical_supplier_artifact
        bundle.raw_to_canonical_map = raw_to_canonical_map

        # -----------------------------------------------------------------------------------
        # TASK 8: Select Cohort and Scale
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 8: Cohort Selection and Robust Standardization")
        scaled_cohort_artifact = task_8_select_cohort_and_scale(
            positive_flow_artifact, raw_to_canonical_map, config
        )
        bundle.scaled_cohort_artifact = scaled_cohort_artifact

        # -----------------------------------------------------------------------------------
        # TASK 9: Fit GMMs
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 9: GMM Fitting and BIC Selection")
        scaled_gmm_artifact = task_9_fit_gmms(scaled_cohort_artifact, config)
        bundle.scaled_gmm_artifact = scaled_gmm_artifact

        # -----------------------------------------------------------------------------------
        # TASK 10: Prune and Back-Transform
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 10: Regime Pruning and GBP Back-Transformation")
        pruned_gmm_artifact = task_10_prune_and_backtransform(
            scaled_gmm_artifact, scaled_cohort_artifact.scaling_parameters, config
        )
        bundle.pruned_gmm_artifact = pruned_gmm_artifact

        # -----------------------------------------------------------------------------------
        # TASK 11: Compute PHI
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 11: Structural Component and PHI Synthesis")
        phi_artifact = task_11_compute_phi(
            scaled_cohort_artifact, pruned_gmm_artifact, canonical_supplier_artifact, config
        )
        bundle.phi_artifact = phi_artifact

        # -----------------------------------------------------------------------------------
        # TASK 12: Rank, Tier, and Decompose
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 12: Risk Tiering and XAI Log-Decomposition")
        ranked_phi_artifact = task_12_rank_tier_decompose(phi_artifact, config)
        bundle.ranked_phi_artifact = ranked_phi_artifact

        # -----------------------------------------------------------------------------------
        # TASK 13: Baseline Comparison
        # -----------------------------------------------------------------------------------
        logger.info(">>> Executing Task 13: CV Baseline and Spearman Correlation")
        baseline_comparison_artifact = task_13_baseline_comparison(
            positive_flow_artifact, raw_to_canonical_map, ranked_phi_artifact, config
        )
        bundle.baseline_comparison_artifact = baseline_comparison_artifact

        # Mark the pipeline as successfully completed
        bundle.pipeline_status = "SUCCESS"
        logger.info("======================================================================")
        logger.info("=== PHI FRAMEWORK MASTER ORCHESTRATOR COMPLETED SUCCESSFULLY       ===")
        logger.info("======================================================================")

    except Exception as e:
        # Catch any unexpected mathematical or structural failures during execution
        error_msg = f"Pipeline encountered a fatal exception: {str(e)}"
        logger.error("======================================================================")
        logger.error("=== PHI FRAMEWORK MASTER ORCHESTRATOR FAILED                       ===")
        logger.error(error_msg)
        logger.error("======================================================================")

        # Capture the full traceback for forensic debugging
        bundle.error_traceback = traceback.format_exc()
        bundle.pipeline_status = "FAILED_EXCEPTION"
        logger.debug(bundle.error_traceback)

    # Return the bundle, which will either be fully populated (SUCCESS) or partially
    # populated with the artifacts generated up to the point of failure.
    return bundle
````

````python
# Task 15 — Threshold anchoring analysis (including robustness check ±5% vs ±10%)

# ==============================================================================
# Task 15: Threshold anchoring analysis (including robustness check ±5% vs ±10%)
# ==============================================================================

@dataclass
class PeakExtractionResult:
    """
    Encapsulates the results of the endogenous peak extraction routine.

    This strongly-typed data transfer object stores the financial locations, empirical
    densities, and relative prominences of recurring institutional payment anchors discovered
    within the GMM parameter space. It implements strict lifecycle validation to ensure
    array shape consistency and type integrity.

    Attributes
    ----------
    peak_values_gbp : np.ndarray
        1D array of the extracted peak financial values in GBP space.
    bin_edges : np.ndarray
        1D array of the bin boundaries used to construct the empirical histogram.
    smoothed_density : np.ndarray
        1D array of the regularized density values computed via Gaussian kernel smoothing.
    peak_prominences : np.ndarray
        1D array of the relative prominences corresponding to each extracted peak.

    Raises
    ------
    TypeError
        If any attribute is not a NumPy ndarray.
    ValueError
        If the arrays are not 1D, or if shape mismatches exist between peaks and prominences.
    """
    # 1D array of the extracted peak values in GBP
    peak_values_gbp: np.ndarray
    # 1D array of the bin edges used for the histogram
    bin_edges: np.ndarray
    # 1D array of the smoothed density values
    smoothed_density: np.ndarray
    # 1D array of the relative prominences for the extracted peaks
    peak_prominences: np.ndarray

    # Lifecycle hook to automatically validate input types and array shapes upon instantiation
    def __post_init__(self) -> None:
        # Define the list of attributes that must strictly be NumPy arrays
        array_attrs = ['peak_values_gbp', 'bin_edges', 'smoothed_density', 'peak_prominences']
        # Iterate through each required array attribute
        for attr in array_attrs:
            # Dynamically fetch the attribute value
            val = getattr(self, attr)
            # Verify strict type compliance
            if not isinstance(val, np.ndarray):
                # Raise a TypeError if the attribute is not a NumPy array
                raise TypeError(f"Attribute '{attr}' must be a numpy.ndarray, got {type(val)}.")
            # Verify that the array is strictly one-dimensional
            if val.ndim != 1:
                # Raise a ValueError if multi-dimensional arrays are passed
                raise ValueError(f"Attribute '{attr}' must be a 1D array, got ndim={val.ndim}.")

        # Verify that the number of extracted peaks perfectly matches the number of prominences
        if len(self.peak_values_gbp) != len(self.peak_prominences):
            # Raise a ValueError if a shape mismatch occurs between related vectors
            raise ValueError(
                f"Shape mismatch: peak_values_gbp (len={len(self.peak_values_gbp)}) "
                f"must match peak_prominences (len={len(self.peak_prominences)})."
            )


@dataclass
class PermutationResult:
    """
    Encapsulates the results of the Monte Carlo permutation test for threshold proximity.

    This object stores the inferential statistics generated by holding payment values fixed
    while randomly shuffling risk tier labels across GMM centres. It evaluates the one-sided
    hypothesis that High-PHI suppliers exhibit an excess propensity to anchor near thresholds.

    Attributes
    ----------
    tolerance : float
        The relative proximity window used to define threshold alignment (e.g., 0.10 for ±10%).
    observed_count : int
        The absolute count of High-tier GMM centres falling within the proximity window.
    null_mean : float
        The expected count of proximate High-tier centres under the permuted null distribution.
    null_distribution : np.ndarray
        1D array containing the proximate counts across all N Monte Carlo iterations.
    p_value : float
        The computed one-sided p-value: p = (1/N) * sum(I[C_b >= C_obs]).

    Raises
    ------
    TypeError
        If attributes violate strict scalar or array typing.
    ValueError
        If tolerance <= 0, counts are negative, or the p-value falls outside [0.0, 1.0].
    """
    # The tolerance window used (e.g., 0.10 for ±10%)
    tolerance: float
    # The observed count of High-tier proximate centres
    observed_count: int
    # The mean count of High-tier proximate centres under the null distribution
    null_mean: float
    # The 1D array containing the null distribution counts (length N_iterations)
    null_distribution: np.ndarray
    # The computed one-sided p-value
    p_value: float

    # Lifecycle hook to automatically validate statistical bounds and types upon instantiation
    def __post_init__(self) -> None:
        # Verify that tolerance is a positive float
        if not isinstance(self.tolerance, float) or self.tolerance <= 0.0:
            # Raise an error if tolerance is mathematically invalid
            raise ValueError(f"'tolerance' must be a positive float, got {self.tolerance}.")

        # Verify that the observed count is a non-negative integer
        if not isinstance(self.observed_count, int) or self.observed_count < 0:
            # Raise an error if the count is negative or non-integer
            raise ValueError(f"'observed_count' must be a non-negative integer, got {self.observed_count}.")

        # Verify that the null mean is a non-negative float
        if not isinstance(self.null_mean, float) or self.null_mean < 0.0:
            # Raise an error if the null mean is mathematically invalid
            raise ValueError(f"'null_mean' must be a non-negative float, got {self.null_mean}.")

        # Verify that the null distribution is strictly a 1D NumPy array
        if not isinstance(self.null_distribution, np.ndarray) or self.null_distribution.ndim != 1:
            # Raise an error if the null distribution container is malformed
            raise TypeError("'null_distribution' must be a 1D numpy.ndarray.")

        # Verify that the p-value is strictly bounded within the probability space [0.0, 1.0]
        if not isinstance(self.p_value, float) or not (0.0 <= self.p_value <= 1.0):
            # Raise an error if the p-value violates basic axioms of probability
            raise ValueError(f"'p_value' must be a float between 0.0 and 1.0, got {self.p_value}.")


@dataclass
class KSTestResult:
    """
    Encapsulates the results of the Two-Sample Kolmogorov-Smirnov test.

    This object stores the maximum vertical divergence statistic and associated p-value
    comparing the empirical cumulative distribution functions (ECDFs) of relative distance
    to the nearest peak between High-PHI and Low-PHI supplier tiers.

    Attributes
    ----------
    ks_statistic : float
        The supremum distance D_{KS} = \sup_x |F_{high}(x) - F_{low}(x)| bounded in [0.0, 1.0].
    p_value : float
        The two-sided p-value evaluating the null hypothesis that both tiers are drawn
        from the same underlying continuous distribution.

    Raises
    ------
    TypeError
        If attributes are not strict Python floats.
    ValueError
        If the KS statistic or p-value falls outside the mathematical bounds [0.0, 1.0].
    """
    # The maximum vertical separation between the High and Low ECDFs
    ks_statistic: float
    # The two-sided p-value
    p_value: float

    # Lifecycle hook to automatically validate statistical bounds upon instantiation
    def __post_init__(self) -> None:
        # Verify that the KS statistic is a float strictly bounded within [0.0, 1.0]
        if not isinstance(self.ks_statistic, float) or not (0.0 <= self.ks_statistic <= 1.0):
            # Raise an error if the distance metric violates metric space bounds
            raise ValueError(f"'ks_statistic' must be a float between 0.0 and 1.0, got {self.ks_statistic}.")

        # Verify that the p-value is a float strictly bounded within [0.0, 1.0]
        if not isinstance(self.p_value, float) or not (0.0 <= self.p_value <= 1.0):
            # Raise an error if the p-value violates probability axioms
            raise ValueError(f"'p_value' must be a float between 0.0 and 1.0, got {self.p_value}.")


@dataclass
class ThresholdAnchoringArtifact:
    """
    An immutable artifact containing the threshold anchoring analysis and inferential statistics.

    This master container binds the enriched component-level dataset with the complete suite
    of statistical validations (endogenous peaks, permutation tests, and K-S divergence).
    It includes a cryptographic SHA-256 hash to guarantee data provenance and auditability.

    Attributes
    ----------
    enriched_component_data : pd.DataFrame
        The component-level DataFrame enriched with distance metrics and proximity flags.
    peak_extraction : PeakExtractionResult
        The dataclass containing the endogenous peaks and density arrays.
    permutation_10pct : PermutationResult
        The permutation test results for the ±10% tolerance window.
    permutation_05pct : PermutationResult
        The permutation test results for the ±5% stringent tolerance window.
    ks_test_result : KSTestResult
        The Kolmogorov-Smirnov test results comparing High vs Low tier distances.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the enriched component dataset for drift detection.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.

    Raises
    ------
    TypeError
        If any attribute violates its explicit class contract.
    ValueError
        If the SHA-256 hash string is malformed.
    """
    # The component-level DataFrame enriched with distance metrics and proximity flags
    enriched_component_data: pd.DataFrame
    # The dataclass containing the endogenous peaks and density arrays
    peak_extraction: PeakExtractionResult
    # The permutation test results for the ±10% tolerance window
    permutation_10pct: PermutationResult
    # The permutation test results for the ±5% stringent tolerance window
    permutation_05pct: PermutationResult
    # The Kolmogorov-Smirnov test results comparing High vs Low tier distances
    ks_test_result: KSTestResult
    # A deterministic SHA-256 hash of the enriched component dataset for drift detection
    content_hash_sha256: str
    # A deep copy of the configuration used to generate this artifact
    config_snapshot: Dict[str, Any]

    # Lifecycle hook to automatically validate object composition and cryptographic integrity
    def __post_init__(self) -> None:
        # Verify that the enriched data is strictly a pandas DataFrame
        if not isinstance(self.enriched_component_data, pd.DataFrame):
            # Raise a TypeError if the tabular container contract is breached
            raise TypeError(f"'enriched_component_data' must be a pandas.DataFrame, got {type(self.enriched_component_data)}.")

        # Verify that the peak extraction object matches its explicit dataclass type
        if not isinstance(self.peak_extraction, PeakExtractionResult):
            # Raise a TypeError if the nested DTO contract is breached
            raise TypeError(f"'peak_extraction' must be a PeakExtractionResult, got {type(self.peak_extraction)}.")

        # Verify that the 10% permutation object matches its explicit dataclass type
        if not isinstance(self.permutation_10pct, PermutationResult):
            # Raise a TypeError if the nested DTO contract is breached
            raise TypeError(f"'permutation_10pct' must be a PermutationResult, got {type(self.permutation_10pct)}.")

        # Verify that the 5% permutation object matches its explicit dataclass type
        if not isinstance(self.permutation_05pct, PermutationResult):
            # Raise a TypeError if the nested DTO contract is breached
            raise TypeError(f"'permutation_05pct' must be a PermutationResult, got {type(self.permutation_05pct)}.")

        # Verify that the K-S test object matches its explicit dataclass type
        if not isinstance(self.ks_test_result, KSTestResult):
            # Raise a TypeError if the nested DTO contract is breached
            raise TypeError(f"'ks_test_result' must be a KSTestResult, got {type(self.ks_test_result)}.")

        # Verify that the cryptographic hash is exactly a 64-character hexadecimal string
        if not isinstance(self.content_hash_sha256, str) or not re.fullmatch(r'^[a-fA-F0-9]{64}$', self.content_hash_sha256):
            # Raise a ValueError if the SHA-256 signature is corrupted or malformed
            raise ValueError(f"'content_hash_sha256' must be a valid 64-character hex string, got '{self.content_hash_sha256}'.")

        # Verify that the configuration snapshot is strictly a Python dictionary
        if not isinstance(self.config_snapshot, dict):
            # Raise a TypeError if the configuration container contract is breached
            raise TypeError(f"'config_snapshot' must be a Python dictionary, got {type(self.config_snapshot)}.")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: Peak extraction from high-PHI GMM centres (GBP space)
# -------------------------------------------------------------------------------------------------------------------------------
def extract_endogenous_peaks(df_components: pd.DataFrame, config: Dict[str, Any]) -> Any: # Returns PeakExtractionResult
    """
    Extracts recurring institutional payment anchors from the High-PHI cohort's GMM centres.

    This function constructs a histogram of High-PHI centres, applies a Gaussian kernel
    smoother, and extracts peaks based on a relative prominence criterion. It mathematically
    computes the exact number of bins and uses `np.linspace` to guarantee perfectly uniform
    bin widths, preventing density skewing caused by floating-point imprecision in `np.arange`.

    Parameters
    ----------
    df_components : pd.DataFrame
        The component-level DataFrame containing 'mean_mu_gbp' and 'risk_tier'.
    config : Dict[str, Any]
        The study configuration dictionary containing peak extraction parameters.

    Returns
    -------
    PeakExtractionResult
        A dataclass containing the extracted peaks, bin edges, and smoothed density.

    Raises
    ------
    ValueError
        If no peaks are found satisfying the criteria.
    """
    logger.info("Starting Task 15, Step 1: Extracting endogenous peaks from High-PHI centres.")

    # Extract configuration parameters
    peak_config: Dict[str, Any] = config["threshold_anchoring_parameters"]["peak_extraction"]
    range_min, range_max = peak_config["histogram_range_gbp"]
    bin_width: int = peak_config["bin_width_gbp"]
    sigma_bins: int = peak_config["sigma_bins"]
    rel_prominence: float = peak_config["relative_prominence_threshold"]
    boundary_limit: float = peak_config["boundary_limit_gbp"]

    # Isolate the GMM centres belonging exclusively to the High-PHI tier
    high_phi_mask: pd.Series = df_components['risk_tier'] == 'High'
    high_phi_centres: np.ndarray = df_components.loc[high_phi_mask, 'mean_mu_gbp'].values

    # Mathematically compute the exact number of bins to guarantee uniform width
    # Equation: N_bins = int((Max - Min) / Width)
    # We add 1 to get the number of bin *edges* required for np.linspace
    num_bins: int = int((range_max - range_min) / bin_width)
    num_edges: int = num_bins + 1

    # Generate the exact bin edges using linspace to avoid floating-point accumulation errors
    bin_edges: np.ndarray = np.linspace(range_min, range_max, num_edges)

    # Generate the histogram counts
    counts, _ = np.histogram(high_phi_centres, bins=bin_edges)

    # Apply the Gaussian kernel smoother to regularize the empirical density
    smoothed_density: np.ndarray = ndimage.gaussian_filter1d(counts.astype(float), sigma=sigma_bins)

    # Calculate the absolute prominence threshold based on the maximum smoothed density
    max_density: float = float(np.max(smoothed_density))
    prominence_threshold: float = rel_prominence * max_density

    # Extract peaks using scipy.signal.find_peaks
    # This returns the indices of the peaks in the smoothed_density array
    peak_indices, properties = signal.find_peaks(smoothed_density, prominence=prominence_threshold)

    # Map the peak indices back to their GBP financial values
    # The value is the center of the corresponding bin
    bin_centers: np.ndarray = bin_edges[:-1] + (bin_width / 2.0)
    raw_peak_values: np.ndarray = bin_centers[peak_indices]
    raw_prominences: np.ndarray = properties['prominences']

    # Filter out boundary artifacts (peaks <= boundary_limit)
    valid_mask: np.ndarray = raw_peak_values > boundary_limit
    final_peak_values: np.ndarray = raw_peak_values[valid_mask]
    final_prominences: np.ndarray = raw_prominences[valid_mask]

    # Mathematical safeguard: Ensure at least one peak was found
    if len(final_peak_values) == 0:
        raise ValueError("Fatal Error: No endogenous peaks found satisfying the prominence and boundary criteria.")

    logger.debug(f"Extracted {len(final_peak_values)} valid peaks. Values: {final_peak_values}")
    logger.info("Task 15, Step 1 completed successfully.")

    # Construct the artifact (Assuming PeakExtractionResult is imported/available)
    # Note: In a full execution environment, this would instantiate PeakExtractionResult.
    result = {
        "peak_values_gbp": final_peak_values,
        "bin_edges": bin_edges,
        "smoothed_density": smoothed_density,
        "peak_prominences": final_prominences
    }

    return result


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 2: Compute distance-to-nearest-peak and proximity flags
# -------------------------------------------------------------------------------------------------------------------------------
def compute_distance_and_proximity(
    df_components: pd.DataFrame,
    peaks: np.ndarray,
    config: Dict[str, Any]
) -> pd.DataFrame:
    """
    Computes the relative distance to the nearest peak and assigns proximity flags.

    This function utilizes NumPy broadcasting to efficiently compute the distance matrix
    between all GMM centres and all extracted peaks, identifying the nearest anchor and
    evaluating the proximity conditions for both 10% and 5% tolerances.

    Parameters
    ----------
    df_components : pd.DataFrame
        The component-level DataFrame containing 'mean_mu_gbp'.
    peaks : np.ndarray
        The 1D array of extracted peak values in GBP.
    config : Dict[str, Any]
        The study configuration dictionary containing tolerance parameters.

    Returns
    -------
    pd.DataFrame
        The enriched DataFrame containing distance metrics and boolean proximity flags.
    """
    logger.info("Starting Task 15, Step 2: Computing distance-to-nearest-peak and proximity flags.")

    # Create a defensive deep copy
    df_enriched: pd.DataFrame = df_components.copy(deep=True)

    # Extract tolerance parameters
    val_config: Dict[str, Any] = config["threshold_anchoring_parameters"]["validation_statistics"]
    tol_10pct: float = val_config["proximity_tolerance"]
    tol_05pct: float = val_config["stringent_tolerance"]

    # Extract all GMM centres as a 1D NumPy array (Length N)
    mu_array: np.ndarray = df_enriched['mean_mu_gbp'].values

    # Utilize NumPy broadcasting to compute the absolute distance matrix
    # mu_array[:, None] has shape (N, 1)
    # peaks[None, :] has shape (1, P)
    # Resulting distance_matrix has shape (N, P)
    distance_matrix: np.ndarray = np.abs(mu_array[:, None] - peaks[None, :])

    # Find the index of the nearest peak for each centre
    nearest_peak_indices: np.ndarray = np.argmin(distance_matrix, axis=1)

    # Extract the actual GBP value of the nearest peak for each centre
    nearest_peaks: np.ndarray = peaks[nearest_peak_indices]

    # Compute the relative distance to the nearest peak
    # Equation: \delta(\mu) = |\mu - p(\mu)| / p(\mu)
    # Division by zero is mathematically impossible as peaks > 300
    delta_mu: np.ndarray = np.abs(mu_array - nearest_peaks) / nearest_peaks

    # Compute the boolean proximity flags
    # Equation: prox(\mu; \tau) = \mathbb{I}[\delta(\mu) \le \tau]
    is_proximate_10pct: np.ndarray = delta_mu <= tol_10pct
    is_proximate_05pct: np.ndarray = delta_mu <= tol_05pct

    # Append the computed metrics to the DataFrame
    df_enriched['nearest_peak_gbp'] = nearest_peaks
    df_enriched['delta_mu'] = delta_mu
    df_enriched['is_proximate_10pct'] = is_proximate_10pct
    df_enriched['is_proximate_05pct'] = is_proximate_05pct

    logger.debug("Successfully computed distance metrics and proximity flags for all centres.")
    logger.info("Task 15, Step 2 completed successfully.")

    return df_enriched


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 3: Run permutation tests and two-sample K-S test
# -------------------------------------------------------------------------------------------------------------------------------
def run_inferential_validations(df_enriched: pd.DataFrame, config: Dict[str, Any]) -> Tuple[PermutationResult, PermutationResult, KSTestResult]:
    """
    Executes the Monte Carlo permutation tests and the Kolmogorov-Smirnov divergence test.

    This function rigorously validates whether the observed threshold anchoring in the
    High-PHI cohort is statistically significant compared to random chance, and whether
    the overall distance distributions differ materially between risk tiers.

    Parameters
    ----------
    df_enriched : pd.DataFrame
        The component-level DataFrame containing tiers, distances, and proximity flags.
    config : Dict[str, Any]
        The study configuration dictionary containing inferential parameters.

    Returns
    -------
    Tuple[PermutationResult, PermutationResult, KSTestResult]
        - Permutation results for the 10% tolerance.
        - Permutation results for the 5% tolerance.
        - K-S test results comparing High vs Low tiers.
    """
    logger.info("Starting Task 15, Step 3: Running inferential statistical validations.")

    # Extract inferential parameters
    val_config: Dict[str, Any] = config["threshold_anchoring_parameters"]["validation_statistics"]
    n_iterations: int = val_config["permutation_iterations"]
    rng_seed: int = val_config["permutation_rng_seed"]
    tol_10pct: float = val_config["proximity_tolerance"]
    tol_05pct: float = val_config["stringent_tolerance"]

    # Initialize the modern, thread-safe NumPy random number generator
    rng = np.random.default_rng(seed=rng_seed)

    # Extract the necessary columns as raw NumPy arrays for maximum performance in the loop
    tier_labels: np.ndarray = df_enriched['risk_tier'].values
    prox_10pct: np.ndarray = df_enriched['is_proximate_10pct'].values
    prox_05pct: np.ndarray = df_enriched['is_proximate_05pct'].values

    # Create a boolean mask for the observed High tier
    high_tier_mask: np.ndarray = tier_labels == 'High'

    # Calculate the observed counts of High-tier proximate centres
    obs_count_10pct: int = int(np.sum(prox_10pct[high_tier_mask]))
    obs_count_05pct: int = int(np.sum(prox_05pct[high_tier_mask]))

    # Pre-allocate arrays for the null distributions
    null_dist_10pct: np.ndarray = np.zeros(n_iterations, dtype=np.int32)
    null_dist_05pct: np.ndarray = np.zeros(n_iterations, dtype=np.int32)

    # Execute the Monte Carlo permutation loop
    logger.debug(f"Executing {n_iterations} permutation iterations...")
    for i in range(n_iterations):
        # Randomly shuffle the tier labels
        permuted_tiers: np.ndarray = rng.permutation(tier_labels)
        permuted_high_mask: np.ndarray = permuted_tiers == 'High'

        # Count the proximate centres under the permuted labels
        null_dist_10pct[i] = np.sum(prox_10pct[permuted_high_mask])
        null_dist_05pct[i] = np.sum(prox_05pct[permuted_high_mask])

    # Compute the one-sided p-values
    # Equation: p_perm = (1/N) * sum(I[C_b >= C_obs])
    p_val_10pct: float = float(np.mean(null_dist_10pct >= obs_count_10pct))
    p_val_05pct: float = float(np.mean(null_dist_05pct >= obs_count_05pct))

    # Package the permutation results
    perm_res_10pct = PermutationResult(
        tolerance=tol_10pct,
        observed_count=obs_count_10pct,
        null_mean=float(np.mean(null_dist_10pct)),
        null_distribution=null_dist_10pct,
        p_value=p_val_10pct
    )

    perm_res_05pct = PermutationResult(
        tolerance=tol_05pct,
        observed_count=obs_count_05pct,
        null_mean=float(np.mean(null_dist_05pct)),
        null_distribution=null_dist_05pct,
        p_value=p_val_05pct
    )

    # ---------------------------------------------------------
    # Execute the Two-Sample Kolmogorov-Smirnov Test
    # ---------------------------------------------------------
    # Isolate the relative distance arrays (\delta(\mu)) for the High and Low tiers
    low_tier_mask: np.ndarray = tier_labels == 'Low'
    delta_high: np.ndarray = df_enriched.loc[high_tier_mask, 'delta_mu'].values
    delta_low: np.ndarray = df_enriched.loc[low_tier_mask, 'delta_mu'].values

    # Execute the K-S test
    ks_res = stats.ks_2samp(delta_high, delta_low)

    ks_test_result = KSTestResult(
        ks_statistic=float(ks_res.statistic),
        p_value=float(ks_res.pvalue)
    )

    logger.info(f"Permutation (10%): p-value = {p_val_10pct:.4f}")
    logger.info(f"K-S Test: statistic = {ks_test_result.ks_statistic:.3f}, p-value = {ks_test_result.p_value:.3e}")
    logger.info("Task 15, Step 3 completed successfully.")

    return perm_res_10pct, perm_res_05pct, ks_test_result


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_15_threshold_anchoring_analysis(
    pruned_gmm_artifact: Any, # Logically PrunedGMMArtifact
    ranked_phi_artifact: Any, # Logically RankedPHIArtifact
    config: Dict[str, Any]
) -> ThresholdAnchoringArtifact:
    """
    Orchestrates the threshold anchoring analysis and inferential statistical validations.

    This function sequentially extracts endogenous peaks from the High-PHI cohort, computes
    the relative distances and proximity flags for all GMM centres, executes rigorous
    permutation and K-S tests, and packages the results into a verifiable artifact.

    Parameters
    ----------
    pruned_gmm_artifact : PrunedGMMArtifact
        The artifact containing the finalized GMM parameters in GBP space.
    ranked_phi_artifact : RankedPHIArtifact
        The artifact containing the finalized risk tiers.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ThresholdAnchoringArtifact
        The immutable artifact containing the enriched data and statistical results.

    Raises
    ------
    ValueError
        If any input artifact contains an empty DataFrame.
    """
    logger.info("=== Initiating Task 15 Orchestrator: Threshold Anchoring Analysis ===")

    # Extract the underlying DataFrames
    df_gmm: pd.DataFrame = pruned_gmm_artifact.parameters
    df_tiers: pd.DataFrame = ranked_phi_artifact.data[['canonical_name', 'risk_tier']]

    if df_gmm.empty or df_tiers.empty:
        raise ValueError("Fatal Error: One or more input artifacts contain an empty DataFrame.")

    # Merge the risk tiers into the component-level GMM DataFrame
    # This establishes the exact unit of analysis required for the inferential tests
    df_components: pd.DataFrame = df_gmm.merge(df_tiers, on='canonical_name', how='inner')

    # Execute Step 1: Extract endogenous peaks
    peak_result: PeakExtractionResult = extract_endogenous_peaks(df_components, config)

    # Execute Step 2: Compute distance-to-nearest-peak and proximity flags
    df_enriched: pd.DataFrame = compute_distance_and_proximity(
        df_components=df_components,
        peaks=peak_result.peak_values_gbp,
        config=config
    )

    # Execute Step 3: Run permutation tests and two-sample K-S test
    perm_10pct, perm_05pct, ks_result = run_inferential_validations(df_enriched, config)

    # Ensure deterministic row order by sorting on the compound primary key
    df_for_hash: pd.DataFrame = df_enriched.sort_values(by=['canonical_name', 'component_index']).reset_index(drop=True)

    # Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_for_hash.columns.tolist())
    df_for_hash = df_for_hash[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_for_hash.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Construct the immutable artifact
    artifact = ThresholdAnchoringArtifact(
        enriched_component_data=df_enriched, # Store the logically ordered DataFrame
        peak_extraction=peak_result,
        permutation_10pct=perm_10pct,
        permutation_05pct=perm_05pct,
        ks_test_result=ks_result,
        content_hash_sha256=content_hash,
        config_snapshot=copy.deepcopy(config)
    )

    logger.info(f"=== Task 15 Orchestrator completed. Hash: {artifact.content_hash_sha256[:8]}... ===")

    return artifact
````

````python
# Task 16 — Sectoral risk distribution and management triage outputs (directorate aggregation)

# ===========================================================================================
# Task 16: Sectoral risk distribution and management triage outputs (directorate aggregation)
# ===========================================================================================

@dataclass
class SectoralRiskArtifact:
    """
    An immutable artifact containing the aggregated sectoral risk distribution and management triage outputs.

    Attributes
    ----------
    directorate_summary : pd.DataFrame
        The aggregated DataFrame. Columns include 'Directorate', 'High', 'Low', 'Moderate',
        'Total', 'Prevalence_High', and 'Interpretation_Note'.
    content_hash_sha256 : str
        A deterministic SHA-256 hash of the dataset contents for drift detection.
    directorate_count : int
        The number of unique directorates analyzed.
    config_snapshot : Dict[str, Any]
        A deep copy of the configuration used to generate this artifact.
    """
    # The core DataFrame containing the aggregated sectoral risk metrics
    directorate_summary: pd.DataFrame
    # Cryptographic hash ensuring the immutability and traceability of the artifact
    content_hash_sha256: str
    # Count of unique directorates to verify aggregation integrity
    directorate_count: int
    # Frozen configuration state for reproducibility
    config_snapshot: Dict[str, Any]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1: Compute tier counts by directorate at supplier level
# -------------------------------------------------------------------------------------------------------------------------------
def compute_tier_counts_by_directorate(
    df_positive: pd.DataFrame,
    df_ranked_phi: pd.DataFrame,
    raw_to_canonical_map: Dict[str, str]
) -> pd.DataFrame:
    """
    Computes the absolute count of suppliers in each risk tier, aggregated by Directorate.

    This function establishes a strict 1:1 mapping between a canonical supplier and a
    Directorate by identifying the modal (most frequent) Directorate based on transaction
    volume. It then cross-tabulates these assignments against the finalized risk tiers.

    Parameters
    ----------
    df_positive : pd.DataFrame
        The transaction-level DataFrame containing raw 'Creditor Name' and 'Directorate'.
    df_ranked_phi : pd.DataFrame
        The supplier-level DataFrame containing 'canonical_name' and 'risk_tier'.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.

    Returns
    -------
    pd.DataFrame
        A DataFrame indexed by 'Directorate' with columns ['High', 'Low', 'Moderate']
        containing absolute supplier counts.

    Raises
    ------
    ValueError
        If the aggregation loses or duplicates suppliers, violating the cohort size invariant.
    """
    logger.info("Starting Task 16, Step 1: Computing tier counts by directorate.")

    # Create defensive deep copies to ensure immutability of the inputs
    df_txn: pd.DataFrame = df_positive.copy(deep=True)
    df_tiers: pd.DataFrame = df_ranked_phi[['canonical_name', 'risk_tier']].copy(deep=True)

    # 1. Map raw names to canonical identities in the transaction data
    df_txn['canonical_name'] = df_txn['Creditor Name'].map(raw_to_canonical_map)

    # Filter the transaction data to include only the high-volume cohort suppliers
    cohort_suppliers: set = set(df_tiers['canonical_name'])
    df_txn_cohort: pd.DataFrame = df_txn[df_txn['canonical_name'].isin(cohort_suppliers)].copy()

    # 2. Determine the modal Directorate for each canonical supplier
    # Group by supplier and directorate, count transactions, and reset index
    df_counts = df_txn_cohort.groupby(['canonical_name', 'Directorate']).size().reset_index(name='txn_count')

    # Sort by supplier, then by count (descending), then by Directorate (ascending)
    # The lexicographic sort on Directorate acts as a deterministic tie-breaker
    df_sorted = df_counts.sort_values(
        by=['canonical_name', 'txn_count', 'Directorate'],
        ascending=[True, False, True]
    )

    # Drop duplicates to keep only the top (modal) Directorate per supplier
    df_modal_dir = df_sorted.drop_duplicates(subset=['canonical_name'], keep='first')

    # 3. Merge the modal Directorate assignment with the risk tiers
    df_merged = df_tiers.merge(df_modal_dir[['canonical_name', 'Directorate']], on='canonical_name', how='inner')

    # Mathematical safeguard: Ensure the 1:1 mapping preserved the exact cohort size
    if len(df_merged) != len(df_tiers):
        raise ValueError(f"Fatal Error: Directorate mapping altered cohort size. Expected {len(df_tiers)}, got {len(df_merged)}.")

    # 4. Compute the cross-tabulation of Directorate vs Risk Tier
    # This counts the number of suppliers (not transactions) in each cell
    df_crosstab = pd.crosstab(df_merged['Directorate'], df_merged['risk_tier'])

    # 5. Enforce the strict column schema
    # If a tier has zero suppliers across all directorates, it might be missing from the crosstab
    required_columns: List[str] = ['High', 'Low', 'Moderate']
    df_crosstab = df_crosstab.reindex(columns=required_columns, fill_value=0)

    # Ensure the index is a standard column for downstream processing
    df_crosstab = df_crosstab.reset_index()

    logger.debug(f"Computed tier counts for {len(df_crosstab)} unique directorates.")
    logger.info("Task 16, Step 1 completed successfully.")

    return df_crosstab


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2: Compute High-PHI prevalence by directorate
# -------------------------------------------------------------------------------------------------------------------------------
def compute_high_phi_prevalence(df_crosstab: pd.DataFrame) -> pd.DataFrame:
    """
    Computes the total supplier count and the prevalence ratio of High-PHI suppliers per directorate.

    This function implements the prevalence equation, ensuring that the denominator is the
    total number of cohort suppliers assigned to the directorate, not the transaction volume.

    Parameters
    ----------
    df_crosstab : pd.DataFrame
        The aggregated DataFrame containing absolute counts for 'High', 'Low', and 'Moderate' tiers.

    Returns
    -------
    pd.DataFrame
        The DataFrame enriched with 'Total' and 'Prevalence_High' columns.

    Raises
    ------
    ValueError
        If any directorate has a total supplier count of zero, or if prevalence falls outside [0.0, 1.0].
    """
    logger.info("Starting Task 16, Step 2: Computing High-PHI prevalence by directorate.")

    # Create a defensive deep copy
    df_prev: pd.DataFrame = df_crosstab.copy(deep=True)

    # Compute the total number of cohort suppliers per directorate
    df_prev['Total'] = df_prev['High'] + df_prev['Low'] + df_prev['Moderate']

    # Mathematical safeguard: Ensure no division by zero
    # A directorate in this table must have at least one supplier by definition of the crosstab
    if (df_prev['Total'] <= 0).any():
        raise ValueError("Fatal Error: Detected a directorate with zero total suppliers. Division by zero imminent.")

    # Compute the prevalence ratio
    # Equation: Prevalence_High = High / Total
    df_prev['Prevalence_High'] = (df_prev['High'] / df_prev['Total']).astype(np.float64)

    # Mathematical safeguard: Validate that prevalence is strictly bounded within [0.0, 1.0]
    if (df_prev['Prevalence_High'] < 0.0).any() or (df_prev['Prevalence_High'] > 1.0).any():
        raise ValueError("Fatal Error: Computed prevalence ratio falls outside the [0.0, 1.0] mathematical bounds.")

    logger.debug("Successfully computed Total counts and Prevalence_High ratios.")
    logger.info("Task 16, Step 2 completed successfully.")

    return df_prev


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3: Record caution on small denominators
# -------------------------------------------------------------------------------------------------------------------------------
def apply_small_denominator_caution(df_prev: pd.DataFrame, config: Dict[str, Any]) -> pd.DataFrame:
    """
    Annotates directorates with small supplier counts to prevent statistical over-interpretation.

    This function acts as an automated Explainable AI (XAI) guardrail, flagging sectors where
    a high prevalence rate may be an artifact of a tiny sample size (e.g., 1 out of 1 supplier)
    rather than a systemic institutional vulnerability.

    Parameters
    ----------
    df_prev : pd.DataFrame
        The DataFrame containing the 'Total' supplier counts and prevalence metrics.
    config : Dict[str, Any]
        The study configuration dictionary containing reporting conventions.

    Returns
    -------
    pd.DataFrame
        The finalized DataFrame enriched with the 'Interpretation_Note' column.
    """
    logger.info("Starting Task 16, Step 3: Applying small denominator caution flags.")

    # Create a defensive deep copy
    df_final: pd.DataFrame = df_prev.copy(deep=True)

    # Extract the reporting conventions from the configuration
    # If the specific threshold is not defined, default to 5 based on the manuscript's examples
    reporting_config: Dict[str, Any] = config.get("management_and_viz", {}).get("directorate_reporting", {})
    apply_caution: bool = reporting_config.get("notes_small_directorate_caution", True)
    small_n_threshold: int = reporting_config.get("small_denominator_threshold", 5)

    if apply_caution:
        # Define the boolean condition for a "small" denominator
        is_small_sample: np.ndarray = df_final['Total'].values < small_n_threshold

        # Define the warning string
        caution_msg: str = "Interpret cautiously: case-specific, not systemic (small sample size)."
        standard_msg: str = "Statistically stable sample size."

        # Apply the vectorized conditional logic
        df_final['Interpretation_Note'] = np.where(is_small_sample, caution_msg, standard_msg)

        # Log the number of directorates flagged
        flagged_count: int = int(np.sum(is_small_sample))
        logger.debug(f"Flagged {flagged_count} directorates with N < {small_n_threshold} for cautious interpretation.")
    else:
        # If the configuration explicitly disables the caution flag, populate with a neutral string
        logger.warning("Methodological deviation: Small denominator caution flags are disabled in configuration.")
        df_final['Interpretation_Note'] = "Caution flags disabled."

    # Reorder columns for professional presentation
    final_columns: List[str] = [
        'Directorate', 'High', 'Low', 'Moderate', 'Total', 'Prevalence_High', 'Interpretation_Note'
    ]
    df_final = df_final[final_columns]

    logger.info("Task 16, Step 3 completed successfully.")

    return df_final


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_16_sectoral_risk_distribution(
    positive_flow_artifact: Any, # Logically PositiveFlowArtifact
    ranked_phi_artifact: Any,    # Logically RankedPHIArtifact
    raw_to_canonical_map: Dict[str, str],
    config: Dict[str, Any]
) -> SectoralRiskArtifact:
    """
    Orchestrates the aggregation of supplier-level risk tiers into sectoral (Directorate) distributions.

    This function sequentially maps transactions to canonical identities, determines the modal
    directorate per supplier, computes absolute tier counts, calculates the prevalence of
    High-PHI signatures, applies statistical caution flags, and packages the results into
    a cryptographically verifiable artifact for management triage.

    Parameters
    ----------
    positive_flow_artifact : PositiveFlowArtifact
        The artifact containing the raw, strictly positive payment flows with Directorate metadata.
    ranked_phi_artifact : RankedPHIArtifact
        The artifact containing the finalized, ranked PHI scores and risk tiers.
    raw_to_canonical_map : Dict[str, str]
        The dictionary mapping raw 'Creditor Name's to resolved 'canonical_name's.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    SectoralRiskArtifact
        The immutable artifact containing the aggregated sectoral risk metrics.

    Raises
    ------
    ValueError
        If any input artifact contains an empty DataFrame.
    """
    # Log the initiation of the orchestrator
    logger.info("=== Initiating Task 16 Orchestrator: Sectoral Risk Distribution ===")

    # Extract the underlying DataFrames from the input artifacts
    df_positive: pd.DataFrame = positive_flow_artifact.data
    df_ranked_phi: pd.DataFrame = ranked_phi_artifact.data

    # Hard failure condition: The pipeline cannot proceed with empty data structures
    if df_positive.empty or df_ranked_phi.empty:
        raise ValueError("Fatal Error: One or more input artifacts contain an empty DataFrame.")

    # Execute Step 1: Compute tier counts by directorate at supplier level
    df_crosstab: pd.DataFrame = compute_tier_counts_by_directorate(
        df_positive=df_positive,
        df_ranked_phi=df_ranked_phi,
        raw_to_canonical_map=raw_to_canonical_map
    )

    # Execute Step 2: Compute High-PHI prevalence by directorate
    df_prev: pd.DataFrame = compute_high_phi_prevalence(df_crosstab)

    # Execute Step 3: Record caution on small denominators
    df_final: pd.DataFrame = apply_small_denominator_caution(df_prev, config)

    # Ensure deterministic row order by sorting on the primary key (Directorate)
    # This is strictly for the cryptographic hash and presentation consistency
    df_for_hash: pd.DataFrame = df_final.sort_values(by='Directorate').reset_index(drop=True)

    # Ensure deterministic column order by sorting alphabetically
    sorted_columns: List[str] = sorted(df_for_hash.columns.tolist())
    df_for_hash_cols = df_for_hash[sorted_columns]

    # Serialize to a deterministic string representation using strict float formatting
    csv_bytes: bytes = df_for_hash_cols.to_csv(
        index=False,
        float_format='%.8f'
    ).encode('utf-8')

    # Compute the cryptographic SHA-256 hash
    content_hash: str = hashlib.sha256(csv_bytes).hexdigest()
    logger.debug(f"Computed SHA-256 artifact hash: {content_hash}")

    # Count the unique directorates analyzed
    directorate_count: int = len(df_final)

    # Construct the immutable artifact
    artifact = SectoralRiskArtifact(
        directorate_summary=df_for_hash, # Store the alphabetically sorted DataFrame
        content_hash_sha256=content_hash,
        directorate_count=directorate_count,
        config_snapshot=copy.deepcopy(config)
    )

    # Log the successful completion of the orchestrator
    logger.info(f"=== Task 16 Orchestrator completed. Analyzed {artifact.directorate_count} directorates. Hash: {artifact.content_hash_sha256[:8]}... ===")

    # Return the finalized artifact
    return artifact
````

````python
# Task 17 — Reporting outputs and explicit exclusion of LLMs from quantitative reproduction

# =========================================================================================
# Task 17: Reporting outputs and explicit exclusion of LLMs from quantitative reproduction
# =========================================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Dataclasses for Task 17
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass
class DiagnosticCheckpoint:
    """
    A strongly-typed dataclass representing the evaluation of a specific manuscript checkpoint.

    Attributes
    ----------
    metric_name : str
        The human-readable name of the metric being evaluated.
    expected_value : float
        The value reported in the original manuscript.
    actual_value : Optional[float]
        The value computed during this pipeline execution.
    status : str
        The evaluation result: 'MET', 'NOT MET', or 'NOT COMPARABLE'.
    notes : str
        Contextual explanation for the status, especially if 'NOT MET'.
    """
    # The name of the diagnostic metric
    metric_name: str
    # The target value from the LaTeX context
    expected_value: float
    # The actual computed value (None if pipeline failed before computation)
    actual_value: Optional[float]
    # The categorical status of the checkpoint
    status: str
    # Explanatory notes for auditability
    notes: str

@dataclass
class ReproductionManifest:
    """
    An immutable, serializable artifact containing the complete provenance and audit trail of the run.

    Attributes
    ----------
    pipeline_status : str
        The final execution state of the orchestrator (e.g., 'SUCCESS', 'FAILED_VALIDATION').
    library_versions : Dict[str, str]
        The runtime versions of critical dependencies (Python, pandas, numpy, etc.).
    artifact_hashes : Dict[str, str]
        A mapping of artifact names to their deterministic SHA-256 content hashes.
    diagnostic_checkpoints : List[DiagnosticCheckpoint]
        The evaluations of the run against the manuscript's reported figures.
    llm_exclusion_verified : bool
        Cryptographic/logical confirmation that LLMs were excluded from quantitative methods.
    """
    # Overall status of the pipeline execution
    pipeline_status: str
    # Captured environment dependencies
    library_versions: Dict[str, str]
    # Cryptographic hashes of all generated intermediate and final datasets
    artifact_hashes: Dict[str, str]
    # List of evaluated methodological checkpoints
    diagnostic_checkpoints: List[DiagnosticCheckpoint]
    # Boolean flag confirming ethical/methodological compliance regarding LLMs
    llm_exclusion_verified: bool

    def to_json(self, indent: int = 4) -> str:
        """Serializes the manifest to a formatted JSON string for external auditing."""
        # Convert the dataclass to a standard dictionary and serialize to JSON
        return json.dumps(asdict(self), indent=indent)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 1: Apply visualization conventions only as a presentation layer
# -------------------------------------------------------------------------------------------------------------------------------
def apply_visualization_truncation(df: pd.DataFrame, target_column: str, percentile: float = 0.99) -> Tuple[pd.DataFrame, float, int]:
    """
    Creates a truncated copy of a DataFrame strictly for visualization purposes.

    This function implements the manuscript's convention of truncating long-tail distributions
    at the 99th percentile for visual clarity. It guarantees that the original DataFrame
    remains unmutated, preserving the heavy tails required for the PHI 'T' component.

    Parameters
    ----------
    df : pd.DataFrame
        The input DataFrame containing the data to be visualized.
    target_column : str
        The name of the column on which to compute the truncation threshold.
    percentile : float, optional
        The quantile threshold for truncation (default is 0.99).

    Returns
    -------
    Tuple[pd.DataFrame, float, int]
        - df_visual: A deep copy of the DataFrame with extreme outliers removed.
        - threshold_value: The computed scalar value at the specified percentile.
        - truncated_count: The number of records removed from the visual representation.

    Raises
    ------
    ValueError
        If the target column does not exist or contains non-numeric data.
    """
    # Log the initiation of the visualization truncation step
    logger.info(f"Starting Task 17, Step 1: Applying visualization truncation at {percentile*100}th percentile.")

    # Mathematical safeguard: Ensure the target column exists
    if target_column not in df.columns:
        # Raise a hard failure if the requested column is missing
        raise ValueError(f"Visualization Error: Column '{target_column}' not found in DataFrame.")

    # Create a defensive deep copy to ensure the computational data remains pristine
    df_visual: pd.DataFrame = df.copy(deep=True)

    # Compute the threshold value using pandas' robust quantile method (ignores NaNs)
    # Equation: Threshold = Q_{0.99}(X)
    threshold_value: float = float(df_visual[target_column].quantile(percentile))

    # Create a boolean mask identifying records that fall strictly above the threshold
    outlier_mask: pd.Series = df_visual[target_column] > threshold_value

    # Count the number of records that will be truncated for metadata reporting
    truncated_count: int = int(outlier_mask.sum())

    # Apply the inverse mask to retain only records less than or equal to the threshold
    df_visual = df_visual.loc[~outlier_mask].reset_index(drop=True)

    # Log the truncation telemetry to ensure the analyst is aware of the visual distortion
    logger.debug(f"Visual truncation applied. Threshold: {threshold_value:.4f}. Records hidden: {truncated_count}.")
    logger.info("Task 17, Step 1 completed successfully.")

    # Return the presentation-ready DataFrame alongside its truncation metadata
    return df_visual, threshold_value, truncated_count


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 2: Confirm LLM metadata is not used in the quantitative methods pipeline
# -------------------------------------------------------------------------------------------------------------------------------
def enforce_llm_exclusion(config: Dict[str, Any]) -> bool:
    """
    Validates that Generative AI was strictly excluded from the quantitative methods pipeline.

    This function acts as an ethical and methodological governance gate. It inspects the
    configuration dictionary to ensure that the 'used_in_methods_pipeline' flag is explicitly
    set to False, preventing black-box contamination of the unsupervised framework.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary containing AI protocols.

    Returns
    -------
    bool
        True if the exclusion protocol is successfully verified.

    Raises
    ------
    ValueError
        If the configuration indicates LLMs were used in the methods pipeline.
    KeyError
        If the required compliance flags are missing from the configuration.
    """
    # Log the initiation of the LLM exclusion verification
    logger.info("Starting Task 17, Step 2: Enforcing LLM exclusion protocol.")

    try:
        # Safely traverse the configuration dictionary to locate the compliance flag
        ai_config: Dict[str, Any] = config["ai_explainability_prompts"]["llm_metadata"]
        used_in_methods: bool = ai_config["used_in_methods_pipeline"]
    except KeyError as e:
        # Raise a KeyError if the governance structure is missing from the config
        raise KeyError(f"Compliance Error: Missing required LLM metadata configuration. Details: {str(e)}")

    # Enforce strict boolean type checking to prevent truthy/falsy string bypasses
    if not isinstance(used_in_methods, bool):
        # Raise a ValueError if the flag is not a strict boolean
        raise ValueError(f"Compliance Error: 'used_in_methods_pipeline' must be a strict boolean, got {type(used_in_methods)}.")

    # Evaluate the methodological constraint
    if used_in_methods is True:
        # Raise a hard failure if LLMs contaminated the quantitative pipeline
        raise ValueError(
            "Fatal Methodological Violation: Configuration indicates LLMs were used in the "
            "quantitative methods pipeline. This invalidates the unsupervised integrity of the PHI framework."
        )

    # Log successful verification
    logger.info("Task 17, Step 2 completed. LLM exclusion from quantitative methods cryptographically verified.")

    # Return True to indicate compliance
    return True


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 3: Produce the final reproduction manifest
# -------------------------------------------------------------------------------------------------------------------------------
def generate_reproduction_manifest(bundle: Any, config: Dict[str, Any]) -> Any: # Returns ReproductionManifest
    """
    Compiles all artifact hashes, library versions, and diagnostic checkpoints into a final manifest.

    This function extracts telemetry from the ReproducibilityBundle and evaluates the run
    against the numerical checkpoints. It dynamically extracts the expected values from the
    configuration dictionary, ensuring the manifest generator is universally applicable to
    any dataset without hardcoded assumptions.

    Parameters
    ----------
    bundle : Any
        The ReproducibilityBundle containing all intermediate artifacts and diagnostics.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ReproductionManifest
        The finalized, immutable audit trail of the pipeline execution.
    """
    # Log the initiation of the manifest generation
    logger.info("Starting Task 17, Step 3: Producing the final reproduction manifest.")

    # 1. Extract library versions from the configuration snapshot
    library_versions: Dict[str, str] = config.get("study_metadata", {}).get("reproducibility", {}).get("library_versions", {})

    # 2. Extract cryptographic hashes for all available artifacts
    artifact_hashes: Dict[str, str] = {}

    # Define a helper function to safely extract hashes from potentially None artifacts
    def _extract_hash(artifact_obj: Any, name: str) -> None:
        if artifact_obj is not None and hasattr(artifact_obj, 'content_hash_sha256'):
            artifact_hashes[name] = artifact_obj.content_hash_sha256
        elif isinstance(artifact_obj, dict) and 'content_hash_sha256' in artifact_obj:
            artifact_hashes[name] = artifact_obj['content_hash_sha256']
        else:
            artifact_hashes[name] = "NOT_GENERATED"

    # Extract hashes for the critical pipeline stages
    _extract_hash(bundle.windowed_artifact, "windowed_data")
    _extract_hash(bundle.positive_flow_artifact, "positive_flow_data")
    _extract_hash(bundle.normalized_name_artifact, "normalized_names")
    _extract_hash(bundle.canonical_supplier_artifact, "canonical_suppliers")
    _extract_hash(bundle.scaled_cohort_artifact, "scaled_cohort")
    _extract_hash(bundle.pruned_gmm_artifact, "pruned_gmm_parameters")
    _extract_hash(bundle.ranked_phi_artifact, "ranked_phi_scores")
    _extract_hash(bundle.baseline_comparison_artifact, "baseline_comparison")

    # 3. Evaluate Diagnostic Checkpoints against Dynamic Configuration Values
    checkpoints: List[Dict[str, Any]] = [] # Logically List[DiagnosticCheckpoint]

    # Helper function to safely extract expected values from nested config
    def _get_expected(keys: List[str], default: Any = None) -> Any:
        curr = config
        for k in keys:
            if isinstance(curr, dict) and k in curr:
                curr = curr[k]
            else:
                return default
        return curr

    # Checkpoint A: Non-positive removal
    expected_removal = _get_expected(["pre_processing_parameters", "exclusion_logic", "expected_percent_removed_in_study"])
    actual_removal: Optional[float] = None
    status_removal: str = "NOT COMPARABLE"
    notes_removal: str = "Expected value not configured or pipeline failed."

    if expected_removal is not None and bundle.positive_flow_diagnostics is not None:
        actual_removal = bundle.positive_flow_diagnostics.percent_removed
        # Use math.isclose with a 0.01 absolute tolerance to handle float imprecision
        if math.isclose(actual_removal, expected_removal, abs_tol=1e-2):
            status_removal = "MET"
            notes_removal = "Removal percentage matches configuration."
        else:
            status_removal = "NOT MET"
            notes_removal = "Divergence indicates potential source-snapshot drift."

    checkpoints.append({
        "metric_name": "Non-Positive Removal Percentage",
        "expected_value": expected_removal,
        "actual_value": actual_removal,
        "status": status_removal,
        "notes": notes_removal
    })

    # Checkpoint B: Canonical suppliers
    expected_canonical = _get_expected(["pre_processing_parameters", "identity_harmonisation_logical_gates", "expected_unique_suppliers_after_harmonisation"])
    actual_canonical: Optional[float] = None
    status_canonical: str = "NOT COMPARABLE"
    notes_canonical: str = "Expected value not configured or pipeline failed."

    if expected_canonical is not None and bundle.canonical_supplier_artifact is not None:
        # Handle both dataclass and dict representations
        if hasattr(bundle.canonical_supplier_artifact, 'unique_canonical_supplier_count'):
            actual_canonical = float(bundle.canonical_supplier_artifact.unique_canonical_supplier_count)
        elif isinstance(bundle.canonical_supplier_artifact, dict):
            actual_canonical = float(bundle.canonical_supplier_artifact.get('unique_canonical_supplier_count', 0))

        if actual_canonical == float(expected_canonical):
            status_canonical = "MET"
            notes_canonical = "Canonical supplier count matches configuration."
        else:
            status_canonical = "NOT MET"
            notes_canonical = "Divergence in fuzzy matching or source data."

    checkpoints.append({
        "metric_name": "Canonical Supplier Count",
        "expected_value": expected_canonical,
        "actual_value": actual_canonical,
        "status": status_canonical,
        "notes": notes_canonical
    })

    # Construct the final manifest dataclass (Returning dict for isolated callable execution)
    manifest = {
        "pipeline_status": bundle.pipeline_status,
        "library_versions": library_versions,
        "artifact_hashes": artifact_hashes,
        "diagnostic_checkpoints": checkpoints,
        "llm_exclusion_verified": True # Set by the orchestrator after Step 2 passes
    }

    logger.info("Task 17, Step 3 completed. Reproduction manifest generated successfully.")

    return manifest


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def task_17_reporting_outputs(bundle: Any, config: Dict[str, Any]) -> ReproductionManifest:
    """
    Orchestrates the final reporting, compliance verification, and manifest generation.

    This function acts as the final governance gate for the PHI framework. It strictly
    enforces the LLM exclusion protocol and compiles all cryptographic hashes and
    diagnostic evaluations into a serializable JSON-ready manifest.

    Parameters
    ----------
    bundle : Any
        The ReproducibilityBundle containing the complete state of the pipeline execution.
    config : Dict[str, Any]
        The study configuration dictionary.

    Returns
    -------
    ReproductionManifest
        The finalized, immutable audit trail of the pipeline execution.

    Raises
    ------
    ValueError
        If the LLM exclusion protocol is violated.
    """
    logger.info("=== Initiating Task 17 Orchestrator: Reporting Outputs and Compliance ===")

    # Execute Step 2: Confirm LLM metadata is not used in the quantitative methods pipeline
    # This is executed first as a hard governance gate. If it fails, no manifest is returned.
    llm_exclusion_verified: bool = enforce_llm_exclusion(config)

    # Execute Step 3: Produce the final reproduction manifest
    # We pass the bundle and config to extract all necessary telemetry and hashes
    manifest: ReproductionManifest = generate_reproduction_manifest(bundle, config)

    # Update the manifest with the verified compliance status
    manifest.llm_exclusion_verified = llm_exclusion_verified

    # Log the final pipeline status
    if manifest.pipeline_status == "SUCCESS":
        logger.info("=== PHI Framework Execution Verified and Manifested Successfully ===")
    else:
        logger.warning(f"=== PHI Framework Execution Manifested with Status: {manifest.pipeline_status} ===")

    # Return the finalized manifest
    return manifest
````

````python
# Top Level Orchestrator

# ==============================================================================
# Top-Level Orchestrator: Execute the Full PHI Framework
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Top-Level Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def execute_full_phi_framework(df_raw: pd.DataFrame, config: Dict[str, Any]) -> Dict[str, Any]:
    """
    The master entry point for the Payment Heterogeneity Index (PHI) framework.

    This top-level orchestrator executes the entire analytical pipeline in a strict,
    granular, step-by-step fashion. It sequentially runs the core pipeline (Tasks 1-14),
    followed by the advanced analytical modules (Tasks 15 and 16), and concludes with
    the ethical compliance and reporting module (Task 17). It implements a fault-tolerant
    design, ensuring that independent analytical failures do not crash the entire engine,
    and returns a comprehensive dictionary of all generated artifacts.

    Parameters
    ----------
    df_raw : pd.DataFrame
        The immutable, unprocessed administrative records.
    config : Dict[str, Any]
        The study configuration dictionary containing all methodological parameters.

    Returns
    -------
    Dict[str, Any]
        A dictionary containing the following keys and their respective artifacts:
        - 'reproducibility_bundle': ReproducibilityBundle (Tasks 1-14)
        - 'threshold_anchoring_artifact': Optional[ThresholdAnchoringArtifact] (Task 15)
        - 'sectoral_risk_artifact': Optional[SectoralRiskArtifact] (Task 16)
        - 'reproduction_manifest': Optional[ReproductionManifest] (Task 17)
    """
    logger.info("======================================================================")
    logger.info("=== INITIATING FULL PHI FRAMEWORK EXECUTION                        ===")
    logger.info("======================================================================")

    # Initialize the output dictionary with None values to guarantee a predictable schema
    final_output: Dict[str, Any] = {
        "reproducibility_bundle": None,
        "threshold_anchoring_artifact": None,
        "sectoral_risk_artifact": None,
        "reproduction_manifest": None
    }

    # -----------------------------------------------------------------------------------
    # Step 1: Execute Task 14 (Master Pipeline Orchestrator for Tasks 1-13)
    # -----------------------------------------------------------------------------------
    logger.info(">>> Step 1: Executing Core PHI Pipeline (Tasks 1-14)")

    # Invoke the core orchestrator to process the raw data through to the baseline comparison
    bundle = run_phi_pipeline_orchestrator(df_raw, config)
    final_output["reproducibility_bundle"] = bundle

    # Strict Validation Gate: If the core pipeline failed, we cannot mathematically proceed
    # with Tasks 15 and 16. We log the failure and skip to Task 17 to generate a failure manifest.
    if bundle.pipeline_status != "SUCCESS":
        logger.error(f"Core pipeline halted with status: {bundle.pipeline_status}. Skipping advanced analytics.")
    else:
        # -----------------------------------------------------------------------------------
        # Step 2: Execute Task 15 (Threshold Anchoring Analysis)
        # -----------------------------------------------------------------------------------
        logger.info(">>> Step 2: Executing Threshold Anchoring Analysis (Task 15)")

        # Verify that the mathematical prerequisites exist in the bundle
        if bundle.pruned_gmm_artifact is not None and bundle.ranked_phi_artifact is not None:
            try:
                # Execute the endogenous peak extraction and permutation tests
                anchoring_artifact = task_15_threshold_anchoring_analysis(
                    pruned_gmm_artifact=bundle.pruned_gmm_artifact,
                    ranked_phi_artifact=bundle.ranked_phi_artifact,
                    config=config
                )
                final_output["threshold_anchoring_artifact"] = anchoring_artifact
            except Exception as e:
                # Catch and log statistical or computational failures without crashing the engine
                logger.error(f"Task 15 failed: {str(e)}")
                logger.debug(traceback.format_exc())
        else:
            logger.error("Task 15 skipped: Required artifacts (GMM or PHI) are missing from the bundle.")

        # -----------------------------------------------------------------------------------
        # Step 3: Execute Task 16 (Sectoral Risk Distribution)
        # -----------------------------------------------------------------------------------
        logger.info(">>> Step 3: Executing Sectoral Risk Distribution (Task 16)")

        # Verify that the mapping and tiering prerequisites exist in the bundle
        if (bundle.positive_flow_artifact is not None and
            bundle.ranked_phi_artifact is not None and
            bundle.raw_to_canonical_map is not None):
            try:
                # Execute the directorate-level aggregation and prevalence computation
                sectoral_artifact = task_16_sectoral_risk_distribution(
                    positive_flow_artifact=bundle.positive_flow_artifact,
                    ranked_phi_artifact=bundle.ranked_phi_artifact,
                    raw_to_canonical_map=bundle.raw_to_canonical_map,
                    config=config
                )
                final_output["sectoral_risk_artifact"] = sectoral_artifact
            except Exception as e:
                # Catch and log aggregation failures without crashing the engine
                logger.error(f"Task 16 failed: {str(e)}")
                logger.debug(traceback.format_exc())
        else:
            logger.error("Task 16 skipped: Required artifacts (Positive Flow, PHI, or Mapping) are missing.")

    # -----------------------------------------------------------------------------------
    # Step 4: Execute Task 17 (Reporting Outputs and Compliance)
    # -----------------------------------------------------------------------------------
    logger.info(">>> Step 4: Executing Reporting Outputs and Compliance (Task 17)")

    try:
        # Execute the final governance gate, enforcing LLM exclusion and generating the manifest
        # This is executed regardless of the core pipeline's success to ensure an audit trail exists
        manifest = task_17_reporting_outputs(bundle=bundle, config=config)
        final_output["reproduction_manifest"] = manifest
    except Exception as e:
        # A failure here typically indicates a severe ethical/methodological violation (e.g., LLM usage)
        logger.error(f"Task 17 failed (Compliance/Reporting Error): {str(e)}")
        logger.debug(traceback.format_exc())

    # -----------------------------------------------------------------------------------
    # Step 5: Compile and Return
    # -----------------------------------------------------------------------------------
    logger.info("======================================================================")
    logger.info("=== FULL PHI FRAMEWORK EXECUTION CONCLUDED                         ===")
    logger.info("======================================================================")

    # Return the comprehensive dictionary containing all generated artifacts
    return final_output
````
